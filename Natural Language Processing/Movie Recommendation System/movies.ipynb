```python
import ast
```

```python
def genre_extract(col):
    L = []
    for i in ast.literal_eval(col):
        L.append(i['name'])
    return L
```

```python
data['genres'] = data['genres'].apply(genre_extract)
```

```python
def keyword_extract(col):
    L=[]
    for i in ast.literal_eval(col):
        L.append(i['name'])
    return L 
```

```python
data['keywords'] = data['keywords'].apply(keyword_extract)
## here we are applying the keyword_extract func to the 'keyword' column and storing it in the 'keyword' column
```

```python
def cast_extract(col):
    L = []
    j = 1
    for i in ast.literal_eval(col):
        if j <= 3:
            L.append(i['name'])
            j+=1
    return L
```

```python
data['cast'] = data['cast'].apply(cast_extract)
```

```python
data['crew'][0]
```

```python
def crew_extract(col):
    L = []
    for i in ast.literal_eval(col):
        if i['job'] == 'Director':
            L.append(i['name'])
            break
    return L    
```

```python
data['crew'] = data['crew'].apply(crew_extract)
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
## we should convert the textual format of 'tags' --> into the List format then only the stemming can be applied
```

```python
data['tags'][0] ## str format ---> list
```

```python
def stemming(text):
    L=[]
    for i in text.split():
        L.append(ps.stem(i))  ## stemming and appending it into the list
    return " ".join(L)
```

#### For Vectorization we need a clean raw data ---> but here if we directly Print the List of words it will be difficult for the vectorization so, we will replace the , and joins the word the words with spaces " " than " , "

```python
data['tags'] = data['tags'].apply(stemming)
```

```python
data.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
vectors = cv.fit_transform(data['tags'])
```

```python
vectors ## we got sparse matrix ---> convert to numpy array
```

```python
vectors = vectors.toarray()
```

```python
vectors
```

```python
## what 5000 features choosen ???
cv.get_feature_names_out()[:200]
```

#### now we should come up with the similarity between the movies

#### movie by movie ---> pair wise checking cosine similarities

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarities = cosine_similarity(vectors)
```

```python
similarities[0]
```

```python
## fetching the movie index
data[data['title'] == 'avatar'].index[0]
```

```python
## function output 5 recommended movies
def recommend(movie):

    movie_index = data[data['title'] == movie].index[0]

    recommendations = similarities[movie_index]

    movie_list = sorted(enumerate(recommendations),reverse=True,key = lambda x:x[1])[1:6]
    ## these are the index of the 5 recommended movies

    for i in movie_list:
        print(data.iloc[i[0]].title) ## it will print the index location of the recommended movies
    
```
