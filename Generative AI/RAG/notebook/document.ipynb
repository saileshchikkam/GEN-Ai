```python
## creating the simple txt file

import os
os.makedirs("../data/text_files",exist_ok = True)
```

```python
sample_texts = {
    "../data/text_files/pharmacovigilance_intro.txt": """Pharmacovigilance Introduction

Pharmacovigilance is the science and activities concerned with the detection, assessment, understanding, and prevention of adverse effects or any other drug-related problems.

The primary goal of pharmacovigilance is to improve patient safety by continuously monitoring the safety and effectiveness of medicines after they are introduced into the market.

Key Areas:

- Adverse Drug Reaction (ADR) detection
- Drug safety monitoring
- Causality assessment
- Signal detection
- Risk assessment and management
- Patient safety

Pharmacovigilance data can be collected from healthcare professionals, patients, clinical trials, electronic health records, and spontaneous adverse event reports.""",

    "../data/text_files/adr_report.txt": """Adverse Drug Reaction Report

Patient: 45-year-old male
Suspected Drug: Metformin 500 mg
Indication: Type 2 Diabetes Mellitus
Treatment Duration: 6 months

The patient reported persistent nausea, abdominal discomfort, and diarrhea after starting metformin therapy. The symptoms gradually increased in severity and were associated with reduced appetite.

Clinical Findings:

- Persistent nausea
- Abdominal discomfort
- Diarrhea
- Reduced appetite
- Occasional dizziness

The physician suspected that the gastrointestinal symptoms could be associated with metformin therapy. The dose was reduced and the patient was advised to take the medication with meals.

Outcome:

The patient's symptoms improved significantly after dose reduction. No hospitalization was required.

Seriousness: Non-serious
Causality Assessment: Possible association with metformin.""",

    "../data/text_files/drug_safety.txt": """Drug Safety Information - Metformin

Metformin is a commonly prescribed medication for the management of Type 2 Diabetes Mellitus. Gastrointestinal adverse reactions are among the most frequently reported side effects.

Common adverse reactions include:

- Nausea
- Diarrhea
- Abdominal discomfort
- Bloating
- Reduced appetite

Gastrointestinal symptoms are often more noticeable when treatment is initiated or when the dosage is increased.

Taking metformin with meals and gradually increasing the dose may help reduce gastrointestinal intolerance.

Healthcare professionals should evaluate persistent or severe adverse reactions and determine whether dose adjustment or alternative treatment is appropriate.

The occurrence of an adverse event after administration of a medicine does not necessarily establish a causal relationship between the medicine and the event.""",

    "../data/text_files/adr_classification.txt": """Adverse Drug Reaction Classification

Adverse drug reactions can be classified according to their characteristics, severity, seriousness, and relationship with the suspected medicine.

Common categories include:

- Type A reactions: Dose-dependent and predictable reactions.
- Type B reactions: Unpredictable reactions that are not usually dose-dependent.
- Type C reactions: Reactions associated with long-term drug exposure.
- Type D reactions: Delayed adverse reactions.
- Type E reactions: Reactions that occur after withdrawal of a medicine.
- Type F reactions: Unexpected therapeutic failure.

ADR classification helps pharmacovigilance systems organize safety information and support further investigation.""",

    "../data/text_files/signal_detection.txt": """Pharmacovigilance Signal Detection

Signal detection is an important activity in pharmacovigilance used to identify previously unknown or insufficiently documented relationships between medicines and adverse events.

A potential safety signal may emerge when an adverse event is reported more frequently than expected for a particular medicine.

Signal detection can involve:

- Statistical analysis
- Disproportionality analysis
- Case report review
- Literature monitoring
- Clinical evaluation
- Expert assessment

Identified signals require further investigation before conclusions about drug safety can be established."""
}

for filepath, content in sample_texts.items():
    with open(filepath,'w', encoding="utf-8") as f:
        f.write(content)

print(" ✅ Sample text files created")
```

```python
## Text Loader

from langchain_community.document_loaders import TextLoader

loader = TextLoader("../data/text_files/adr_classification.txt",
                    encoding="utf-8")
document = loader.load()
print(document) # now it is documnet structure
```

```python
## Directory Loader

from langchain_community.document_loaders import DirectoryLoader

## load all the text files from the directory
dir_loader = DirectoryLoader(
    "../data/text_files",
    glob = "**/*.txt", ## Pattern to match files
    loader_cls = TextLoader, ## loader class to use
    loader_kwargs={"encoding" : "utf-8"},
    show_progress=False
)
documents = dir_loader.load()
documents
```

```python
## Pdf Loader

from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader

## load all the pdf files from the directory
dir_loader = DirectoryLoader(
    "../data/pdf",
    glob = "**/*.pdf", ## Pattern to match files
    loader_cls = PyMuPDFLoader, ## loader class to use
    show_progress=False
)
pdf_documents = dir_loader.load()
pdf_documents
```
