# Object Oriented Programming

### In this we are trying to match up the real world scenarios by using the programming

### By a class we mean to provide a blue print that must remain the same for each and every single object for us

### These Objects can have different functionalities from each other and different characteristics form each other as well but they follow the similar blueprint

### By a blue print or a class we mean to say that similar means of providingthe functionalities and the characteristic to our objects

### class ClassName :
### def _init__(self,name,weapon,nationality):
###     self.name = name
###     self.power = power
###     self.weapon = weapon
###     self.nationality = nationality
### def introduction(self):
###     print(f"Hello my name is {self.name}")



```python
class Avengers:
    def __init__(self,name,power,weapon,nationality = "American"):
        self.name = name
        self.power = power
        self.weapon = weapon
        self.nationality = nationality
    def introduction(self): #instance method
        print(f"Hello my name is {self.name}")
    def carrying_weapon(self):
        print(f"I am carrying the weapon {self.weapon}")
```

```python
#object for the class Avengers
obj1 = Avengers("Iron man","Iron power","Repulsive beams")
print(obj1.introduction())
print(obj1.carrying_weapon())
```

```python
obj1.name
```

```python
obj1.power
```

```python
obj2 = Avengers("Captain America","Super Strength","shield")
print(obj2.introduction(),end = " ")
obj2.power
```

## In Object Oriented Programming, we are trying to structure the code in such a way that we can match up with the real world scenarios

```python
class Cartoon:
    def __init__(self,name,show,catchphrase): #This is constructor
        # It is the Parameterised constructor
        self.name=name
        self.show=show
        self.catchphrase=catchphrase

    def introduction(self):
        print(f"Hello my name is {self.name}")
    def speak(self):
        print(f"{self.name} says {self.catchphrase}")
```

```python
cartoon = Cartoon("chotabheem","DC","Chota bheem chota bheem")
print(cartoon.introduction())
print(cartoon.speak())
```

## Types of Constructors

### 1. Default Constructor
It doesn't have parameters inside it and it is only having self keyword inside it. It assigns a default values to all os the objects

### 2. Parameterized Constructor
It consists of parameters inside it so that every object can have a specific attributes associated to it.

It  allows us to pass arguments by creating an object

```python
class Dog:
    def __init__(self):
        self.name = "Shitzhu"
    def displaydetails(self):
        print(f"this dog is a {self.name}")
            
```

```python
d1 = Dog()
d1.name
d1.displaydetails()
```

```python
# Multiple objects can't get diff values we get only default values
d2= Dog()
d2.name
```

```python
class Dog:
    def __init__(self,bread,color):
        self.bread = bread
        self.color = color
    def display(self):
        print(f"The dog bread is {self.bread}, the colour is {self.color}")
           
```

## Types of Methods

### 1. Instance Method

It operate on object attributes

It is going to use self keyword to access and modify the object attributes

These methods are called using an objects

### 2. Class Method

It is the method of the whole class or the whole blueprint which will remain the same for all of the objects then such kind of method is beign reffered as class method

In class Method , These methods work with the class level attributes. 

They are going to use the cls instead of self keyword

They are called using the class name as well

For specifying the current method is a class method we have to use the abbreviation as @classmethod

### 3. Static Method

Static Methods do not use self or cls

They work like normal functions but belong to a class

They are going to be designed by abbreviation @staticmethod

## 1. Instance Method Example

```python
class Avengers:
    def __init__(self,name,power):
        self.name = name
        self.power = power
    def show_details(self):
        print(f"My name is {self.name} and i am having power {self.power}")    
```

```python
iron_man = Avengers("iron man", "Fire")
iron_man.name
iron_man.show_details()
```

## 2. Class Method Example

```python
class Avengerss:

    origin = "Avenger" # class attribute

    def __init__(self,name,power):
        self.name = name
        self.power = power

    @classmethod
    # to access the class attribute origin we use cls keyword
    def origin_info(cls):
        print(f"My origin is {cls.origin}")

    def show_details(self):
        print(f"My name is {self.name} and i am having power {self.power}, {self.origin}")
```

```python
c1 = Avengerss("Thor","God of Thunder")
c1.origin_info()
```

```python
c1.show_details()
```

## 3. Static Method Example

```python
class Avengersss:

    @staticmethod
    def mission():
        print("All Avnegers must protect the earth")

    origin = "Avenger" # class attribute

    def __init__(self,name,power):
        self.name = name
        self.power = power

    @classmethod
        # to access the class attribute origin we use cls keyword
    def origin_info(cls):
        print(f"My origin is {cls.origin}")

    def show_details(self):
        print(f"My name is {self.name} and i am having power {self.power}")
    
```
