# Map Function

## Map()

### Map function applies a given function operation to All items in an iterable or a List and returns a Map Object(An Iterator)

### -->> Map function is particularly useful for transforming the data or applying an operation on All items of the iterable

### -->> Map() is going to help us out with unnecessary lines of code and we can apply the operation on all values or all items in an iterable and this sign of a good programmer

### Syntax for Map()

#### list(map(function,iterable))

```python
## Squaring all the items in the list

squared_list = []
def square(list_numbers):
    for items in list_numbers:
        squared_item = items ** 2
        squared_list.append(squared_item)
    print(squared_list)
square([1,2,3,4,5])
```

### Solving the above problem in simple way with a Map() , lambda()

```python
print(list(map(lambda x:x**2,[1,2,3,4,5])))
```

### Map function with basic function init

```python
def squared_num(num):
    return num ** 2

print(list(map(squared_num,[1,2,3,4,5])))
```

### Whenever you are going to pass the general function inside your map  function as the parameter then you dont havae to use parenthesis

## While maping the value of one iterable with another iterable we can actuslly apply the operation in between the values of the two iterables in a synchronus way

### [1,2,3,4] <----> [5,6,7,8] <


### -->> Map function with multiple Iteration

```python
numbers_1 = [1,2,3,4]
numbers_2 = [5,6,7,8]
print(list(map(lambda x,y: x+y, numbers_1,numbers_2)))
```

```python
print(list(map(lambda x,y : x*y, numbers_1, numbers_2)))
```

```python
### Map()  to convert all string vaalues into the integer value
ans = list(map(int,['1','2']))
print(ans)
print(type(ans[0])) ## int
```

```python
## Map() to convert the small letters to upper case
employee_details = ["sailesh","jayanth"]
print(list(map(str.upper,employee_details)))
```

```python
## example iterating over a dictionary
## Retrieve the values inside a dictionary using a Map()
employee = [{
    'Name' : 'Sailesh',
    'Age' : 21,
    'Qualification' : 'btech'
},
{
    'Name' : 'kiran',
    'Age' : 22,
    'Qualification' : 'b.com'
},
{
    'Name' : 'Roshan',
    'Age' : 27,
    'Qualification' : 'Mtech'
}]
```

```python
def age_retriever(list_item):
    return list_item['Age']
```

```python
list(map(age_retriever, employee))
```
