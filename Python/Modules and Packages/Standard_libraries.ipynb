## 3. csv library ---> Most  of the time we aare going to work with csv files that's why knowledge with csv libarary functionalities matters

```python
import csv 
```

```python
with open('example.txt', mode = 'w') as file:
    # file.write("Sailesh is a good boy")
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Sailesh',20])
```

```python
with open('example.txt','r') as file:
    # read = file.read()
    # print(read)
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## 4. DateTime Library 

### --> It is going to helps us out with figuring out the current date and time

```python
from datetime import datetime,timedelta
```

```python
time_current = datetime.now()
```

```python
print(time_current)
```

```python
Yesterday_time = time_current - timedelta(days=1)
print(Yesterday_time)
```
