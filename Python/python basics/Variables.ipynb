## We are building a simple calculator here 

```python
value_1 = int(input("Enter first number: "))
value_2 = int(input("Enter second number: "))
print("The sum of two numbers is: ", value_1 + value_2)
print("The subtraction of two numbers is: ", value_1 - value_2)
print("The product of two numbers is: ", value_1 * value_2)
print("The division of two numbers is: ", value_1 / value_2)
```
