# Type of Operators

## 1.Arthimatic Operators

Addition (+)
Subtraction (-)
Multiplication (*)
Division (/)
Modulus (%)
floor (//)
Exponent (^)

## 2.Comaprision Operators

Greater than (>)
Less than (<)
Greater than or equal (>=)
Less than or equal (<=)
double equal to (==) for comparision
Not equal (!=)

## 3.Logical Operator

and --> whether both conditions are true or not
or --> whether if any of the operation is true or not
not --> negation or to inverse the result of the given operation

```python
#Addition operator (+)
a = 18
b = 5
print(a+b)
# Subtraction operator (-)
print(a-b)
# Multiplication operator (*)
print(a*b)
# Division operator (/)
print(a/b)
# Floor Division operator (//)
print(a//b)
# Modulus operator (%)
print(a%b)
```

```python
#comparision operator
a= 5
b= 6
print(a==b) 
print(a!=b)
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
```

```python
#Logical Operators
a = 10
b = 20
c = 30
print(a<b and b<c) # True and True = True
print(a<b or b>c) # True or False = True
print(not(a<b)) # not True = False
```

```python
#Simple Calculator as of now
value_1 = int(input("Enter first number: "))
value_2 = int(input("Enter second number: "))
operation = input("Enter operation (+, -, *, /): ")
if operation == "+":
    print("The sum of two numbers is: ", value_1 + value_2)
elif operation == "-":
    print("The subtraction of two numbers is: ", value_1 - value_2)
elif operation == "*":
    print("The product of two numbers is: ", value_1 * value_2)
elif operation == "/":
    print("The division of two numbers is: ", value_1 / value_2)
```
