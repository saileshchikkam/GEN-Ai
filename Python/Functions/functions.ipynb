# Recursive function
## We can say that a function is going to call itself again and again

## 1. Functions with a single parameters


## Normal Method

```python
# Factorial of number
def factorial_of_num(num1):
    if num1<0:
        print("Invalid number")
        return
    elif(num1 == 0 or num1 == 1):
        print(f"Factorial of {num1} is: 1")
        return
    else:
        factorial = 1
        for i in range(1,num1+1):
            factorial *= i
        print(f"Factorial of {num1} is: {factorial}")

factorial_of_num(int(input("Enter a number to find its factorial: ")))
```

## with recursive method

```python
def factorial_recursive(num2):
    if num2<0:
        print("Invalid number")
    elif num2==0 or num2==1:
        return 1
    else:
        return num2 * factorial_recursive(num2-1)

print(factorial_recursive(int(input("Enter a number to calculate the factorial: "))))        
```

## --> Positional Arguments === *args

```python
## function to add multiple numbers irrespective hoe many numbers  we are passing
def addition_pos_arguments(*num3):
    add = 0
    for i in num3:
        add += i
    print(add)
addition_pos_arguments(1,2,1,2,1,2)
```

## --> Positional Arguments === **kwargs
## key word arguments takes the input values as key,value pair, as many as

```python
def data(**user_data):
    ## user_data.items
    for key,value in user_data.items():
        print(f"{key} : {value}")

data(name = "Sai",age=21,gender="male")
```

## We want to pass keyword values pairs as tour input along with positional arguments as well

```python
def variable(*positional,**keywords):
    for i in positional:
        print(f"positional values: ",i)
    for key,value in keywords.items():
        print(f"{key} : {value}")

variable(1,2,3,4,5, name = "sailesh",Age = 20)
```
