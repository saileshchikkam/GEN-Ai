##  Q1. Function that converts temp to celsius to fahrenheit and reverse

```python
def temp_convertor(**temp):
    for key,value in temp.items():
        if key =="C":
            temperature = value * 9/5 +32
            print(f"Temperature is: ",temperature,"F")
        elif key == "F":
            temperature = (value-32) *5/9
            print(f"Temperature is: ",temperature,"C")
        else:
            print("Invalid temp value")

temp_convertor(C = 35,F = 95)
```

## Q2. define a function that would behave as a Password Strength Checker

### Strong password-->

### If your password:
###    1. length is greater than 8 characters
###    2. password must consist of atleast a lowercase character (a,b,c...)
###    3. password must consist of atleast a uppercase character (A,B,C...)
###    4. password must consists of atleast a digit (1,2,)
###    5. password must consists of atleast a special character (!,@,#,$,^,&,*,(),_)

### Weak passwords-->

### If a password doesn't match any above condition then it will be considered as a weak password.

#### 1. isLower() ---> checks that whether if the current is a lowercase character or not

#### 2. isUpper() ---> checks that whether current character is upper or not

#### 3. isdigit() ---> checks that whether if the current character is digit or not

```python
def password_strength(password):
    if len(password) < 8:
        print(f"The entered password {password} is a Weak password")

    elif not any (char.islower() for char in password):
        print(f"The entered password {password} is a Weak password")

    elif not any(char.isupper() for char in password):
        print(f"The entered password {password} is a Weak password")

    elif not any(char.isdigit() for char in password):
        print(f"The entered password {password} is a Weak password")

    elif not any(char for char in "!,@,#,$,%,^,&,*,(,)"):
        print(f"The entered password {password} is a Weak password")
        
    else:
        print(f"The entered password {password} is a Strong password")

password_strength("Hello")
```

```python
password_strength("HelloSailesh")
```

```python
password_strength("Hello123 @ amit")
```

## Q3. Check the whether if the current string is a palindrome or not

### Palindrome String
### Plaindrome string is going to have characters similar to it from both the sided or even if you will reverse the string you will be getting the given string itself

### 1 .lower()
### 2 .replace()
### 3. slicing --> [ : :-1] for reversal of string


```python
def palindrome(string):
    string_value = string.lower().replace(" ","")
    if string_value==string_value[::-1]:
        print("Palindrome")
    else:
        print("Not Palindrome")
palindrome("Mango")
```
