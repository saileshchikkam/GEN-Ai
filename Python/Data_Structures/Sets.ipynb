# Sets


### which consists of values and store values inside the braces format{}, which are not duplicate and mutable

```python
a = set()
print(type(a))
```

## Methods in set

### add() --> like append, adding at end

```python
set_value = {1, 2, 3, 4, 5}
set_value.add(6)
print(set_value)
```

## no duplicates are allowed


```python
set_value.add(1)
print(set_value)
```

```python
set_empty = set()
type(set_empty)
set_empty.add(1)
print(set_empty)
```

```python
# user asking to add the values inside a set which must be unique and non duplicate
for i in range(1, 21):
    set_empty.add(i)
print(set_empty)
```

```python
fruits = {"apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"}
print(fruits)
```

```python
veges = ["potato", "tomato", "onion", "carrot", "cabbage", "spinach"]
veges_set = set(veges)
print(veges_set)
```

```python
for items in veges:
    fruits.add(items)
fruits
```
