# Dictionary Methods

## 1. keys()
## 2. values()
## 3. items()
## 4. get()


```python
dictionary={"Name": "sailesh",
            "Phone": 9705783417,
            "Email": "sailesh@example.com",
            "Fruit" : "Apple",
            "Cell": "Oneplus"
            }
dictionary
```

## 1. Keys() --> it is a method in dictionary which returns all of the keys present inside our dictionary

```python
print(dictionary.keys())
```

## 2. Values() --> it will return all values

```python
print(dictionary.values())
```

## 3. items() --> will return the key value pairs together

```python
print(dictionary.items())
```

## 4. get() --> returns the value of the key the we are going to pass in the get method

```python
print(dictionary.get("Name"))
```

## shallow copy --> shallow copy is being used for creating a copy of the original dictionary in a way such that the changes that we are going to make the original dict wont be getting reflected into our copied dictionary and shallow copy is being done by copy()

```python
dict_1 = {"Name": "sailesh",
          "Phone": 9705783417,
          "Email": "sailesh.ch118@gmail.com"}
dict_2 = dict_1
print(dict_2)
```

```python
dict_1["Phone"] = 1234567890
dict_1
```

### changes are copied owwwwww

```python
dict_2
```

```python
dict_3 = dict_1.copy()
dict_3
dict_1["Phone"] = 9876543210
dict_3
```

```python
## I want to store the info about the multiple students
Nested_dict = {
    "Student_1": {"Name":"Sailesh","Marks": 90, "Grade": "A"},
    "Student_2": {"Name":"Ramesh","Marks": 80, "Grade": "B"},
    "Student_3": {"Name":"Suresh","Marks": 70, "Grade": "C"}
}
Nested_dict
```

```python
for outer_key,outer_value in Nested_dict.items():
    print(f"Student: {outer_key}, Details: {outer_value}")
    for inner_key,inner_value in outer_value.items():
        print(f"  {inner_key}: {inner_value}")
```

# Dictionary Comprehension

## Dictionary comprehensions helps us with using Looping statement along with conditional statement inside our dictionary

## Syntax for dictionary Comprehension

```python
set_comprehension = {x**2 for x in range(5)}
set_comprehension
```

```python
dict_comprehension = {x: x**2 for x in range(5)}
dict_comprehension
```

```python
even_dict = {x : x**2 for x in range(10) if x%2==0 }
print(even_dict)
```

# Merged Dictionary

```python
dictionary_1 = {"Name": "sailesh",
                "Phone": 9705783417,
                }
dictionary_2 = {
                "Phone": 123456789,
                "Email": "ramesh@gmail.com"}
Merged_dict = {**dictionary_1, **dictionary_2}
print(Merged_dict)
```
