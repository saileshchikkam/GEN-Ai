## Loop Control Statements
### Used for terminating the loop

## 1. break --> terminates loop when a certain condition is met

```python
for i in range(21):
    if i== 6:
        break
    else:
        print(i)
```

## 2. continue --> which helps us in iterating over a series of numbers but helps us with skipping the values in between once we meet a condition

```python
for i in range(20):
    if(i%2==0):
        continue

    else:
        print(i)
```

```python
## print no from 0 to 20 just skip 17
for i in range(21):
    if i == 17:
        continue
    print(i)
```

## 3. Pass --> which do nothing

```python
for i in range(21):
    if i == 17:
        pass # just do nothing
    print(i)
```

## Nested Loops
### Loop inside a loop

```python
for i in range(3):
    for j in range(3):
        print(f"i value is {i}, j value is {j}")
```

```python
## print the sum of n natural numbers
n = int(input("Enter the n value"))
sum = 0
for i in range(n+1):
    sum = sum + i
print(sum)
```

```python
## check whether it is prime or not
for num in range(2,20):
    if(num % 2 == 1):
        print(num)
    
```
