## with ---> It will be used for first of all opening the file as a object and there is no requirement for closing the file as well

```python
with open("example.txt", 'r') as f:
    content_data = f.read()
    print(content_data)
```

## readline() ---> reads the line in a way where in it is like the lines are stored inside the list

```python
## Reading the total lines, total words, total  characters from a file
def count_lines(file):
    with open(file,'r') as files:
        lines = files.readlines()
        total_lines = len(lines)
        total_words = sum(len(line.split()) for line in lines)
        total_characters = sum(len(line) for line in lines)
    return total_lines,total_words,total_characters

file_path = 'Total_data.txt'
Total_lines, Total_Words, Total_characters = count_lines(file_path)
print(f"Total Lines are : {Total_lines} and Total Words are: {Total_Words} and Total characters are: {Total_characters}") 
```

## w+ ---> first writing and the reading

## seek() ---> it is going to heelp us out with reading the content after writing, from the starting only but not from the cursor pointing area

```python
## write the content into a file, after that we are going to 
## Read the content from the file as well --->seek()

with open('Total_data.txt', 'w+') as write_read_file:
    write_read_file.write("We are looking into the w+ operation")
    write_read_file.write("\nIt means writing and reading")
    write_read_file.seek(0)
    content_complete_data = write_read_file.read()
    print(content_complete_data)
```
