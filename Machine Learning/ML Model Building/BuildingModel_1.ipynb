### Feature Engineering and Model Building

```python
df.head()
```

```python
## addition of Tuition_Fee & Rent_USD
## New output column = Total_Cost_USD
df['Total_Cost_USD'] = df['Tuition_USD'] + df['Rent_USD']
```

```python
df.info()
```

```python
df.head()
```

```python
df['Total_Cost_USD']
```

```python
## Feature Selection ----> Mention all column names inside a list and the list should be inside a dataframe
df = df[['Country', 'City', 'University', 'Program', 'Level', 'Duration_Years', 'Living_Cost_Index', 'Visa_Fee_USD', 'Insurance_USD', 'Total_Cost_USD']]
```

```python
df.head()
```

```python
df.info()
```

```python
## Label Encoding ----> Convert Categorical data into Numerical data
## or Feature Encoding ---> we want to feed this data to a mathematical model
import sklearn
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder() ## object of LabelEncoder class
```

```python
df['Country'] = le.fit_transform(df['Country']) # categorical to numerical
```

```python
df.head()
```

```python
df['City']=le.fit_transform(df['City'])
df['University']=le.fit_transform(df['University'])
df['Program']=le.fit_transform(df['Program'])
df['Level']=le.fit_transform(df['Level'])
```

```python
df.head() ## see everything is numerical 😁
```

```python
## now we have proper columns lets feed
## Before that lets separete the input and output columns
```

```python
x = df.iloc[:, :-1] ## input columns[rows,cols]
y = df['Total_Cost_USD'] ## output column
```

```python
x
```

```python
y
```

```python
## Feature Scaling ----> Standardization or Normalization data
from sklearn.preprocessing import StandardScaler

s = StandardScaler()
```

```python
s.fit_transform (x) ##inputs 'x'
```

```python
## no standard scalar for output column 'y' because we are predicting the cost in USD
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

### 13. Stacking Regressor ---> Ensemble learning
### ---> Combination of the multiple models of the own choice

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import StackingRegressor
from sklearn.ensemble import BaggingRegressor
```

```python
estimators = [('rf',RandomForestRegressor()),('br',BaggingRegressor())]
```

```python
stack = StackingRegressor(estimators=estimators,final_estimator=DecisionTreeRegressor())
```

```python
stack
```

```python
stack.fit(x_train,y_train)
```

```python
y_predict = stack.predict(x_test)
```

```python
mean_absolute_error(y_test,y_predict) ## ok
```
