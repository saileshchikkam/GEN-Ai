### Data Cleaning

```python
## isnull() function is used to find the missing values in the dataset. It returns a boolean value indicating whether each element is null or not.
df.isnull().sum() ## there is no null values in the dataset. So, we don't need to handle missing values.
```

```python
df.duplicated().sum() ## there is no duplicate values in the dataset. So, we don't need to handle duplicate values.
```

```python
df.dtypes
```

```python
## astype() function is used to convert the data type of a column in a DataFrame. It takes the desired data type as an argument and returns a new DataFrame with the specified column converted to that data type.
df['Duration_Years'] = df['Duration_Years'].astype('float32')
df['Rent_USD'] = df['Rent_USD'].astype('float32')
df['Tuition_USD'] = df['Tuition_USD'].astype('int32')
df['Visa_Fee_USD'] = df['Visa_Fee_USD'].astype('int32')
df['Insurance_USD'] = df['Insurance_USD'].astype('int32')
```
