# What is Seaborn ?

## Seaborn is Data Visaulization Library that is built on Top of Matplotlib and is integrated with Pandas.

# How Seaborn is better than Matplotlib ?

### 1. Syntax --- > Seaborn is Simpler --- > less lines of code is required whereas in Matplotlib a lot of Lines of code are required for bringing customization.
### 2. By default --- > Seaborn Plots looks Classic and that also without even bringing any additional customization but in Matplotlib by default its plots looks similar to those of pandas plots and only adding additional customization in Matplotlib we can bring those visuals to Life.

### 3. Plot Types --- > Matplotlib --- > General Plots (Line Plot, scatter, bar, histogram, pie ) but in Seaborn --- > We'll have all of the visuals for statistical Analysis (Kde, Box, Regression Plot,etc )
### 4. Integration --- > Matplotlib integrates with All Data Types while Seaborn integrates with Pandas DataFrame
### 5. Seaborn ---- >More Information we can extract from the plots or visuals as compared to Matplotlib
### 6. A Lot of New plots are supported in Seaborn that are not supported in Matplotlib
