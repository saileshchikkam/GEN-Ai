## Facet Plot

### Lot of information in above plots are getting overlayed one on right to solve it we can use the Facet Plot

### facet Plot can help us divide the plot as per the multiple categories of a categorical column such as let us say , we dont want to use hue and we want to divide the plot or graph as per multiple categories of a categorical column as row wise or column wise then we can go with facet plot

#### facet plot can help us seperate the info Rowwise or Columnwise

#### We can use Row or Column Parameter to mention that if we want to seperate the info rowwise or column wise

#### facet plot functunality is only possible in only figure  level plot

#### As figure level plot provides us control over the whole figure, where as axis level plot gives control over the axis itself

### 1. Axes Level Plot -- scatterplot()

```python
sns.scatterplot(data=tips,x='total_bill',y='tip',hue='sex')
```

### 2. Figure Level Plot -- relplot()

```python
sns.relplot(data=tips,x='total_bill',y='tip',row='sex',kind='scatter')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',col='sex',kind='scatter')
```

```python
## row wise ---> smoker or Not
## column wise ---> Male or female

## row 1 Smoker <------> Male  Smoker <-----> female
## row 2 NonSmoker <---> Male  NonSmoker <--> female
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',row='smoker',col='sex',kind='scatter')
```
