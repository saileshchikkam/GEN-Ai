#### 1. sum() ---- > it calculates the sum of all values present in a column or in muliple columns

#### 3. min() ---- > it calculates the minimum values of a column

#### 4. max() --- > it calculates the maximum value of a column

#### 5. median() --- > it calculates the median value of a column

#### 6. var() ---- > it calculates the variance of all values of a column
