## Pandas Series

### Pandas Series is a column of data in a Table like in Excel or Google Sheets with labels called as indexes attached to each value .

### Data stored in Pandas series in labelled

```python
pd.Series([40,50,60])
```

#### Labelling

```python
pd.Series([40,50,60],index = ['Bread','Milk','Buttter'])
```

#### naming

```python
pd.Series([40,50,60],index = ['Bread','Milk','Buttter'], name = 'price of Grocessiries items')
```

## Series is a 1D Array that can hold any type of the data like integers, float, strings (objects), boolean and mixed data types

#### Data type --- > Integers ----- > Data in a Series(Column) --- > Quantity of items in the Stock

#### Data type --- > Float ---- > Data in a Series(Column) --- > Price or Product Ratings

#### Data type --- > Strings --- > Column --- > Product Name

#### Data type --- > Boolean --- > Column --- > Is item in Stock ?

#### Data type --- > Dates/ Times --- > Column --- > Expiry Dates or Sale Dates

#### Mixed Types --- > Milk,40, True

## Creating a Series in Multiple Ways
1. List
2. Numpy Array
3. Dictionary
4. Custom Index
5. Scalar Value

### 1. List

```python
country = ['India','china','USA','UK']

pd.Series(country)
```

```python
runs_batsman = [100, 200, 300, 400]
pd.Series(runs_batsman)
```

```python
subject_names = ['Maths','Physics','Chemistry','Biology']
marks = [90, 80, 70, 60]
pd.Series(data = marks, index = subject_names, name = 'Marks of Students')
```

### 2. Numpy

```python
values = np.arange(10, 21)
pd.Series(values)
```

### 3. Dictionary to series

### Dictionary is another data Structure that stores the data in a key value pair and just because it stroes the data in a key value pair so we don't have to explicitly mention the index

```python
data_items = {'Maths': 90, 'Physics': 80, 'Chemistry': 70, 'Biology': 60}
pd.Series(data_items)
```

### 4. Custom Index

```python
values = [1,2,3,4,5]
pd.Series(data = values, index = ['a','b','c','d','e'])
```

### 5. Scalar Value

```python
## Scalar Values in Series
## same value is assigned to all the indexes in the series.

pd.Series(5, index = [0,1,2,3,4])
```

## Method Chaining

### ---> We will use multiple methods to get the required value in a way the first method's output will be the second method's input then 2nd method output will be 3rd method's input

```python
yt_updated.sort_values(ascending = False).head(1)
```

```python
yt_updated.sort_values(ascending = False).head(1).values[0]
```
