#### Numpy arrays have a fixed size at creation, unlike python lists which grows dynamically

## Changingthw size of an ndim array will create a new array and delete the original

#### The elements in a Numpy Array are required to be of the same data types (no mixed datatypes) and thus will be of the same size in memory.

#### Numpy Arrays facilitate advanced mathematical and other types of operations on large number of data typically that operations are executed more efficiently on less code by using the python built in sequence libraries
