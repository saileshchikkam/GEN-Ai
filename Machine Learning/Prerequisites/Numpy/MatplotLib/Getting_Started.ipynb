## Functions in Matplotlib

## 1. plt.figure()

#### plt.figure() --> figure() ---> parameter ---> figsize --->(len,breadth) ---> figure dimension of our graph

### plt.figure(figsize = (10,6))
 

```python
plt.figure(figsize=(10,6))
plt.plot(x,y)
```

## 2. plt.title() 
### --- > title() --- > It takes the value as the Title of your plot that what it isrepresnting as what insight we are getting through this graph , colour 



```python
plt.figure(figsize=(10,5))
plt.title("Count of Population as per the gender", color = 'red')
plt.plot(x,y)
```

## 3. plt.ylable()

```python
plt.figure(figsize=(10,5))
plt.title("Count of Population as per the gender", color = 'red')
plt.ylabel("Count of Population",color='green')
plt.plot(x,y)
```

## 4. plt.xlable()

```python
plt.figure(figsize=(10,5))
plt.title("Count of Population as per the gender", color = 'red')
plt.xlabel("Gender",color='green')
plt.ylabel("Count of Population",color='green')
plt.plot(x,y)
```

## 5. plt.grid() ---> bring the grid on our visualization

```python
plt.figure(figsize=(10,5))
plt.title("Count of Population as per the gender", color = 'red')
plt.xlabel("Gender",color='green')
plt.ylabel("Count of Population",color='green')
plt.plot(x,y)
plt.grid()
```

## 6. plt.show() ---> helps in representing the chart

```python
plt.figure(figsize=(10,5))
plt.title("Count of Population as per the gender", color = 'red')
plt.xlabel("Gender",color='green')
plt.ylabel("Count of Population",color='green')
plt.plot(x,y)
plt.grid()
plt.show()
```

## 7. plt.xlim() ---> X values [10, 1000] ---> [700, 1000]

## 8. plt.ylim() ---> X values [10, 1000] ---> [200, 1000]

## Customization for Line

#### Parameters:

#### 1. color --- > What is going to be the color of the Line --- > for eg , red , green , blue OR HTML hash code

#### 2. marker --- > how point of intersection appears -- > o , ^ , s , + , * , d, h , p

#### 3. linestyle --- > it will be defining the style of the line -

#### 4. linewidth ---- > it shows the width of the line --- > 1,1.5,2

#### 5. markersize --- > it shows the markersize or it can change the size of the marker ,default -- > 6 , --- >10,9

#### 6. lable ---> Label taht what this specific line is representing but is only helful while plotting multiple lines in a single visualization

```python
plt.plot(x,y,color='red',marker = 'o',linestyle = '--',linewidth = 3, markersize = 10, label = 'Gender Population')
```
