# Stacked Bar Chart

#### Instead of Keeping the Bars in Side by Side Direction , we can keep the Bars in the Stacked Manner i.e, One on Top of Other and in order to Figure out which category is having Maximum Sale , we can bave a look over the Length of the Bar

## Stacked Bar Chart --- > bar() --- > parameter --- > bottom --- > This bottom parameter specific that what bar i want to keep below my current bar

### Red Bar
### Green Bar ----> bottom ----> Red  Bar
### Blue Bar -----> bottom ----> Green Bar + Red Bar

#### As we want to differentiate on the basis of Bar Length Value so that's why we have to keep y_value of previous bar at the bottom

```python
plt.figure(figsize=(15,10))
plt.xlabel('Month Name')
plt.ylabel('Toatal Sales')
plt.bar(x,y,color = 'green',label = 'Mint Chip Choco',width = 0.4)
plt.bar(a,b,color = 'red',label = '85% Dark Bars',width = 0.4, bottom = y)
plt.bar(f,g,color = 'blue',label = 'Peanut Butter Cubes',width = 0.4, bottom = (y+b))
plt.xticks(np.arange(m.shape[0])-0.2,x)
plt.xticks(rotation = 'vertical')
plt.legend()
plt.show()
```
