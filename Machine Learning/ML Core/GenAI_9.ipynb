# 1. 🌐 What is an API?

**API** stands for **Application Programming Interface**.

An API provides a defined way for one software application to communicate with another application or service.

### Simple idea

```text
Application 1
      ↓
     API
      ↓
Application 2
```

The API acts as a communication interface between the two systems.

### ⭐ Easy definition

> **An API is a bridge that allows applications or services to communicate and exchange data with each other.**

# 2. 🔄 API as a Bridge Between Applications

An API allows one application to request information or an operation from another application.

```text
┌─────────────┐
│   App 1     │
└──────┬──────┘
       │ Request
       ↓
┌─────────────┐
│     API     │
└──────┬──────┘
       │
       ↓
┌─────────────┐
│   App 2     │
└──────┬──────┘
       │ Response
       └──────────────→ App 1
```

The applications do not need to directly access each other's internal implementation.

The API defines how they communicate.

# 3. 🍔 Example — Zomato and Google Maps

Suppose a food-delivery application wants to show the location of a restaurant or delivery destination on a map.

A simplified flow is:

```text
Zomato
   │
   │ Request
   ↓
 Google Maps API
   │
   │ Response
   ↓
Zomato
```

The food-delivery application can use the mapping service through its API.

### Example request

```text
"Show the location of this restaurant."
```

### API response

The mapping service can return information needed by the application to display the location/map.

### ⭐ Key idea

> **An API allows one application to use functionality or data provided by another service.**

# 4. 📡 API Request and Response

API communication commonly follows a **request-response** pattern.

### Request

The client/application sends a request:

```text
Application
     ↓
   Request
     ↓
    API
```

### Response

The service processes the request and returns a response:

```text
    API
     ↓
  Response
     ↓
Application
```

### Complete flow

```text
Client / App
     ↓
  API Request
     ↓
  API / Server
     ↓
Process request
     ↓
 API Response
     ↓
Client / App
```

The request and response often contain structured data such as JSON.

# 5. 🗄️ APIs and Application Data

An API can provide controlled access to information or functionality exposed by an application/service.

A simplified architecture is:

```text
Application
     ↓
    API
     ↓
Backend / Database
     ↓
Data
     ↓
    API
     ↓
Application
```

The application normally should not directly expose its database to another application.

Instead, an API can provide controlled access.

### ⭐ Important

```text
App → API → Backend/Database
```

The API acts as the interface through which the required information or operation is exposed.

# 6. 🧩 API as a Bridge — App 1 to App 2

Consider two applications:

```text
┌──────────┐
│  App 1   │
└────┬─────┘
     │
     ↓
┌──────────┐
│   API    │  ← Bridge
└────┬─────┘
     │
     ↓
┌──────────┐
│  App 2   │
└──────────┘
```

### Why use an API?

- Applications can communicate.
- Data can be exchanged in a controlled way.
- Services can expose selected functionality.
- Internal implementation can remain hidden.
- Different applications can be integrated together.

### Memory trick

> **API = Bridge between software systems.**

# 10. 🧠 RAG — Retrieval-Augmented Generation

**RAG** stands for:

> **Retrieval-Augmented Generation**

RAG combines two important stages:

```text
Retrieval
    +
Generation
```

### Retrieval

Retrieve relevant information from an external source.

### Generation

Use an AI/LLM to generate a response using the retrieved information.

### ⭐ Core idea

> **RAG retrieves relevant information first and then augments the AI model's generation with that information.**

# 11. 🌦️ Weather App Example

Consider a user asking:

> **"What will the weather be next Saturday?"**

A normal LLM may not have access to the latest weather information.

So the application can retrieve current/future weather data from a weather service API.

### Simplified flow

```text
User
 ↓
AI Application
 ↓
Weather API
 ↓
Weather Data
 ↓
AI Model
 ↓
Summarized Answer
 ↓
User
```

For example, the retrieved information could contain:

```text
Date
Temperature
Humidity
Rain probability
Weather condition
Wind speed
```

The AI model can then summarize the information for the user.

# 12. 🔍 Retrieval vs Generation

RAG has two distinct stages.

## A. Retrieval

The system obtains relevant information from an external source.

Example:

```text
Weather API
     ↓
Weather information
```

This is the **retrieval** stage.

---

## B. Generation

The AI model uses the retrieved information to produce a user-friendly answer.

Example:

```text
Retrieved weather information
          ↓
        AI Model
          ↓
"Next Saturday is expected to be..."
```

This is the **generation** stage.

### Memory trick

```text
Retrieval  → Get information

Generation → Create answer
```

# 13. 🔄 RAG Workflow

A simplified RAG workflow is:

```text
                USER
                  │
                  ↓
             User Query
                  │
                  ↓
             AI / Agent
                  │
                  ↓
            RETRIEVAL
                  │
          ┌───────┴────────┐
          ↓                ↓
       API / DB         Documents
          │                │
          └───────┬────────┘
                  ↓
           Relevant Data
                  │
                  ↓
              LLM / AI
                  │
             GENERATION
                  │
                  ↓
           Final Response
                  │
                  ↓
                 USER
```

### ⭐ Main principle

The model does not need to rely only on information stored in its parameters.

Relevant external information can be retrieved and supplied as context.

# 14. 🧾 JSON in API Communication

The notes connect APIs with JSON-based data exchange.

A request may contain JSON such as:

```json
{
    "location": "Hyderabad",
    "date": "Saturday"
}
```

The weather service can return JSON such as:

```json
{
    "location": "Hyderabad",
    "date": "Saturday",
    "temperature": 29,
    "rain_probability": 40,
    "condition": "Cloudy"
}
```

The AI application can then use the returned data.

### Flow

```text
User
 ↓
AI Application
 ↓
JSON Request
 ↓
Weather API
 ↓
JSON Response
 ↓
AI Model
 ↓
Summarized Information
 ↓
User
```

# 15. 🤖 RAG Is More Than Just Retrieval

An important point from the notes is that simply calling an API and returning its result is **not necessarily RAG**.

### Simple retrieval

```text
User
 ↓
Weather API
 ↓
Weather data
 ↓
User
```

This is basically retrieval.

---

### RAG-style flow

```text
User
 ↓
Retrieve relevant information
 ↓
Provide retrieved information as context
 ↓
AI Model
 ↓
Generate / summarize answer
 ↓
User
```

### ⭐ Key difference

> **Retrieval gets the information; generation uses that information to produce the final response.**

# 16. 🆚 Retrieval vs RAG vs Agent

These concepts are related but not identical.

| Concept | Main purpose |
|---|---|
| Retrieval | Obtain relevant information |
| RAG | Retrieve information + generate an answer using it |
| AI Agent | Use tools and take steps/actions toward a goal |

### Example

#### Retrieval

```text
User → Weather API → Weather Data
```

#### RAG

```text
User
 ↓
Weather API
 ↓
Weather Data
 ↓
LLM
 ↓
Summarized Answer
```

#### Agentic system

```text
User
 ↓
Agent
 ↓
Decide which tool is needed
 ↓
Weather API
 ↓
Possibly perform additional tool calls
 ↓
LLM
 ↓
Final response/action
```

# 20. 🧠 Why APIs Are Important for RAG

RAG applications often need information that is:

- current,
- external to the model,
- frequently changing,
- private or organization-specific,
- too large to place directly into the prompt.

APIs can provide access to such information.

### Example

Weather information changes frequently:

```text
LLM alone
   ↓
May not know current forecast

LLM + Weather API
   ↓
Retrieve current data
   ↓
Generate useful response
```

# 21. ⚠️ Important Point — API Data Must Be Controlled

When an AI application uses an API, developers should consider:

```text
✔ Authentication
✔ API keys
✔ Rate limits
✔ Error handling
✔ Input validation
✔ Output validation
✔ Data privacy
✔ Access permissions
```

Never expose private API keys directly in client-side code or public repositories.

### Typical secure pattern

```text
User
 ↓
Frontend
 ↓
Your Backend
 ↓
API
 ↓
Response
```

Sensitive credentials should normally be managed on the server side.

# 22. 📝 Important Definitions

### API

> **An API is an interface that allows software applications or services to communicate and exchange data or functionality.**

---

### JSON

> **JSON is a lightweight structured data format commonly used for exchanging information between applications and APIs.**

---

### Retrieval

> **Retrieval is the process of obtaining relevant information from an external source such as a database, document store or API.**

---

### Generation

> **Generation is the process in which an AI model produces an answer or content based on the available input/context.**

---

> **RAG (Retrieval-Augmented Generation) combines information retrieval with AI generation by supplying retrieved information to a generative model as context.**

### Agent

# 23. 🎓 Interview Questions & Answers

### Q1. What is an API?

**Answer:**  
An API is an interface that enables software applications or services to communicate with each other and exchange data or functionality.

---

### Q2. Why are APIs important in AI applications?

**Answer:**  
APIs allow AI applications to connect with external services such as LLMs, weather systems, maps, databases and other tools.

---

### Q3. What is the request-response model?

**Answer:**  
A client sends a request to an API, the server processes it and returns a response containing the requested data or result.

---

### Q4. What is RAG?

**Answer:**  
RAG stands for Retrieval-Augmented Generation. It retrieves relevant external information and provides it to a generative AI model so that the model can generate a more relevant answer.

---

### Q5. What is the difference between retrieval and generation?

**Answer:**  
Retrieval obtains relevant information, while generation uses that information to create the final response.

---

### Q6. Can an API be used as a RAG data source?

**Answer:**  
Yes. An API can provide external information that is retrieved and then supplied as context to an LLM for generating an answer.

---

### Q7. Is every API call RAG?

**Answer:**  
No. Calling an API and directly returning its result is retrieval. It becomes RAG when retrieved information is used as context for a generative model to produce the answer.

---

### Q8. What is the role of JSON in API communication?

**Answer:**  
JSON provides a structured and widely supported format for sending requests and receiving responses between applications and APIs.

---

### Q9. Give a real-world RAG example.

**Answer:**  
A weather assistant can retrieve forecast information from a weather API and provide that retrieved information to an LLM, which summarizes it into a natural-language response.

---

### Q10. What is the difference between RAG and an AI agent?

**Answer:**  
RAG primarily focuses on retrieving information and using it during generation. An AI agent can additionally reason about a task, choose tools and perform multiple actions to accomplish a goal.
