# 1. 🎯 Problem Framing

The first step is to clearly understand the **business problem**.

A business problem is converted into an appropriate **Machine Learning problem**.

```text
Business Problem
      │
      ▼
Understand the goal
      │
      ▼
Can ML solve it?
      │
      ▼
Define ML problem
      │
      ▼
Define inputs + target + success metric
```

### Example

Suppose a company receives a large number of customer emails.

**Business problem:**
> Automatically identify unwanted/spam emails.

**ML problem:**
> Build a **classification model** that predicts whether an email is `Spam` or `Not Spam`.

| Business Requirement | ML Formulation |
|---|---|
| Identify spam emails | Classification |
| Predict house price | Regression |
| Group customers | Clustering |
| Detect unusual transactions | Anomaly Detection |
| Recommend products | Ranking / Recommendation |

### ✅ Questions to answer during problem framing

1. What exactly are we trying to solve?
2. Is ML actually needed?
3. What is the input data?
4. What should the model predict?
5. What metric defines success?
6. What are the constraints?
7. How will the prediction be used in the product?

# 3. 🧹 Data Preprocessing / Data Cleaning

Raw data is rarely ready for direct model training.

After gathering data, we may need to combine datasets and clean them.

```text
Dataset 1 ──┐
            ├──► Merge / Concatenate ──► Clean Data
Dataset 2 ──┘
```

### Common data-cleaning problems

| Problem | Example | Possible Action |
|---|---|---|
| Missing values | Age = `NaN` | Impute / remove |
| Wrong data types | `"25"` stored as text | Convert type |
| Duplicate values | Same customer twice | Remove duplicates |
| Outliers | Extremely unusual value | Investigate / transform |
| Inconsistent values | `India`, `india`, `IND` | Standardize |
| Different columns | Dataset A ≠ Dataset B | Align / transform |

### Concatenation vs Merge

**Concatenation** generally stacks datasets along rows or columns.

```text
Data A
Data B
Data C
   ↓
Combined Data
```

**Merge** combines datasets using one or more common keys.

```text
Customers              Orders
customer_id ────────── customer_id
       │                    │
       └────── MERGE ───────┘
```

### 🎯 Goal

> Produce a reliable, consistent dataset that can safely move into analysis and modeling.

````python
import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "score": [80, 90, 75]
})

df2 = pd.DataFrame({
    "id": [4, 5],
    "score": [88, 92]
})

combined = pd.concat([df1, df2], ignore_index=True)

print(combined)
````

# 5. 🧩 Feature Engineering & Feature Selection

After understanding the data, we prepare the features that will be given to the model.

### Feature Engineering

**Feature engineering** creates useful new features from existing data.

```text
Existing Data
      │
      ▼

      │
      ├── transform
      ├── combine
      ├── extract
      └── create
      │
      ▼
New / Better Features
      │
      ▼

```

### Example

Suppose we have:

| Feature | Value |
|---|---:|
| Rooms | 3 |
| Area | 1500 sq.ft |
| Bathrooms | 2 |

We could create:

```text

             = 1500 / 3
             = 500 sq.ft
```

That newly created feature may contain useful information.

---

## 🎯 Feature Selection

Feature selection means choosing the **most useful existing features** for the model.

Suppose we start with:

```text
20 Features
     │
     ▼
Feature Selection
     │
     ▼
8 Important Features
     │
     ▼
Model
```

### Why avoid unnecessary features?

Too many irrelevant features can:

- increase training time,
- increase computational cost,
- make models harder to interpret,
- increase the risk of overfitting.

> **Feature engineering creates useful features. Feature selection chooses useful features.**

# 8. 📏 Model Evaluation

Evaluation tells us how well the model's predictions perform.

The correct metric depends on the ML task.

## Regression Metrics

### MSE — Mean Squared Error

\[
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
\]

### RMSE — Root Mean Squared Error

\[
RMSE = \sqrt{MSE}
\]

### MAE — Mean Absolute Error

\[
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
\]

Lower values generally indicate smaller prediction errors.

---

## Classification Metrics

Common classification metrics include:

- **Accuracy**
- **Precision**
- **Recall**
- **F1-score**
- **Confusion Matrix**

### Confusion Matrix

```text
                  Actual
                Positive Negative
Predicted Pos.     TP       FP
Predicted Neg.     FN       TN
```

### Accuracy

\[
Accuracy = \frac{TP+TN}{TP+TN+FP+FN}
\]

### Precision

\[
Precision = \frac{TP}{TP+FP}
\]

### Recall

\[
Recall = \frac{TP}{TP+FN}
\]

> ⚠️ Accuracy alone may not be suitable for imbalanced datasets. Select metrics according to the actual business objective.

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

actual = [100, 120, 90, 150]
predicted = [105, 115, 95, 145]

mae = mean_absolute_error(actual, predicted)
mse = mean_squared_error(actual, predicted)
rmse = np.sqrt(mse)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
````

# 9. 🎛️ Hyperparameter Tuning

A model has:

- **Parameters** learned from training data.
- **Hyperparameters** selected before/during the training process.

Hyperparameter tuning searches for a configuration that produces better validation performance.

```text
Many possible hyperparameters
          │
          ▼
   Search configurations
          │
     ┌────┴────┐
     ▼         ▼
 Grid Search  Random Search
     │         │
     └────┬────┘
          ▼
   Best configuration
          │
          ▼
   Better validation
   performance
```

### Common approaches

#### Grid Search

Tests combinations from a predefined parameter grid.

#### Random Search

Samples combinations from specified distributions/ranges.

In scikit-learn:

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
```

> **Important:** Hyperparameter tuning should be performed using validation / cross-validation rather than repeatedly optimizing against the final test set.

# 10. 🧠 Ensemble Learning

**Ensemble learning** combines predictions from multiple models to obtain a stronger overall model.

Suppose:

```text
Model 1 → 90% validation score
Model 2 → 93%
Model 3 → 91%
```

Instead of relying on only one model, an ensemble can combine their information.

```text
Model 1 ──┐
Model 2 ──┼──► Ensemble ──► Final Prediction
Model 3 ──┘
```

### Main idea

> **Many useful models can sometimes produce a stronger and more robust prediction together than a single model.**

### Two important ensemble families

| Method | Basic Idea |
|---|---|
| **Bagging** | Train models independently/in parallel and combine predictions |
| **Boosting** | Train models sequentially so later models focus on previous errors |

### Random Forest

Random Forest is a well-known **bagging-based ensemble** of decision trees.

```text
Tree 1 ──┐
Tree 2 ──┤
Tree 3 ──┼──► Combine ──► Final Output
Tree 4 ──┤
Tree 5 ──┘
```

### Boosting

Examples include:

- XGBoost
- AdaBoost
- Gradient Boosting

```text
Model 1
   ↓
Errors
   ↓
Model 2 learns from errors
   ↓
Errors
   ↓
Model 3
   ↓
Combined stronger model
```

# 13. 🔄 Optimization & Retraining

Real-world data changes.

Customer behavior, product usage, market conditions, and data distributions can evolve over time.

Therefore, an ML system may require continuous improvement.

```text
Production Model
      │
      ▼
New Data
      │
      ▼
Monitor Performance
      │
      ├── Good → Continue
      │
      └── Degraded
             │
             ▼
          Retrain
             │
             ▼
        Re-evaluate
             │
             ▼
          Redeploy
```

### Why optimize?

- Improve prediction quality
- Reduce errors
- Handle changing data
- Reduce latency/cost
- Incorporate new training data
- Prevent performance degradation

### ♻️ ML is a cycle

```text
Build → Deploy → Test → Monitor → Retrain → Redeploy
  ↑                                      │
  └──────────────────────────────────────┘
```

> **Key lesson:** An ML product is not “train once and forget.” Production ML requires monitoring, feedback, maintenance, and controlled retraining.

# 17. 🎓 Exam-Ready Definitions

### Problem Framing
> Problem framing is the process of converting a real-world business problem into a well-defined machine learning problem with clear inputs, outputs, constraints, and evaluation criteria.

### Data Gathering
> Data gathering is the process of collecting relevant data from sources such as CSV files, APIs, databases, and web pages.

### Data Preprocessing
> Data preprocessing is the process of cleaning, transforming, integrating, and preparing raw data for machine learning.

### EDA
> Exploratory Data Analysis is the process of analyzing datasets using statistics and visualizations to understand distributions, relationships, patterns, and anomalies.

### Feature Engineering
> Feature engineering is the process of creating useful features from existing data to improve machine learning performance.

### Feature Selection
> Feature selection is the process of selecting the most relevant features for model training.

### Model Evaluation
> Model evaluation measures how well an ML model performs using suitable metrics such as MAE, RMSE, accuracy, precision, recall, or F1-score.

### Deployment
> Model deployment is the process of integrating a trained ML model into a production application so that it can generate predictions for real users or systems.

### Optimization
> ML optimization involves improving model performance, efficiency, reliability, and maintaining performance as production data changes.
