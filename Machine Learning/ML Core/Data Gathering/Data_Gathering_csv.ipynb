## read_csv() --- > 1. sep (seperation) --- > how values are seperated from each other in a given file

#### csv --- > amit,28 --- > sep=","
##### tsv -- > amit\t28 ---- > sep ="\t"

```python
pd.read_csv('katalog_gempa_v2.tsv',sep='\t')
```

### 2. names ---> names is going to take a list of names that we can provide to our column names

#### ----> names = ['Name','Age']

#### we can change the names

```python
pd.read_csv('Titanic-Dataset.csv',names = ['P_ID','Survived','P_Class','P_Name','Gender','Age','Sibling or Spouse','Parch','Ticket_ID','Fare','Cabin','Starting Journey'])
```
