### 12. Convertor
### it is acheived by the functions
### to change the value in specific row and column

```python
pd.read_csv('Titanic-Dataset.csv')
```

```python
def station(value):
    if value == 'Q':
        return 'Queenstown'
    else:
        return value
```

```python
pd.read_csv('Titanic-Dataset.csv',converters = {'Embarked':station})
```

### 13. na_values
### Handling with misssing values like NaN values
### Considering all missing vaalues not only NaN values
### na_values helps us out with mentioning what values to be considered as a missing value
### syntax = na_values['\',':','No',NaN]

```python
pd.read_csv('Titanic-Dataset.csv',na_values = ['NaN','No'])
```
