# 1. 📊 Data Gathering

## Definition

**Data gathering** is the process of collecting the data required to solve a machine-learning problem.

Once the business problem has been converted into an ML problem, we need to determine:

- What data is required?
- Where is the data available?
- How can we access it?
- Is the data relevant?
- Is the data in a usable format?

### Example

Suppose our ML problem is:

> **Predict customer churn.**

We may need:

```text
Customer information
+
Usage information
+
Subscription information
+
Payment information
+
Historical churn information
```

These data may come from different sources.

---

## ⭐ Key Principle

> **Do not collect data just because it is available. Collect data because it helps answer the ML problem.**

# 3. 📄 CSV Files

## What is a CSV?

**CSV** stands for **Comma-Separated Values**.

It is a simple tabular data format where values are generally separated by commas.

Example:

```text
name,age,city
Amit,22,Hyderabad
Rahul,24,Bengaluru
Priya,23,Chennai
```

The comma separates columns.

### Python example

```python
import pandas as pd

df = pd.read_csv("data.csv")

print(df.head())
```

### Typical use

CSV files are commonly used when:

- datasets are shared as files,
- data is exported from another system,
- a project already contains a prepared dataset,
- data is small or moderate enough to handle as files.

---

## ⚠️ Important

Do not assume that every file containing tabular data is a standard comma-separated CSV.

# 4. 📝 CSV vs TSV

The notes highlight that there can be different separators.

## CSV

**CSV → comma-separated**

```text
name,age,city
Amit,22,Hyderabad
```

Separator:

```text
,
```

## TSV

**TSV → tab-separated values**

```text
name    age    city
Amit    22     Hyderabad
```

Separator:

```text
\t
```

### Reading TSV using pandas

```python
import pandas as pd

df = pd.read_csv("data.tsv", sep="\t")
```

### Comparison

| Format | Typical separator |
|---|---|
| CSV | `,` |
| TSV | Tab (`\t`) |

### ⭐ Important lesson

Before loading a file, understand its actual structure and delimiter.

# 5. ⚠️ `pd.read_csv()` Is Not the Whole Data-Gathering Problem

A common beginner assumption is:

```python
pd.read_csv("file.csv")
```

and then thinking the data-gathering problem is finished.

In real ML projects, there can be many challenges.

### Questions to consider

```text
Where is the data?
What format is it?
What delimiter is being used?
Is the data complete?
Is the data current?
Are there multiple files?
Do the columns match?
Can the data be accessed reliably?
```

### Example

You may receive:

```text
customers.csv
transactions.csv
subscriptions.csv
```

The final ML dataset may require combining information from all three.

Therefore:

> **Loading a file is only one part of data gathering.**

# 6. 🗄️ Databases

A **database** stores and manages structured information so that applications and users can retrieve it efficiently.

Instead of receiving one static CSV file, a company may keep its operational data inside a database.

### Example

A company might have:

```text
customers
subscriptions
payments
transactions
support_tickets
```

These tables can contain information needed for an ML system.

---

## Why databases are useful

They are useful when:

- data is continuously generated,
- many applications use the same data,
- the dataset is too large or dynamic for simple files,
- structured queries are required.

### Typical workflow

```text
Database
   ↓
SQL Query
   ↓
Required records
   ↓
Python / ML pipeline
   ↓
Dataset
```

# 7. 🔎 SQL and Database Data Gathering

**SQL (Structured Query Language)** is commonly used to retrieve data from relational databases.

For example:

```sql
SELECT customer_id, age, monthly_usage, churn
FROM customers;
```

The result can then be loaded into a Python-based ML workflow.

### Conceptual flow

```text
Database
    ↓
SQL
    ↓
Selected data
    ↓
DataFrame
    ↓
Machine Learning
```

### ⭐ Why use SQL?

Instead of pulling everything from a database, we can request only the information that is relevant to the ML problem.

This can reduce:

- unnecessary data transfer,
- memory usage,
- processing time,
- complexity.

# 8. 🔌 APIs

## What is an API?

**API** stands for **Application Programming Interface**.

An API provides a structured way for one software system to communicate with another.

For data gathering, an API can provide data to our application or ML pipeline.

### Typical flow

```text
Our ML System
      ↓
     API Request
      ↓
External Application / Service
      ↓
    API Response
      ↓
      JSON
      ↓
Our Data Pipeline
```

---

## Example

A weather service may expose an API that returns:

```json
{
    "city": "Hyderabad",
    "temperature": 30,
    "humidity": 65
}
```

Our Python program can request the data and process it.

---

## Common Python library

```python
import requests

response = requests.get("API_ENDPOINT")

data = response.json()
```

> The actual endpoint, authentication and request parameters depend on the API provider.

# 9. 🧾 JSON — Common API Data Format

APIs frequently return data in **JSON** format.

**JSON** stands for **JavaScript Object Notation**.

Example:

```json
{
    "name": "Amit",
    "age": 22,
    "city": "Hyderabad"
}
```

JSON represents information using key-value pairs and nested structures.

### Python

```python
import requests

response = requests.get("API_ENDPOINT")
data = response.json()

print(data)
```

### API → JSON → Python

```text
API
 ↓
JSON response
 ↓
Python object
 ↓
Data processing
```

### ⭐ Important

API responses are not automatically ready for ML.

We may still need to:

```text
extract fields
→ flatten nested data
→ convert types
→ handle missing values
→ create a tabular dataset
```

# 10. 🌐 Web Scraping

Another method of data gathering is **web scraping**.

### Definition

Web scraping is the process of programmatically extracting useful information from web pages.

Typical flow:

```text
Website
   ↓
Fetch web page
   ↓
Read HTML
   ↓
Find required elements
   ↓
Extract data
   ↓
Store dataset
```

### Common Python tools

```text
BeautifulSoup
Requests
Selenium
Playwright
```

For example, **BeautifulSoup** can help parse HTML after a page has been retrieved.

---

## Why use web scraping?

Sometimes the required information:

- is publicly displayed on websites,
- is not provided through a convenient downloadable dataset,
- is not available through an accessible API.

In such cases, web scraping may be considered, subject to the website's terms, access rules and applicable laws.

# 11. 🧭 Web Scraping — The Basic Idea

The notes mention that web scraping involves navigating to the **HTML code of a website**.

A simplified structure is:

```text
Website
   ↓
HTML
   ↓
Identify required elements
   ↓
Extract information
   ↓
Save data
```

### Example HTML

```html
<div class="product">
    <h2>Laptop</h2>
    <span class="price">₹50,000</span>
</div>
```

A scraper can locate the relevant HTML elements and extract:

```text
Product → Laptop
Price   → ₹50,000
```

The extracted records can then be converted into a structured dataset.

---

## ⚠️ Important

Web scraping can be technically difficult because websites may have:

- changing HTML structures,
- dynamically generated content,
- pagination,
- rate limits,
- authentication,
- anti-bot mechanisms.

Always respect applicable terms, robots directives where relevant, access restrictions, and laws.

# 12. 🧠 "Poor Algorithm + Lots of Data" vs "Great Algorithm + Less Data"

The notes introduce an important philosophy:

```text
Poor algorithm
+
Lots of data
```

versus:

```text
Great algorithm
+
Less data
```

The broader ML lesson is that **data quality and quantity can strongly influence model performance**, but there is no universal rule that more data always beats a better algorithm.

### Better way to think about it

```text
Useful data
+
Good data quality
+
Appropriate features
+
Suitable model
+
Correct evaluation
=
Strong ML system
```

### ⭐ Key takeaway

Do not focus only on the algorithm.

Data quality, relevance, coverage and representativeness are also extremely important.

# 13. 📈 More Data Is Not Automatically Better

Suppose we want to predict customer churn.

### Dataset A

```text
1,000,000 records
```

but contains:

```text
irrelevant features
duplicate records
incorrect labels
large amounts of missing data
```

### Dataset B

```text
100,000 records
```

with:

```text
relevant features
accurate labels
representative customers
good data quality
```

Dataset B may be much more useful.

Therefore:

> **Quality and relevance matter alongside quantity.**

### Data-gathering checklist

```text
✔ Relevant
✔ Representative
✔ Accurate
✔ Consistent
✔ Sufficient
✔ Accessible
✔ Legally/ethically usable
```

# 14. 🧩 Multiple Sources Can Be Combined

A real ML project may require data from several sources.

For example:

```text
CSV
 ↓
Customer profile

Database
 ↓
Transactions

API
 ↓
External information

Web scraping
 ↓
Public market information
```

These can eventually be combined:

```text
CSV ───────────┐
Database ──────┤
API ───────────┼──→ Unified Dataset
Web Scraping ──┘
                       ↓
                Data Cleaning
                       ↓
                    EDA
                       ↓
                Feature Engineering
```

### ⭐ Important

Different sources often have:

- different formats,
- different column names,
- different identifiers,
- different update frequencies.

Therefore, combining them requires careful data engineering.

# 18. 🧠 Data Gathering vs Data Cleaning

These are different stages.

## Data Gathering

Focuses on:

```text
Where do we get the data?
How do we access it?
How do we collect it?
```

Examples:

```text
CSV
Database
API
Web scraping
```

---

## Data Cleaning / Preprocessing

Focuses on:

```text
Is the collected data usable?
Are values missing?
Are there duplicates?
Are data types correct?
Are there incorrect values?
```

### Flow

```text
Data Sources
     ↓
Data Gathering
     ↓
Raw Data
     ↓
Data Cleaning / Preprocessing
     ↓
Clean Data
```

> **Gather first, then clean and prepare the data for analysis and modelling.**

# 19. ⚠️ Common Data-Gathering Challenges

Real-world data gathering can be much harder than loading a local file.

### CSV challenges

```text
Different delimiters
Missing files
Encoding problems
Inconsistent columns
Large files
```

### Database challenges

```text
Complex schemas
Multiple tables
SQL joins
Access permissions
Large query results
```

### API challenges

```text
Authentication
Rate limits
Pagination
Changing response formats
Network failures
API quotas
```

### Web-scraping challenges

```text
Changing HTML
Dynamic pages
Pagination
Rate limiting
Access restrictions
Terms of service
```

### ⭐ Professional lesson

> **The hardest part is often obtaining reliable data, not writing the first model.**

# 20. 🔐 Data Access and Responsible Collection

Data gathering should also consider responsible data use.

Before collecting data, ask:

```text
Do we have permission to access it?
Is there an official API?
Are there usage restrictions?
Does the data contain sensitive information?
Can we legally and ethically use it?
How should credentials be protected?
```

### Never hard-code secrets

❌ Avoid:

```python
API_KEY = "my-secret-key"
```

Prefer environment variables or a secure secret-management system.

```python
import os

API_KEY = os.getenv("API_KEY")
```

### ⭐ Rule

> **Technical ability to access data does not automatically mean permission to use it.**
