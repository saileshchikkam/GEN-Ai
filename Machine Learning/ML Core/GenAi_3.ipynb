# 1. 🟦 Supervised Machine Learning

In **supervised learning**, the model is trained using both **input features (X)** and a **known target/output (y)**. The model learns the relationship between them and predicts outputs for new inputs.

```text
Training Data → Input (X) + Correct Output (y)
                         ↓
                  🧠 Model Training
                         ↓
                  Learned Patterns
                         ↓
                    New Input
                         ↓
                 Predicted Output (ŷ)
```

### 👨‍🏫 Analogy

Think of a **senior/teacher** training a **junior/learner**: examples and correct answers are provided first; later, the junior is given a new problem and must produce the answer.

| Property | Supervised Learning |
|---|---|
| Input features | Available |
| Correct output | Available |
| Objective | Learn X → y relationship |
| Main tasks | Regression, Classification |
| Evaluation | Compare prediction with actual value |

# 2. 📊 Supervised Dataset Example

| IQ | CGPA | Placement |
|---:|---:|---:|
| 111 | 8.8 | 1 |
| 78 | 7.4 | 1 |
| 65 | 5.8 | 0 |
| … | … | … |

Here **IQ** and **CGPA** are input features and **Placement** is the target. `1` may represent placed and `0` not placed.

```text
New Student → [IQ, CGPA] → Trained Model → Placement Prediction
```

````python
from sklearn.linear_model import LogisticRegression

X = [[111,8.8],[78,7.4],[65,5.8],[100,8.0],[90,7.8]]
y = [1,1,0,1,1]

model = LogisticRegression()
model.fit(X, y)
print("Prediction:", model.predict([[85,7.2]])[0])
````

# 3. 🔢 Regression

**Regression** is a supervised learning task where the target is a **numerical/continuous value**.

### Examples
- 🏠 House-price prediction
- 📈 Stock-price prediction
- 🌡️ Temperature prediction
- 💰 Salary prediction
- 📦 Demand forecasting

```text
Input Features → Regression Model → Numerical Output
```

> **Rule of thumb:** If the required output is a number or continuous quantity, think **Regression**.

# 4. 🏷️ Classification

**Classification** is a supervised learning task where the output is a **category/class**.

| Problem | Classes |
|---|---|
| 🐶 Animal recognition | Dog / Cat |
| 📧 Email filtering | Spam / Not Spam |
| 🩺 Medical screening | Positive / Negative |
| 💳 Transaction analysis | Fraud / Genuine |

```text
New Input → Classifier → Class A / Class B / ...
```

### Regression vs Classification

| Aspect | Regression | Classification |
|---|---|---|
| Output | Numerical | Categorical |
| Example | House price | Spam detection |
| Question | “How much?” | “Which class?” |

# 5. 🟨 Unsupervised Machine Learning

In **unsupervised learning**, the model receives input data but no predefined target labels. It searches for useful structure or patterns.

```text
Input Data
   ↓
Unsupervised Model
   ↓
Patterns / Groups / Structure
```

### Major tasks

- 🔵 **Clustering**
- 📉 **Dimensionality Reduction**
- 🚨 **Anomaly Detection**
- 🛒 **Association Rule Learning**

> Unsupervised learning does not mean “no information”; it means there is no supplied target label guiding the learning process.

# 6. 🔵 Clustering

**Clustering** groups similar data points into clusters without being given the class names beforehand.

```text
                 IQ
                  ↑
        ● ● ●          ● ●
       ● ● ●           ● ●
          ● ●       ● ●
────────────────────────────→ CGPA

        Cluster 1       Cluster 2
```

### Student example

Using features such as IQ, CGPA, attendance and study hours, an algorithm may discover groups such as:

1. High CGPA + High IQ
2. High CGPA + Moderate IQ
3. Low CGPA + High IQ
4. Low CGPA + Low IQ

The labels/names are interpreted **after** clusters are discovered.

> **Core idea:** Similar datapoints → same cluster/group.

````python
from sklearn.cluster import KMeans

students = [[8.8,111],[8.5,105],[7.4,78],[7.2,80],[5.8,65],[6.0,68]]
model = KMeans(n_clusters=2, random_state=42, n_init=10)
labels = model.fit_predict(students)
for student, label in zip(students, labels):
    print(student, "-> Cluster", label)
````

# 7. 📉 Dimensionality Reduction

Real-world datasets can have hundreds or thousands of features. Dimensionality reduction transforms high-dimensional data into a smaller representation while trying to preserve important information.

### MNIST intuition

A `28 × 28` grayscale image contains **784 pixel positions**.

```text
28 × 28 image
     ↓
784 pixel values
     ↓
Dimensionality Reduction
     ↓
2 / 3 / other compact representation
```

A common classical technique is **PCA (Principal Component Analysis)**.

> Goal: reduce unnecessary/redundant dimensions while retaining useful information.

# 8. 🚨 Anomaly Detection

**Anomaly detection** identifies observations that behave very differently from the normal pattern. An anomaly is also called an **outlier**.

```text
Normal:   ● ● ● ● ● ● ●
           ● ● ● ● ●

                          ✕
                    Anomaly / Outlier
```

### Applications
- Fraud detection
- Machine-failure detection
- Network/security monitoring
- Sensor-error detection
- Unusual financial activity

> **Goal:** Find observations that do not fit the expected pattern.

# 9. 🛒 Association Rule Learning

Association rule learning discovers useful relationships between items/events.

### Market-basket example

Suppose transaction data contains purchases from **10,000 customers**. If many customers purchase combinations such as milk and butter, we can investigate what other products frequently occur with them.

```text
Milk + Butter → Bread
```

This represents an **association**, not necessarily causation.

### Uses
- 🛍️ Market-basket analysis
- 🛒 Product recommendation
- 📦 Cross-selling
- 🌐 Web usage analysis

> **Core idea:** Discover co-occurrence relationships in data.

# 10. 🟩 Semi-Supervised Machine Learning

Semi-supervised learning combines a **small labeled dataset** with a **large unlabeled dataset**.

```text
Small labeled dataset ─┐
                       ├──► Semi-Supervised Model → Prediction
Large unlabeled dataset┘
```

### Why use it?

Manual labeling is often expensive and time-consuming. If only some examples can be labeled, the unlabeled examples can still provide useful information about the data distribution.

### Example

```text
10,000 images
├── 500 labeled
└── 9,500 unlabeled
```

Facial-recognition/photo systems are an intuitive application area for this idea.

# 11. 🟥 Reinforcement Learning

**Reinforcement Learning (RL)** trains an agent through interaction with an environment. The agent takes actions, receives rewards/penalties, and improves its strategy.

```text
        ┌───────────────┐
        │  ENVIRONMENT  │
        └───────┬───────┘
                │ State / Feedback
                ▼
          ┌──────────┐
          │  AGENT   │
          └────┬─────┘
               │ Action
               ▼
        ┌───────────────┐
        │  ENVIRONMENT  │
        └───────┬───────┘
                │
                ▼
              Reward
```

### Game example

```text
Action → Good outcome → Positive reward
Action → Bad outcome  → Negative reward
```

The objective is to learn a policy that maximizes expected long-term reward.

> ⚠️ RL is better described as learning from **interaction and feedback**, rather than simply “training without data.”

````python
# Tiny reinforcement-learning-style feedback example
actions = ["move_left", "move_right", "stay"]
rewards = {"move_left": -1, "move_right": 10, "stay": 0}
best_action = max(actions, key=lambda a: rewards[a])
print("Best action:", best_action)
print("Reward:", rewards[best_action])
````

# 12. ⚖️ Comparison of Machine Learning Types

| Feature | 🟦 Supervised | 🟨 Unsupervised | 🟩 Semi-Supervised | 🟥 Reinforcement |
|---|---|---|---|---|
| Labeled outputs | ✅ Yes | ❌ No | 🟡 Some | ❌ Not traditional labels |
| Main signal | Target `y` | Data structure | Labels + data structure | Reward |
| Typical tasks | Regression, Classification | Clustering, PCA, Anomaly, Association | Recognition / Classification | Games, Robotics, Control |
| Objective | Predict | Discover | Learn with fewer labels | Maximize reward |

# 14. 🐍 ML Workflow in Python

```text
Data → Preprocess → Choose Model → Train → Predict → Evaluate
```

For classical ML, `scikit-learn` provides implementations for many supervised and unsupervised algorithms.

The exact workflow changes by problem, but the core idea remains:

**Data → Learning → Prediction/Structure → Evaluation**

````python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

X = [[111,8.8],[78,7.4],[65,5.8],[100,8.0],[90,7.8],[70,6.0],[105,8.5],[60,5.5]]
y = [1,1,0,1,1,0,1,0]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

model = LogisticRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)

print("Predictions:", predictions)
print("Accuracy:", accuracy_score(y_test, predictions))
````

### ⭐ Central lesson

The main distinction is the **learning signal available to the model**:

- **Supervised:** correct outputs are available.
- **Unsupervised:** target labels are not provided.
- **Semi-supervised:** only some examples have labels.
- **Reinforcement:** learning comes from interaction and reward feedback.

# 16. 📌 Exam-Ready Definitions

### Supervised Learning
A machine learning approach where a model learns from labeled input-output examples to predict outputs for new inputs.

### Regression
A supervised learning task where the output is numerical or continuous.

### Classification
A supervised learning task where the output belongs to one of several categories.

### Unsupervised Learning
A machine learning approach where the model discovers patterns or structure without predefined target labels.

### Clustering
An unsupervised technique that groups similar data points into clusters.

### Dimensionality Reduction
A technique that reduces the number of features while attempting to preserve important information.

### Anomaly Detection
A technique for identifying observations that significantly differ from normal patterns.

### Association Rule Learning
A technique for discovering useful co-occurrence relationships among items/events.

### Semi-Supervised Learning
A learning approach that combines a small amount of labeled data with a larger amount of unlabeled data.

### Reinforcement Learning
A learning approach where an agent learns through interaction with an environment and feedback in the form of rewards or penalties.
