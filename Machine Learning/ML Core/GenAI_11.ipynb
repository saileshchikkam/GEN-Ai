<div class="v11-flow">Raw / Dirty Data

        ↓

<p>The second approach is to <b>fill missing values</b> rather than remove the records.</p>

```python
# Fill missing values with a chosen value
df.fillna(value)
```

<span class="v11-tag">TOPIC 15</span>
<h2 class="v11-topic">⚙️ Manual vs Automated Data Cleaning</h2>
<table class="v11-table">
<tr><th>Manual</th><th>Automated</th></tr>
<tr>
<td>Cleaning is performed manually, such as identifying and handling missing values.</td>
<td>Cleaning steps can be implemented programmatically so the same process can be repeated.</td>
</tr>
<tr>
<td>More repetitive when the same operation is required many times.</td>
<td>Useful for repeated datasets / repeated cleaning operations.</td>
</tr>
</table>

<div class="v11-flow">Manual
Dataset → Human checks → Handle missing values

Automated
Dataset → Cleaning logic / automation → Handle missing values</div>

<div class="v11-key">
<b>Main takeaway:</b> Automation can reduce repeated manual work when the cleaning rules are known.
