## 2. Machine Learning — the basic idea

The notes describe Machine Learning (ML) as a process in which an algorithm learns a **pattern or relationship** from data.

Instead of manually writing every rule, we provide examples and allow the model to learn a relationship between **inputs** and **outputs**.

### Core idea

```text
Data → Algorithm → Learned pattern/model → Prediction
```

For example, if we provide:

| Input `x` | Output `y` |
|---:|---:|
| 2 | 4 |
| 3 | 9 |
| 4 | 16 |

the model may discover the relationship:

\[
y = x^2
\]

Then, for a new input such as `5`, the learned relationship can be used to predict `25`.

### Important terminology

- **Input / feature:** information given to the model.
- **Output / target:** the value we want the model to predict.
- **Training data:** examples used to learn.
- **Pattern / relationship:** the relationship discovered from the examples.
- **Model:** the learned mathematical representation used for prediction.


````python
# A tiny example of the pattern written in the notes: y = x²

def predict(x):
    return x ** 2

inputs = [2, 3, 4, 5]
outputs = [predict(x) for x in inputs]

print(outputs)
````

## 3. Algorithm and Data

The notes separate two important ingredients:

### Algorithm

The **algorithm** is the procedure used to learn or find a useful relationship from data.

### Data

The **data** provides the examples from which the algorithm can learn.

The handwritten note summarizes this as:

```text
Algorithm → figures out the pattern
Data      → provides the examples needed to discover the pattern
```

In a real ML system, the algorithm is not simply “looking at” the data. It optimizes model parameters according to an objective or loss function so that the model captures useful relationships in the training examples.


## 4. How an ML algorithm learns a pattern

The notes say that an ML algorithm first receives input and output data. Once it learns a useful pattern, we can give it new data and check whether it produces an accurate output.

### Learning process

```text
Training examples
(input + known output)
          ↓
       ML algorithm
          ↓
    Learn relationship
          ↓
     Trained model
          ↓
       New input
          ↓
      Prediction
          ↓
Compare prediction with actual result
```

The key question is:

> **Can the learned model make accurate predictions on data it has not seen before?**

That last point is important because a model that only memorizes training data is not necessarily useful. We generally care about **generalization** to unseen data.

## 7. Traditional Programming vs Machine Learning

A major idea in the notes is the difference between **traditional rule-based programming** and **Machine Learning**.

### Traditional programming

A programmer explicitly writes rules:

```text
Rules + Input → Output
```

For example:

```text
IF condition A is true:
    perform action A
ELSE:
    perform action B
```

### Machine Learning

Instead of manually specifying all rules, we provide examples:

```text
Input + Expected Output → Learning algorithm → Learned model
```

Then:

```text
New Input → Learned model → Prediction
```

## 10. Machine Learning as statistical learning

The notes describe ML as being connected with **statistical operations/tools** and mathematical operations that map inputs to outputs.

A useful formulation is:

\[
\hat{y}=f(x;	heta)
\]

where:

- \(x\) = input/features
- \(y\) = actual target
- \(\hat{y}\) = model prediction
- \(f\) = model/function
- \(	heta\) = parameters learned from training data

Training tries to find parameters that make predictions useful, usually by minimizing a **loss function**.

Examples of ML tasks from the notes:

- House-price prediction
- Email spam classification
- Cat/dog classification
- Placement prediction

### Text must be represented numerically

Machine-learning models generally operate on numerical representations. Therefore, text is transformed into numerical representations such as:

```text
Text
  ↓
Tokens / linguistic units
  ↓
Numerical representation
  ↓
ML / DL model
  ↓
Prediction or generated output
```

Modern NLP commonly uses **embeddings**, where words, sentences, documents, or other pieces of text are represented as vectors.


````python
# A very small illustration of turning text into numbers.
# This is only a conceptual example, not a production NLP tokenizer.

text = "machine learning"
tokens = text.split()

token_ids = {token: i for i, token in enumerate(sorted(set(tokens)))}
encoded = [token_ids[token] for token in tokens]

print("Tokens:", tokens)
print("Token IDs:", encoded)
````

## 17. Generative Adversarial Networks (GANs)

The notes introduce **GAN** as:

> **Generative Adversarial Network**

GANs are a class of deep-learning generative models consisting of two neural networks trained in competition:

1. **Generator** — creates synthetic samples.
2. **Discriminator** — tries to distinguish real samples from generated samples.

Conceptually:

```text
Random input
     ↓
 Generator
     ↓
Generated sample ───────┐
                         ↓
                    Discriminator
                         ↑
Real sample ─────────────┘
```

Through adversarial training, the generator learns to produce increasingly realistic samples.

GANs can be used for tasks such as image generation and image-to-image translation.

The handwritten notes connect GANs with the broader idea of **Generative AI**, which generates new content such as images, text, or audio.

## 19. Types of Machine Learning

The final pages introduce the question:

> **What are the different training processes / types of Machine Learning?**

The standard three major paradigms are:

### 1. Supervised Learning

The model learns from **labelled examples**.

```text
Input + Correct output
        ↓
     Training
        ↓
   Learned model
        ↓
New input → Prediction
```

Examples:

- Spam vs. not spam
- House-price prediction
- Cat vs. dog classification
- Placement prediction

### 2. Unsupervised Learning

The training data does not provide explicit target labels. The algorithm attempts to discover structure such as clusters or lower-dimensional representations.

Examples:

- Customer segmentation
- Clustering
- Dimensionality reduction

### 3. Reinforcement Learning

An agent learns by interacting with an environment and receiving **rewards or penalties**.

```text
Agent → Action → Environment
  ↑                 ↓
  └──── Reward ─────┘
```

Examples:

- Game-playing agents
- Robotics
- Sequential decision-making

## 21. Placement Prediction — supervised learning example

The notes give a small student dataset:

| IQ | CGPA | Placement |
|---:|---:|---:|
| 111 | 8.8 | 1 |
| 89 | 7.4 | 1 |
| 65 | 5.6 | 0 |

Here:

- `IQ` and `CGPA` are **features**.
- `Placement` is the **label/target**.
- `1` can represent placed.
- `0` can represent not placed.

A trained classifier learns a decision boundary or probability relationship from many labelled examples.

The handwritten note also indicates that a much larger dataset could be used for training. The exact row counts written in the notebook are not fully clear, so they are not treated as fixed requirements here.


````python
# Tiny supervised-learning example based on the placement table.
# This demonstrates the workflow; it is not a meaningful real placement model.

from sklearn.tree import DecisionTreeClassifier

X = [
    [111, 8.8],
    [89, 7.4],
    [65, 5.6],
]

y = [1, 1, 0]

model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)

new_student = [[95, 7.8]]
prediction = model.predict(new_student)

print("Predicted placement class:", prediction[0])
````
