# 1. 🌐 Data Gathering Using APIs

APIs can be used to gather data from external services in a controlled and structured way.

Instead of directly accessing a server's internal database, an application communicates through an API.

### Basic idea

```text
My Computer
     │
     │ Request
     ↓
   API / Server
     │
     ↓
    Data
```

### ⭐ Key point

> **APIs provide a controlled interface for applications to request and receive data.**

# 2. 🖥️ Client Computer and Database Server

Consider a simple architecture:

```text
┌──────────────────┐
│  My Computer     │
│  (Client)        │
└────────┬─────────┘
         │
       Request
         ↓
┌──────────────────┐
│ Database /       │
│ Server           │
└────────┬─────────┘
         │
         ↓
       Data
```

The client sends a request to a server.

However, applications generally should not receive unrestricted access to the server's internal database.

Instead, controlled interfaces such as APIs are used.

# 3. 🔐 Why Direct Database Access Is Restricted

Direct access to a database can create security risks.

For example, a malicious user or attacker could try to access information that they are not authorized to see.

### Simplified problem

```text
Client
  │
  │ Direct database access ❌
  ↓
Database
```

If unrestricted access is provided, sensitive information may be exposed.

### Controlled approach

```text
Client
  │
  ↓
API
  │
  ↓
Database / Backend
```

The API can enforce permissions and expose only the information or operations that are allowed.

### ⭐ Important

> **The API acts as a controlled gateway between the client and backend/database.**

# 4. 🛠️ Tools / APIs for Controlled Data Gathering

The notes highlight the use of tools and APIs to gather data from services.

A general architecture is:

```text
Application
      │
      ↓
     API
      │
      ↓
External Service
      │
      ↓
    Data
```

The application does not need to know the internal implementation of the service.

It only needs to follow the API's defined interface.

# 5. 🌍 Internet Access vs API Access

An important distinction is that **being able to access a website through a browser does not automatically mean that a program can freely access all of its data**.

### Browser

```text
User
 ↓
Browser
 ↓
Website
 ↓
Web page
```

### Program

```text
Program
 ↓
HTTP/API Request
 ↓
Server
 ↓
Response
```

The server may treat browser requests and automated/programmatic requests differently.

Access can depend on authentication, permissions, headers, cookies, rate limits and other server-side controls.

# 6. 🌐 Browser vs Programmatic Access

### Browser-based access

A browser automatically handles many details of a web request.

For example:

```text
Browser
 ├── URL
 ├── Headers
 ├── Cookies
 ├── Session information
 └── Other request details
```

### Programmatic access

When using Python or another programming language, these request details may need to be supplied explicitly when required.

```text
Python Program
      │
      ├── URL
      ├── Headers
      ├── Authentication
      └── Cookies (when applicable)
              │
              ↓
             Server
```

### ⭐ Key idea

> **A browser may automatically provide information that a programmatic request needs to handle explicitly.**

# 7. 🍪 Cookies and Headers

Two important concepts when working with web requests are:

## Cookies

Cookies can store information associated with a user's browser/session.

Examples include:

```text
Session information
Login/session identifiers
User preferences
```

## Headers

HTTP headers provide additional information about a request or response.

Examples include:

```text
Authorization
Content-Type
User-Agent
Accept
```

### Simplified request

```text
URL
 +
Headers
 +
Cookies (if required)
       ↓
    Server
```

### ⚠️ Security note

Do not expose private authentication tokens, session cookies or API keys in public notebooks or repositories.

# 8. 🔗 URL + Headers in API Requests

When a program makes a request, the URL identifies the resource or endpoint.

Additional request information can be sent through headers.

### Conceptual example

```python
url = "https://example.com/api/data"

headers = {
    "Authorization": "Bearer <TOKEN>",
    "Accept": "application/json"
}
```

Then the application can make an HTTP request using an appropriate HTTP client.

### General structure

```text
Program
   │
   ├── URL
   ├── Headers
   └── Request data
         │
         ↓
       API
         │
         ↓
      Response
```

> **The exact headers and authentication method depend on the API being used.**

# 12. 🆚 Browser Request vs API Request

| Feature | Browser | Program/API Request |
|---|---|---|
| User interface | Usually available | Usually not required |
| URL | Used | Used |
| Headers | Often handled automatically | May need to be specified |
| Cookies | Browser manages them | Program may need to manage them |
| Authentication | Often handled through browser session | Usually implemented explicitly |
| Data handling | Page displayed to user | Data processed by program |
| Automation | Limited/controlled | Designed for programmatic workflows |

### ⭐ Remember

A browser and an API client both communicate with servers, but they may differ in how requests are constructed and how authentication/session information is handled.

# 13. 🔒 Security Considerations

When gathering data through APIs or web requests, always consider:

### 1. Authentication

Use the required authentication method.

```text
API Key
OAuth
Bearer Token
Session Authentication
```

### 2. Authorization

Only access resources that the application/user is permitted to access.

### 3. Rate Limits

APIs may restrict the number of requests that can be made within a period.

### 4. Secrets

Never hard-code private credentials into public code.

❌ Avoid:

```python
API_KEY = "my-real-secret-key"
```

Prefer environment variables or a secure secret-management mechanism.

### 5. Terms and Policies

Follow the API provider's usage rules and website/API policies.

# 15. 🎓 Interview Questions & Answers

### Q1. Why should applications use APIs instead of directly accessing databases?

**Answer:**  
APIs provide a controlled interface through which applications can access permitted data or functionality while protecting the internal database and business logic.

---

### Q2. What is the role of an API in data gathering?

**Answer:**  
An API provides a structured interface through which an application can request and receive data from an external service.

---

### Q3. What is the difference between browser access and programmatic access?

**Answer:**  
A browser automatically manages many request details such as cookies and some headers. In programmatic access, the developer may need to explicitly provide the required URL, headers, authentication and session information.

---

### Q4. What are HTTP headers?

**Answer:**  
HTTP headers are metadata sent with HTTP requests or responses. They can provide information such as content type, authorization and accepted response formats.

---

### Q5. What are cookies?

**Answer:**  
Cookies are pieces of data stored by a browser and sent with applicable requests. They are commonly used for sessions, preferences and authentication-related state.

---

### Q6. Why are headers sometimes required in API requests?

**Answer:**  
Headers can provide required information such as authentication credentials, content type and the expected response format.

---

### Q7. What is Google Colab?

**Answer:**  
Google Colab is a cloud-based notebook environment that allows users to execute Python code in a browser and work interactively with data and machine-learning workflows.

---

### Q8. What is Kaggle used for?

**Answer:**  
Kaggle provides datasets, notebooks and tools that are widely used for data analysis, machine learning and experimentation.

---

### Q9. What security precautions should be followed when using APIs?

**Answer:**  
Protect API keys and tokens, follow authentication and authorization requirements, respect rate limits, validate inputs and outputs, and follow the provider's usage policies.

---

### Q10. Can an API access a database?

**Answer:**  
Yes. An API can communicate with a backend/database and return authorized information to the client without exposing direct database access.
