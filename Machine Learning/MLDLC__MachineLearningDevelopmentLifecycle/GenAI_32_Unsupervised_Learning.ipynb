# 3. Classification Evaluation Metrics

The notes list these evaluation metrics:

- Accuracy
- Precision
- Recall
- F1-score
- Confusion Matrix

## Confusion Matrix

For binary classification:

```text
                         ACTUAL
                   Positive   Negative
                ┌───────────┬───────────┐
Predicted       │           │           │
Positive        │    TP     │    FP     │
                ├───────────┼───────────┤
Predicted       │           │           │
Negative        │    FN     │    TN     │
                └───────────┴───────────┘
```

| Term | Meaning |
|---|---|
| **TP — True Positive** | Positive sample correctly predicted positive |
| **TN — True Negative** | Negative sample correctly predicted negative |
| **FP — False Positive** | Negative sample incorrectly predicted positive |
| **FN — False Negative** | Positive sample incorrectly predicted negative |

### Accuracy

$$
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
$$

### Precision

$$
Precision = \frac{TP}{TP + FP}
$$

**Question answered:** Of everything predicted as positive, how much was actually positive?

### Recall

$$
Recall = \frac{TP}{TP + FN}
$$

**Question answered:** Of all actual positive cases, how many did the model identify?

### F1-score

$$
F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}
$$

F1-score combines precision and recall into a single measure.

# 10. Supervised vs Unsupervised Learning

The notes introduce unsupervised learning by contrasting it with supervised learning.

```text
Input Features ─────► Model ─────► Known Target
```

The model learns a relationship between inputs and the target.

# 18. K-Means Clustering

K-Means is a clustering algorithm that groups observations around **centroids**.

The handwritten notes emphasize:

- Selecting the number of clusters
- Using centroids
- Assigning points to clusters
- Updating cluster centers

## Basic idea

```text
Data Points
    │
    ▼
Choose K
    │
    ▼
Initialize K Centroids
    │
    ▼
Assign Points to Nearest Centroid
    │
    ▼
Recalculate Centroids
    │
    ▼
Repeat
    │
    ▼
Final Clusters
```

# 19. What Is a Centroid?

A **centroid** represents the center of a cluster.

Conceptually:

```text
      ●
   ●  ●  ●
      X
   ●  ●  ●

      X
   Centroid
```

The centroid is recalculated based on the points assigned to its cluster.

# 20. Choosing K in K-Means

The notes raise the important question:

> How many clusters should be created?

In K-Means:

```text
K = number of clusters
```

The number of centroids corresponds to the number of clusters.

```text
K = 2  →  2 clusters  →  2 centroids
K = 3  →  3 clusters  →  3 centroids
K = 4  →  4 clusters  →  4 centroids
```

### Example

```text
K = 3

        Data
         │
    ┌────┼────┐
    ▼    ▼    ▼
  C1     C2    C3
```

# 21. K-Means: Step-by-Step

## Step 1 — Initialize centroids

Choose `K` initial centroid positions.

```text
● ● ●             ● ●
  X                 X
● ●               ● ●
```

`X` represents an initial centroid.

## Step 2 — Assign each point

Each point is assigned to the nearest centroid.

```text
Point
  │
  ├── closer to C1 → Cluster 1
  │
  └── closer to C2 → Cluster 2
```

## Step 3 — Recalculate centroids

After assigning points, calculate the new center of each cluster.

```text
Cluster points
      │
      ▼
Calculate center
      │
      ▼
New centroid
```

## Step 4 — Repeat

Repeat assignment and centroid recalculation until the clusters stabilize.

```text
Initialize
    ↓
Assign
    ↓
Recalculate
    ↓
Assign
    ↓
Recalculate
    ↓
Converged clusters
```

---

# 22. K-Means Mathematical Intuition

K-Means attempts to make points within each cluster close to their corresponding centroid.

Conceptually, it minimizes the total within-cluster squared distance:

$$
\sum_{k=1}^{K}\sum_{x_i \in C_k} ||x_i-\mu_k||^2
$$

where:

- $K$ = number of clusters
- $C_k$ = cluster $k$
- $x_i$ = data point
- $\mu_k$ = centroid of cluster $k$

> The handwritten notes primarily focus on the practical centroid-based procedure; this formula expresses the same idea mathematically.

# 29. Exam & Interview Quick Questions

### Q1. What is classification?

Classification is a supervised-learning task where a model predicts a categorical class from input features.

### Q2. Name common classification algorithms.

Logistic Regression, Decision Tree, Random Forest, XGBoost, AdaBoost, CatBoost, KNN, SVC, Naive Bayes, Bagging, and Stacking.

### Q3. What is hyperparameter tuning?

It is the process of testing different hyperparameter configurations to identify a suitable model configuration.

### Q4. Give two examples of hyperparameters.

`max_depth` in a Decision Tree and `n_estimators` in a Random Forest.

### Q5. What is unsupervised learning?

A type of machine learning in which the model learns patterns or structure from data without provided target labels.

### Q6. What is clustering?

Clustering groups observations according to similarities in their features.

### Q7. Name three clustering algorithms.

K-Means, Hierarchical Clustering, and DBSCAN.

### Q8. What is a centroid?

A centroid is the representative center of a cluster in K-Means.

### Q9. What does K represent in K-Means?

`K` represents the number of clusters to be created.

### Q10. What is the basic K-Means process?

```text
Choose K
  ↓
Initialize centroids
  ↓
Assign points
  ↓
Recalculate centroids
  ↓
Repeat
```

### Q11. Classification vs clustering?

> **Classification predicts predefined classes; clustering discovers groups without predefined labels.**
