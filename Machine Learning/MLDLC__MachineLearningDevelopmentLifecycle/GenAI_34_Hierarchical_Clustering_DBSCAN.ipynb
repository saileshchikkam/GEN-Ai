# 1. Hierarchical Clustering

Hierarchical clustering is an **unsupervised learning** technique that builds a hierarchy of clusters.

There are two main approaches:

```text
                    Hierarchical Clustering
                             │
                ┌────────────┴────────────┐
                ▼                         ▼
        Agglomerative                 Divisive
          Clustering                 Clustering
                │                         │
          Bottom → Top               Top → Bottom
```

| Method | Direction | Basic idea |
|---|---|---|
| **Agglomerative** | Bottom → Top | Start with individual points and repeatedly merge clusters |
| **Divisive** | Top → Bottom | Start with one large cluster and repeatedly split it |

---

# 2. Agglomerative Hierarchical Clustering

Agglomerative clustering is a **bottom-up** approach.

### Core idea

> Start by treating every data point as a separate cluster and repeatedly combine the closest clusters.

```text
Individual points
      │
      ▼
Merge closest clusters
      │
      ▼
Larger clusters
      │
      ▼
Continue merging
      │
      ▼
One large cluster
```

---

# 3. Initial State of Agglomerative Clustering

Suppose there are several data points:

```text
P1   P2   P3   P4   P5   P6
```

Initially, every point is treated as its own cluster:

```text
{P1} {P2} {P3} {P4} {P5} {P6}
```

Therefore:

```text
Number of data points = Number of initial clusters
```

### Example

For 10 data points:

```text
10 data points
      ↓
10 initial clusters
```

---

# 4. Step 1 — Calculate Distances

The notes describe calculating the distance between data points using **Euclidean distance**.

For two points:

```text
P1 = (x1, y1)
P2 = (x2, y2)
```

Euclidean distance is:

$$
d(P_1,P_2)
=
\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
$$

### Visual intuition

```text
P1 ●
   │\
   │ \
   │  \  distance
   │   \
   │    \
   └─────● P2
```

A smaller distance means the points are more similar/closer in feature space.

---

# 5. Step 2 — Find the Nearest Clusters

After calculating distances, identify the closest two clusters.

```text
Cluster A ● ●

             ← smallest distance →

                    ● ● Cluster B
```

These two clusters are merged into a single cluster.

```text
Before:

{A}     {B}     {C}     {D}

          ↓ merge closest

After:

{A,B}          {C}          {D}
```

---

# 6. Step 3 — Continue Merging

The same process continues repeatedly.

```text
Start:

{P1} {P2} {P3} {P4} {P5} {P6}

          ↓

{P1,P2} {P3} {P4,P5} {P6}

          ↓

{P1,P2} {P3,P4,P5} {P6}

          ↓

{P1,P2,P3,P4,P5} {P6}

          ↓

{P1,P2,P3,P4,P5,P6}
```

The important idea is:

> **Agglomerative clustering keeps joining the nearest clusters until all points belong to one hierarchy.**

---

# 7. Agglomerative Clustering — Full Flow

```text
             Data points
                  │
                  ▼
      Each point = separate cluster
                  │
                  ▼
        Calculate distances
                  │
                  ▼
       Find nearest clusters
                  │
                  ▼
          Merge the pair
                  │
                  ▼
       Recalculate distances
                  │
                  ▼
            Repeat
                  │
                  ▼
        One large hierarchy
```

---

# 8. Why Don't We Simply Keep One Cluster?

The notes raise an important question:

> If the process eventually creates one large cluster, what number of clusters should we actually use?

The answer is:

> We choose an **optimal number of clusters** from the hierarchy instead of simply using the final single cluster.

A **dendrogram** helps us make this decision.

---

# 9. Dendrogram

A dendrogram is a tree-like visualization of hierarchical clustering.

It shows:

- which points/clusters were merged,
- the order of merging,
- and the distance at which the merging happened.

```text
Distance
   ↑
   │                 ┌───────────────┐
   │          ┌──────┤               │
   │     ┌────┤      │               │
   │     │    │      │               │
   │  ┌──┤    │      │               │
   │  │  │    │      │               │
   └──┴──┴────┴──────┴────────────────→ Data points
     P1 P2   P3 P4   P5 P6
```

The **vertical height** at which two branches join represents the distance at which those clusters were merged.

---

# 10. Reading a Dendrogram

A dendrogram can be interpreted from bottom to top:

```text
Individual points
       │
       ▼
Small clusters
       │
       ▼
Larger clusters
       │
       ▼
One large cluster
```

The lower a merge occurs, the closer the corresponding points/clusters are.

A higher merge indicates that the groups being joined are farther apart.

---

# 11. Dendrogram and Cluster Selection

One of the main purposes of a dendrogram is to determine:

> **How many clusters should we create?**

We introduce a horizontal threshold/cut line.

```text
Distance
   ↑
  6│                 ┌───────────────┐
   │                 │               │
  5│─────────────── CUT / THRESHOLD ─┤
   │          ┌──────┤               │
  3│     ┌────┤      │               │
   │  ┌──┤    │      │               │
  1│  │  │    │      │               │
   └──┴──┴────┴──────┴────────────────→
     P1 P2   P3 P4   P5 P6
```

The number of branches intersected by the horizontal threshold gives the selected number of clusters.

---

# 12. Threshold Concept

The handwritten notes describe searching for an appropriate **threshold value**, based on the distance shown in the dendrogram.

```text
Dendrogram
     │
     ▼
Choose threshold
     │
     ▼
Draw horizontal line
     │
     ▼
Count intersections
     │
     ▼
Number of clusters
```

### Memory rule

> **Horizontal cut → Count vertical branches → Number of clusters**

---

# 13. How to Choose a Useful Threshold

Suppose the dendrogram has several merge levels:

```text
Distance
  ↑
  │             ┌───────────┐
  │             │           │
  │─────────────┼───────────┤  ← possible threshold
  │       ┌─────┤           │
  │   ┌───┤     │           │
  │ ┌─┤   │     │           │
  └─┴─┴───┴─────┴──────────────→
```

A useful threshold is generally chosen where a horizontal cut creates meaningful groups while avoiding unnecessary merging.

In the handwritten notes, this is described as finding an **optimal threshold** from the dendrogram.

# 14. Agglomerative Clustering Example

Suppose:

```text
P1, P2, P3, P4, P5, P6
```

Initially:

```text
{P1} {P2} {P3} {P4} {P5} {P6}
```

Assume the closest pair is:

```text
P1 ↔ P2
```

Merge:

```text
{P1,P2} {P3} {P4} {P5} {P6}
```

Then another closest pair may be:

```text
P4 ↔ P5
```

Merge:

```text
{P1,P2} {P3} {P4,P5} {P6}
```

The process continues until the hierarchy is complete.

# 15. Divisive Hierarchical Clustering

Divisive clustering is the **reverse of agglomerative clustering**.

It follows a **top-down** strategy.

### Core idea

> Start with all data points in one cluster and repeatedly split the cluster into smaller groups.

```text
                 All data
                    │
                    ▼
              One cluster
                    │
               Split
              ┌─────┴─────┐
              ▼           ▼
          Cluster A    Cluster B
              │           │
            Split        Split
             │             │
          Smaller       Smaller
          clusters      clusters
```

---

# 16. Agglomerative vs Divisive

| Feature | Agglomerative | Divisive |
|---|---|---|
| Direction | Bottom → Top | Top → Bottom |
| Starting point | Each data point is its own cluster | All points are one cluster |
| Main operation | Merge | Split |
| Initial clusters | `n` for `n` points | `1` |
| Final hierarchy | One large cluster | Individual/smaller clusters |
| Concept | Build upward | Break downward |

### Easy memory trick

```text
AGGLOMERATIVE
Aggregate → Join → Merge
       ↑
 Bottom → Top

DIVISIVE
Divide → Split
       ↓
 Top → Bottom
```

---

# 17. Why Hierarchical Clustering Is Useful

Hierarchical clustering is useful when we want to understand the **nested structure** of groups.

```text
                 All data
                    │
             ┌──────┴──────┐
             │             │
          Group A        Group B
          /    \         /    \
        A1      A2      B1     B2
```

This gives more information than simply returning one fixed grouping.

# 18. DBSCAN

DBSCAN stands for:

> **Density-Based Spatial Clustering of Applications with Noise**

The notes describe it as a:

> **Density-based spatial clustering method**

Its main idea is to identify groups of points based on **density**.

```text
Dense region              Dense region

 ● ● ●                     ● ●
● ● ● ●                   ● ● ●
 ● ● ●                     ● ●

          sparse/noise

             ×
```

---

# 19. DBSCAN vs K-Means

K-Means is centroid-based.

DBSCAN is density-based.

| Feature | K-Means | DBSCAN |
|---|---|---|
| Main idea | Centroid-based grouping | Density-based grouping |
| Needs `K` beforehand | Yes | No fixed `K` required |
| Cluster shape | Often suited to compact/centroid-like groups | Can discover irregular shapes |
| Noise/outliers | Not naturally handled as noise | Can explicitly identify noise |
| Main parameters | `K` | Density parameters such as `eps` and `min_samples` |

> The last two columns go slightly beyond the exact handwritten wording, but clarify the density-based distinction present in the notes.

# 21. DBSCAN — Radius Concept

DBSCAN uses a neighborhood around a point.

Conceptually:

```text
                 ●
             ●   │   ●
               \ │ /
                \│/
          ●──────P──────●
                /|\
               / │ \
             ●   │   ●

       ←──── neighborhood ────→
```

The neighborhood is determined using a distance/radius concept.

The notes refer to this distance as the **radius** and identify density around points as the key idea.

# 23. Core Point

A **core point** is a point located in a sufficiently dense region.

Conceptually:

```text
      ● ● ●
    ● ● P ● ●
      ● ● ●

        P
     Core point
```

The exact classification depends on the density parameters.

---

# 24. Border Point

A border point is near a dense cluster but does not itself have enough neighboring points to satisfy the core-point condition.

```text
       Dense cluster

       ● ● ●
      ● ● ● ●
       ● P ●

             ○
             │
        Border point
```

The border point can belong to the nearby cluster even though it is not itself a core point.

---

# 25. Noise / Outlier

A point that does not belong to any sufficiently dense region can be treated as noise.

```text
Dense cluster

 ● ● ●
● ● ● ●
 ● ● ●


                         ×
                       Noise
```

This is one of the major conceptual differences between DBSCAN and centroid-based clustering.

---

# 26. DBSCAN Parameters

Two important parameters are:

| Parameter | Meaning |
|---|---|
| **`eps`** | Radius/neighborhood distance used to search for nearby points |
| **`min_samples`** | Minimum number of nearby points needed to consider a region sufficiently dense |

Conceptually:

```text
DBSCAN
  │
  ├── eps
  │     └── How far do we look?
  │
  └── min_samples
        └── How many nearby points are enough?
```

---

# 27. DBSCAN Classification Logic

A simplified conceptual flow:

```text
                Start with point P
                       │
                       ▼
             Find neighbors within eps
                       │
                       ▼
        Are there enough nearby points?
                 /             \
               Yes              No
                │                │
                ▼                ▼
          Core point        Border / Noise
                │
                ▼
        Expand the cluster
```

The precise classification of a non-core point depends on whether it is reachable from a core point.

---

# 28. DBSCAN Cluster Expansion

Once a core point is found:

```text
Core point
    │
    ▼
Find nearby points
    │
    ▼
Add reachable points
    │
    ▼
If another core point appears
    │
    ▼
Expand again
```

Conceptually:

```text
       ● ●
     ● ● ● ●
    ● ● P ● ●
      ● ● ●
          \
           \ expand
            \
             ○
```

The cluster can grow through connected dense regions.

---

# 29. DBSCAN Can Find Irregularly Shaped Clusters

Because the grouping is based on density rather than distance from a single centroid, DBSCAN can represent cluster structures that are not naturally circular/centroid-shaped.

```text
      ● ● ●
    ● ●
  ● ●
 ● ●
  ● ● ● ●
       ● ●
```

The important concept is:

> **Connected dense regions can form a cluster.**

# 36. Interview Questions

### 1. What is hierarchical clustering?

It is an unsupervised clustering approach that creates a hierarchy of groups through repeated merging or splitting.

### 2. What are the two types of hierarchical clustering?

- Agglomerative
- Divisive

### 3. What is agglomerative clustering?

A bottom-up method that starts with each point as an individual cluster and repeatedly merges the closest clusters.

### 4. What is divisive clustering?

A top-down method that starts with all data points in one cluster and repeatedly splits the cluster.

### 5. What is a dendrogram?

A tree-like diagram that represents the hierarchy of cluster merges and their associated distances.

### 6. How do you select clusters from a dendrogram?

Draw a horizontal threshold/cut and count the branches intersected by the cut.

### 7. What is DBSCAN?

DBSCAN is a density-based unsupervised clustering algorithm that identifies dense groups and can distinguish noise/outliers.

### 8. What does `eps` mean in DBSCAN?

It represents the neighborhood radius used to search for nearby points.

### 9. What is `min_samples`?

It represents the minimum number of nearby points used to determine whether a region is sufficiently dense.

### 10. What is a core point?

A point with enough neighboring points within the specified neighborhood.

### 11. What is a border point?

A point that is associated with a dense cluster but does not itself satisfy the core-point density requirement.

### 12. What is noise?

A point that does not belong to any sufficiently dense cluster.
