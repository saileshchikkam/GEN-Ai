# 1. K-Means Clustering — Core Idea

K-Means is an **unsupervised learning** algorithm used to divide data points into groups called **clusters**.

The main idea is:

> Points that are close/similar to one another are grouped into the same cluster.

```text
                    Feature 2
                       ↑
                       │      ● ● ●
                       │    ●   X   ●
                       │
                       │
                       │                 ● ●
                       │               ●  X  ●
                       │
                       └────────────────────────→ Feature 1

                              X = Centroid
```

A cluster therefore contains points that show a similar pattern in the selected features.

---

# 2. Important Terms

| Term | Meaning |
|---|---|
| **Data point** | One observation/customer/sample in the dataset |
| **Cluster** | A group of similar data points |
| **Centroid** | The representative center of a cluster |
| **K** | Number of clusters to create |
| **Iteration** | One repeated cycle of assigning points and updating centroids |
| **Distance** | A measure used to determine which centroid is closest to a point |

# 7. Step 1 — Initialize Centroids

First, choose the number of clusters **K** and initialize the centroids.

For example:

```text
K = 3

Data points:

● ● ●
  ●

             ● ●
              ●

        ● ● ●

Initial centroids:

      X₁          X₂          X₃
```

The notes describe the initial centroids as starting positions that will later be moved according to the data.

# 8. Step 2 — Assign Points to the Nearest Centroid

Each data point is compared with the centroids.

The point is assigned to the centroid it is closest to.

```text
             ● ●
            ● X₁ ●
             ●

                    ●
                   X₂
                  ● ● ●
```

Conceptually:

```text
Data point
    │
    ├── closer to C1 → Cluster 1
    │
    ├── closer to C2 → Cluster 2
    │
    └── closer to C3 → Cluster 3
```

### Result

The points become grouped around their nearest centroid.

# 9. Step 2 — Grouping Based on Distance

The key idea from the notes is:

> Points nearest to a centroid are grouped together.

```text
             Cluster 1
          ● ● ●
        ●   X₁  ●


                         Cluster 2
                       ● ● ●
                      ● X₂ ●


          Cluster 3
       ● ● ●
         X₃
```

Here:

- `X₁` = centroid of Cluster 1
- `X₂` = centroid of Cluster 2
- `X₃` = centroid of Cluster 3

# 10. Step 3 — Recalculate the Centroid

After points are assigned to clusters, the centroid location is changed.

The notes describe this as:

> **Change the location of centroids by calculating the mean of the data points.**

For a cluster, the new centroid is calculated from the mean of its points.

### 2D centroid intuition

For points:

```text
P₁ = (x₁, y₁)
P₂ = (x₂, y₂)
P₃ = (x₃, y₃)
```

the new centroid is:

$$
C = \left(
\frac{x_1+x_2+x_3}{3},
\frac{y_1+y_2+y_3}{3}
\right)
$$

More generally:

$$
C =
\left(
\frac{\sum x_i}{n},
\frac{\sum y_i}{n}
\right)
$$

where `n` is the number of points in that cluster.

# 11. Why Does the Centroid Move?

Initially, a centroid may not be positioned at the best location.

After points are assigned, we calculate their mean.

```text
Initial centroid
       X
       │
       │ moves toward
       ▼
Mean position of
cluster points
```

Conceptually:

```text
Before:

● ● ●
   X
 ● ●

After calculating the mean:

● ● ●
  X'
● ●
```

`X'` represents the updated centroid.

# 12. Centroid Movement

The handwritten notes explain that the initial centroid may need to shift based on the arrangement/density of data points.

```text
Initial position
       X
       │
       │  centroid moves
       ▼
      X'
       │
       ▼
Mean position of
assigned points
```

This movement is repeated until the cluster assignments become stable.

# 13. Step 4 — Reassign Data Points

After the centroids move, the distances to the centroids change.

Therefore, we again check the data points.

```text
Updated centroids
       │
       ▼
Calculate nearest centroid
       │
       ▼
Reassign points
       │
       ▼
New clusters
```

The handwritten notes specifically mention:

> For new data points, check the nearest centroid.

# 14. What Happens When a New Data Point Arrives?

Suppose the model has already learned the cluster centroids.

A new point arrives:

```text
Existing clusters
       │
       ├── C1
       ├── C2
       └── C3
             │
             ▼
       New data point P
             │
             ▼
   Find nearest centroid
             │
             ▼
     Assign P to that cluster
```

### Example

```text
              ● ●
             ● C1 ●

                     P ●
                       \
                        \ closest
                         \
                          C2 ● ●
```

The new point is assigned to the cluster whose centroid is closest.

# 15. Step 5 — Repeat the Process

The handwritten notes emphasize repeating the previous steps.

The overall cycle is:

```text
        ┌──────────────────────────┐
        │                          ▼
Initialize Centroids → Assign Points
        ▲                  │
        │                  ▼
        └──── Update Centroids
```

Expanded:

```text
1. Initialize
       ↓
2. Assign points
       ↓
3. Calculate mean
       ↓
4. Move/update centroids
       ↓
5. Reassign points
       ↓
6. Check changes
       ↓
   Repeat if necessary
```

---

# 16. When Does K-Means Stop?

The notes describe stopping when the clustering/centroid arrangement no longer changes.

```text
New iteration
     │
     ▼
Compare with previous result
     │
 ┌───┴────┐
 │        │
Changed   Same
 │        │
 ▼        ▼
Repeat   Stop
```

### Intuition

If:

- cluster assignments remain the same, and/or
- centroid positions have effectively stabilized,

there is no useful reason to continue iterating.

# 18. K-Means: One Complete Example

Suppose:

```text
K = 2
```

We want two customer groups.

### Initial state

```text
● ● ●                 ● ●
  X₁                   X₂
● ●                   ● ●
```

### Assign points

Each point is assigned to its closest centroid.

```text
Cluster 1              Cluster 2

● ● ●                  ● ●
  X₁                    X₂
● ●                    ● ●
```

### Calculate means

The average location of the points in each cluster is calculated.

```text
Cluster 1 → new centroid C₁'
Cluster 2 → new centroid C₂'
```

### Reassign

```text
Updated centroids
       │
       ▼
Check every point again
       │
       ▼
Move points if necessary
```

### Repeat

```text
Assignment
    ↓
Mean
    ↓
Centroid movement
    ↓
Reassignment
    ↓
Stable result
```

# 19. Why Reassignment Is Necessary

Moving the centroids can change which centroid is closest to a data point.

Therefore:

```text
Old centroid positions
        │
        ▼
Points assigned
        │
        ▼
Centroids move
        │
        ▼
Distances change
        │
        ▼
Some points may change clusters
```

This is why K-Means is iterative.

# 24. K-Means vs Classification

Although both can produce groups/classes, they solve different problems.

| Aspect | Classification | K-Means Clustering |
|---|---|---|
| Learning | Supervised | Unsupervised |
| Target label | Available | Not provided |
| Main objective | Predict a known class | Discover groups |
| Example | Spam vs not spam | Customer segments |
| Training output | Predicted class | Cluster assignment |
| Main idea | Learn from labelled examples | Find similarity structure |

### Memory trick

> **Classification = predefined classes.**  
> **Clustering = discovered groups.**

# 25. Distance-Based Assignment

The handwritten notes repeatedly emphasize assigning points according to the nearest centroid.

Conceptually:

```text
                 P
                / \
               /   \
              /     \
             ▼       ▼
            C1       C2

If distance(P, C1) < distance(P, C2)

        P → Cluster 1
```

In a typical K-Means implementation, distance is commonly based on Euclidean distance, although the handwritten pages themselves mainly emphasize the nearest-centroid concept.

# 26. K-Means Mathematical View

For each cluster, calculate the mean position of the points assigned to it.

For one-dimensional data:

$$
\mu = \frac{x_1+x_2+\cdots+x_n}{n}
$$

For two-dimensional data:

$$
\mu_x = \frac{1}{n}\sum_{i=1}^{n}x_i
$$

$$
\mu_y = \frac{1}{n}\sum_{i=1}^{n}y_i
$$

So:

$$
C=(\mu_x,\mu_y)
$$

This explains why the notes repeatedly refer to **calculating the mean** to update the centroid.

# 31. Interview Questions

### 1. What is K-Means?

K-Means is an unsupervised learning algorithm that groups data points into `K` clusters based on similarity/proximity.

### 2. What does K represent?

`K` represents the number of clusters to create.

### 3. What is a centroid?

A centroid is the representative center of a cluster.

### 4. How is a centroid updated?

It is updated using the mean of the data points assigned to that cluster.

### 5. Why do we repeatedly assign points?

Because when centroids move, the nearest centroid for some points can change.

### 6. When does K-Means stop?

When the cluster assignments/centroids become stable according to the stopping condition.

### 7. Give a real-world application.

Customer segmentation based on features such as salary and spending score.

### 8. What happens when a new data point arrives?

It can be assigned to the nearest existing centroid.

### 9. What are the clustering algorithms mentioned in the notes?

- K-Means
- Hierarchical Clustering
- DBSCAN

### 10. Classification vs clustering?

> Classification learns from labelled data to predict predefined classes, whereas clustering works without target labels to discover groups.
