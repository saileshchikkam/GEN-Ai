## 2. Bayes' Theorem

The handwritten notes use Bayes' theorem as the mathematical foundation.

$$
P(A|B)=\frac{P(B|A)\,P(A)}{P(B)}
$$

Where:

| Term | Meaning |
|---|---|
| $P(A\mid B)$ | Probability of `A` given `B` |
| $P(B\mid A)$ | Probability of `B` given `A` |
| $P(A)$ | Prior probability of `A` |
| $P(B)$ | Evidence / marginal probability of `B` |

### Interpretation

Bayes' theorem allows us to update the probability of a hypothesis after observing evidence.

```text
Prior belief
     │
     │ + observed evidence
     ↓
Bayes' theorem
     │
     ↓
Updated probability
```


## 3. Conditional Probability

Conditional probability describes the probability of one event when another event is already known.

$$
P(A|B)=\frac{P(A\cap B)}{P(B)}
$$

The notes use probability relationships such as:

$$
P(A\cap B)=P(A|B)P(B)
$$

and equivalently:

$$
P(A\cap B)=P(B|A)P(A)
$$

Therefore:

$$
P(A|B)P(B)=P(B|A)P(A)
$$

This relationship leads directly to Bayes' theorem.

# 5. Naive Bayes Assumption — Feature Independence

The important idea in the notes is that Naive Bayes treats features as conditionally independent.

Suppose we have:

```text
Feature 1 ──┐
Feature 2 ──┼──→ Class
Feature 3 ──┘
```

The Naive Bayes assumption can be written as:

$$
P(x_1,x_2,\ldots,x_n|C)
=
\prod_{i=1}^{n}P(x_i|C)
$$

Therefore:

$$
P(C|x_1,x_2,\ldots,x_n)
\propto
P(C)\prod_{i=1}^{n}P(x_i|C)
$$

### Why "Naive"?

The independence assumption is often unrealistic in real-world data, but it makes the probability calculation much simpler and can still produce very useful classifiers.


## 6. Naive Bayes Classification

For an input vector:

$$
X=(x_1,x_2,\ldots,x_n)
$$

the model evaluates each possible class.

$$
P(C_k|X)
\propto
P(C_k)
\prod_{i=1}^{n}P(x_i|C_k)
$$

The predicted class is the class with the highest posterior probability:

$$
\hat C
=
\arg\max_{C_k}
P(C_k)
\prod_{i=1}^{n}P(x_i|C_k)
$$

### Classification pipeline

```text
Training data
     ↓
Estimate class probabilities
     ↓
Estimate feature probabilities
     ↓
Receive new sample
     ↓
Calculate posterior for each class
     ↓
Choose highest probability
     ↓
Predicted class
```

# 10. Gaussian Naive Bayes

Gaussian Naive Bayes is appropriate for **continuous numerical features**.

The handwritten notes connect it with a Gaussian / normal distribution.

For a feature $x$ in class $C$:

$$
P(x|C)
=
\frac{1}{\sqrt{2\pi\sigma_C^2}}
\exp
\left(
-\frac{(x-\mu_C)^2}{2\sigma_C^2}
\right)
$$

where:

- $\mu_C$ = mean for the class
- $\sigma_C^2$ = variance for the class

### Visual idea

```text
Probability
   │
   │             /\
   │           /    \
   │         /        \
   │_______/____________\________
            mean (μ)
                    → feature value
```

The model estimates a probability density for each feature within each class.

# Part II — K-Nearest Neighbors (KNN)

# 13. K-Nearest Neighbors

KNN is a supervised learning algorithm that can be used for:

- **Classification**
- **Regression**

The core idea from the notes is:

> Find the nearest data points to a new point and use them to make the prediction.

### Basic flow

```text
Training data
      │
      ↓
New / unknown point
      │
      ↓
Calculate distance to training points
      │
      ↓
Find K nearest points
      │
      ↓
Use their labels / values
      │
      ↓
Prediction
```


## 14. KNN Classification

For classification, KNN looks at the classes of the nearest neighbors.

Example:

```text
        ● ●
      ● ●

          ×  ← new point

      ○ ○
        ○
```

Here:

- `×` → new sample
- `●` → one class
- `○` → another class

If most of the nearest neighbors are `●`, the new point is classified as `●`.

### Majority voting

```text
Nearest neighbors:
●  ●  ○  ●  ○

Class ● = 3
Class ○ = 2

Prediction → ●
```


# 15. KNN Regression

KNN can also predict a numerical value.

Instead of taking a majority vote, the model can use the values of the nearest neighbors.

A simple approach is:

$$
\hat y=
\frac{1}{K}
\sum_{i=1}^{K}y_i
$$

### Example

Suppose the nearest neighbors have house prices:

```text
2
3
4
```

Then:

$$
\hat y=\frac{2+3+4}{3}=3
$$

The handwritten notes connect KNN regression with predicting values such as **house prices**.


# 16. KNN — Step-by-Step Algorithm

The notes describe the process as a sequence of steps.

### Step 1 — Choose `K`

Select the number of neighbors to consider.

### Step 2 — Calculate distances

Calculate the distance between the new point and training examples.

### Step 3 — Sort by distance

Arrange the training points from nearest to farthest.

### Step 4 — Select the nearest `K`

Keep only the closest `K` samples.

### Step 5 — Predict

- **Classification:** majority vote
- **Regression:** aggregate neighbor values, commonly by averaging

```text
New point
   ↓
Calculate distances
   ↓
Sort distances
   ↓
Select K nearest
   ↓
┌───────────────┐
│ Classification │ → Majority vote
│ Regression     │ → Average / aggregate
└───────────────┘
```


# 17. Distance Measures

The notes discuss distance as a central component of KNN.

Common distance measures include:

1. **Euclidean distance**
2. **Manhattan distance**

The choice of distance affects which observations are considered neighbors.


## 18. Euclidean Distance

For two points:

$$
A=(x_1,y_1)
$$

and

$$
B=(x_2,y_2)
$$

Euclidean distance is:

$$
d(A,B)
=
\sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
$$

For $n$ dimensions:

$$
d(A,B)
=
\sqrt{
\sum_{i=1}^{n}(x_i-y_i)^2
}
$$

### Geometric interpretation

```text
y
↑
│        B ●
│         |\
│         | \
│         |  \  Euclidean distance
│         |   \
│        A ●
│
└────────────────→ x
```

It is the straight-line distance between two points.


## 19. Manhattan Distance

Manhattan distance measures movement along coordinate directions.

For two-dimensional points:

$$
d(A,B)
=
|x_2-x_1|+|y_2-y_1|
$$

For $n$ dimensions:

$$
d(A,B)
=
\sum_{i=1}^{n}|x_i-y_i|
$$

### Geometric interpretation

```text
y
↑
│        ● B
│        │
│        │
│   ─────┘
│  ● A
│
└────────────────→ x
```

Instead of taking the diagonal path, Manhattan distance follows horizontal and vertical movements.


## 20. Euclidean vs Manhattan Distance

| Property | Euclidean | Manhattan |
|---|---|---|
| Formula | $\sqrt{\sum(x_i-y_i)^2}$ | $\sum|x_i-y_i|$ |
| Path | Straight line | Coordinate-wise |
| 2D intuition | Diagonal | Grid / city-block |
| Used in KNN | Yes | Yes |

### Quick memory

```text
Euclidean → straight-line distance
Manhattan → city-block distance
```


# 21. Why Feature Scaling Matters in KNN

KNN is distance-based.

Suppose one feature has values:

```text
Age:     18 → 60
```

while another has:

```text
Income:  10,000 → 1,000,000
```

The large-scale feature can dominate the distance calculation.

Therefore, the notes emphasize the need to consider **normalization / standardization** before applying KNN when feature scales differ.

### Concept

```text
Raw features
     ↓
Different scales
     ↓
Distance becomes dominated by large-scale feature
     ↓
Scale / standardize features
     ↓
More balanced distance calculation
```


# 22. Choosing the Value of `K`

The value of `K` controls how many neighbors participate in the prediction.

### Small `K`

Example:

```text
K = 1
```

The prediction depends heavily on a single nearest point.

**Effect:** more sensitive to noise and local fluctuations.

### Larger `K`

A larger neighborhood considers more points.

**Effect:** prediction becomes smoother, but very large values can ignore useful local structure.

### Conceptual trade-off

```text
Small K
  │
  ├── very local
  ├── sensitive to noise
  └── potentially high variance

Large K
  │
  ├── smoother
  ├── less sensitive to individual points
  └── potentially too generalized
```

The handwritten notes explicitly emphasize **choosing an appropriate K**.


## 23. Odd `K` and Tie Avoidance

For binary classification, choosing an odd value of `K` can help avoid a tie in majority voting.

Example:

```text
K = 5

Class A → 3
Class B → 2

Prediction → Class A
```

With an even `K`, a tie can occur:

```text
K = 4

Class A → 2
Class B → 2

Prediction → tie
```

This is a practical reason for commonly trying odd values of `K` in binary classification.

# 26. KNN — Training and Prediction

KNN is often described as a **lazy learning** or **instance-based** method because it does not build a complicated explicit model during training.

Conceptually:

```text

     ↓

     ↓
Very little model fitting
     ↓
Prediction request arrives
     ↓
Calculate distances

```

This means much of the computational work occurs at prediction time.


# 27. Computational Cost

The notes highlight an important KNN issue:

> The prediction step can become expensive when the dataset is large.

For a new query point, the algorithm may need to calculate distances to many training samples.

```text
Small dataset
New point → distances → manageable

Large dataset
New point → many distance calculations → expensive
```

Therefore:

- Prediction can be slower as the dataset grows.
- Storing the training data is necessary.
- Efficient search structures or approximate nearest-neighbor methods can be useful for larger systems.


# 28. Curse of Dimensionality

The notes connect KNN with the **curse of dimensionality**.

As the number of features increases:

```text
2 dimensions
      ↓
3 dimensions
      ↓
10 dimensions
      ↓
100+ dimensions
```

the space becomes increasingly sparse and distance-based neighborhood relationships become less informative.

### Intuition

```text
Low dimensions

● ● ●
 ● ●
  ●


High dimensions

Points become sparse
      ↓
Nearest and farthest points
can become harder to distinguish
      ↓
KNN performance can degrade
```

This is particularly important because KNN relies heavily on meaningful distances.


## 29. Why High Dimensions Can Hurt KNN

KNN assumes that nearby points are useful for predicting a new point.

With many irrelevant or weakly informative features:

```text
More dimensions
      ↓
More possible directions
      ↓
Data becomes sparse
      ↓
Distances become less discriminative
      ↓
Neighborhood quality decreases
      ↓
Prediction can become worse
```

### Practical response

Possible approaches include:

- Feature selection
- Dimensionality reduction
- Removing irrelevant features
- Scaling features appropriately

# 34. Interview / Exam Perspective

### Naive Bayes

**Q: Why is it called "Naive"?**

Because it makes the simplifying assumption that features are conditionally independent given the class.

**Q: Which Naive Bayes model is useful for text counts?**

Multinomial Naive Bayes.

**Q: Which variant is suitable for binary features?**

Bernoulli Naive Bayes.

**Q: Which variant is suitable for continuous numerical data?**

Gaussian Naive Bayes.

---

### KNN

**Q: Is KNN supervised or unsupervised?**

Supervised.

**Q: Can KNN perform regression?**

Yes.

**Q: Why is scaling important?**

Because KNN uses distances, and a feature with a much larger numerical scale can dominate the distance.

**Q: What happens when `K` is too small?**

The model can become sensitive to noise and individual observations.

**Q: What happens when `K` is too large?**

The model can become overly smooth and may lose local patterns.

**Q: Why can high dimensionality hurt KNN?**

Because data becomes sparse and distances become less useful for identifying meaningful neighbors.


````python
# Naive Bayes example: simple binary classification
from sklearn.naive_bayes import BernoulliNB
import numpy as np

# Features: [contains_free, contains_offer, contains_meeting]
X = np.array([
    [1, 1, 0],
    [1, 0, 0],
    [0, 0, 1],
    [0, 1, 1],
    [1, 1, 0],
    [0, 0, 1],
])

# 1 = spam, 0 = not spam
y = np.array([1, 1, 0, 0, 1, 0])

model = BernoulliNB()
model.fit(X, y)

new_email = np.array([[1, 1, 0]])

print("Predicted class:", model.predict(new_email)[0])
print("Class probabilities:", model.predict_proba(new_email)[0])
````

````python
# KNN classification example
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

X = np.array([
    [1, 1],
    [1, 2],
    [2, 1],
    [5, 5],
    [6, 5],
    [5, 6],
])

y = np.array([0, 0, 0, 1, 1, 1])

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)

new_point = np.array([[4, 4]])

print("Predicted class:", model.predict(new_point)[0])
````

````python
# KNN regression example
from sklearn.neighbors import KNeighborsRegressor
import numpy as np

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 3, 4, 5, 6])

model = KNeighborsRegressor(n_neighbors=3)
model.fit(X, y)

new_point = np.array([[3.5]])

print("Predicted value:", model.predict(new_point)[0])
````
