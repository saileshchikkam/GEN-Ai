## 1. What is DBSCAN?

**DBSCAN** stands for:

> **Density-Based Spatial Clustering of Applications with Noise**

DBSCAN is an **unsupervised machine learning clustering algorithm** based on the **density of data points**.

The central idea is:

> Points that are sufficiently close to one another and form a dense region are grouped into the same cluster.

```text
        Dense region
      ● ● ● ●
    ● ● ● ● ●
      ● ● ●

                     ×
                   Noise
```

Unlike a centroid-based approach, DBSCAN focuses on **how densely points are packed together**.

---

# 2. Why Use DBSCAN?

DBSCAN is useful when:

- We want to discover clusters from data without predefined class labels.
- The data may contain **noise/outliers**.
- The clusters may have **arbitrary shapes**.
- We want the algorithm to discover the clusters from the density structure of the data.

```text
Input data
    │
    ▼
Examine local density
    │
    ├── Dense region ──► Cluster
    │
    └── Isolated region ──► Noise / outlier
```

---

# 3. Important DBSCAN Parameters

DBSCAN mainly uses two important parameters:

| Parameter | Meaning | Easy way to remember |
|---|---|---|
| **`eps` (ε)** | Radius/neighborhood distance around a point | **How far do we look?** |
| **`min_samples` / `minPts`** | Minimum number of points required in that neighborhood to form a dense region | **How many points are enough?** |

### Visual interpretation

```text
                 ●
             ●   ●   ●
           ●     P     ●
             ●   ●   ●
                 ●

             ← ε → 
        neighborhood radius
```

`eps` defines the size of the neighborhood around a point.

`min_samples` defines how many points must be present in that neighborhood for it to be considered sufficiently dense.

---

# 4. DBSCAN Neighborhood

For every point, DBSCAN checks the points that lie within the selected `eps` radius.

```text
              ●
         ●         ●

      ●      P      ●

         ●         ●
              ●

        ←── eps ──→
```

Conceptually:

```text
Point P
   │
   ▼
Find points within ε
   │
   ▼
Count neighboring points
   │
   ▼
Compare with min_samples
```

# 5. How DBSCAN Works

The basic process from the notes can be organized as follows:

```text
                 Start
                   │
                   ▼
             Select a point
                   │
                   ▼
        Find its ε-neighborhood
                   │
                   ▼
       Count points in that region
                   │
             ┌─────┴─────┐
             │            │
        Enough points   Not enough
             │            │
             ▼            ▼
        Core point     Border/Noise
             │
             ▼
       Expand the cluster
             │
             ▼
      Check neighboring points
             │
             ▼
        Continue expansion
```

# 6. Core Point

A **core point** is a point whose `eps` neighborhood contains at least the required number of points.

The notes summarize the condition as:

```text
Number of points within ε ≥ minPts
                 ↓
             Core point
```

### Diagram

```text
             ● ● ●
          ● ● P ● ●
             ● ● ●

               P
          = Core Point
```

The surrounding points indicate that the region around `P` is sufficiently dense.

---

# 7. Border Point

A **border point** is associated with a cluster but does not itself have enough neighboring points to satisfy the core-point condition.

Conceptually:

```text
        Dense cluster

        ● ● ●
      ● ● ● ●
        ● P ●
             \
              \
               ○
            Border point
```

A border point can be reached from a nearby core region even though its own neighborhood is not dense enough to make it a core point.

---

# 8. Noise / Outlier

A point that is **not reachable from any core point** is treated as noise.

```text
        Cluster

      ● ● ● ●
    ● ● ● ● ●
      ● ● ●


                         ×
                       Noise
```

### Key idea

```text
Core point      → belongs to dense region
Border point    → near/reachable from a cluster
Noise           → not reachable from a core region
```

---

# 9. Core, Border and Noise — Comparison

| Point type | Density condition | Cluster membership |
|---|---|---|
| **Core point** | Has enough points within `eps` | Forms/expands a cluster |
| **Border point** | Not dense enough itself, but associated with a core region | Belongs to a nearby cluster |
| **Noise** | Not sufficiently connected to a core region | Treated as outlier/noise |

### Memory trick

```text
CORE   = Dense
BORDER = Near a dense cluster
NOISE  = Isolated
```

---

# 10. DBSCAN Cluster Expansion

Once DBSCAN identifies a core point, it can expand the cluster.

```text
             Core point
                 ●
              ↙  ↓  ↘
            ●    ●    ●
           ↙           ↘
         ●               ●
          \             /
           ● ● ● ● ● ●
```

Conceptually:

1. Find a core point.
2. Add points in its `eps` neighborhood.
3. If a neighboring point is also a core point, expand through that point.
4. Continue until no more density-connected points can be added.
5. Start searching for another unvisited dense region.

---

# 11. DBSCAN Algorithm — High-Level Flow

```text
                 DATASET
                    │
                    ▼
              Choose a point
                    │
                    ▼
           Find ε-neighborhood
                    │
                    ▼
        Count neighboring points
                    │
          ┌─────────┴─────────┐
          │                   │
      count ≥ minPts      count < minPts
          │                   │
          ▼                   ▼
      Core point        Not a core point
          │                   │
          ▼             ┌─────┴─────┐
   Create/expand        │           │
      cluster       Reachable   Not reachable
                         │           │
                         ▼           ▼
                       Border      Noise
```

# 12. DBSCAN Example

Suppose:

```text
eps = chosen neighborhood radius
minPts = minimum required number of points
```

For a point `P`:

```text
Points inside ε-neighborhood = 6
minPts = 4
```

Since:

```text
6 ≥ 4
```

`P` is classified as a **core point**.

Another point may have:

```text
Points inside ε-neighborhood = 2
minPts = 4
```

Then:

```text
2 < 4
```

It is **not a core point**.

If it is reachable from a core point, it can be a **border point**; otherwise it is **noise**.

# 13. Density-Based Thinking

The main difference in DBSCAN is that we do not simply ask:

> "Which centroid is this point closest to?"

Instead, we ask:

> "Is this point located in a sufficiently dense region?"

```text
Centroid-based thinking
        │
        ▼
Nearest center?


DBSCAN thinking
        │
        ▼
Enough nearby points?
        │
        ▼
Dense region?
```

# 23. DBSCAN Parameter Selection — Intuition

The notes emphasize searching for suitable values for:

- `eps`
- `minPts`

Think of them as two questions:

```text
eps
 │
 └── "How far should I look?"

minPts
 │
 └── "How many points should I require?"
```

### If the neighborhood is too small

```text
eps too small
     ↓
Very few neighbors
     ↓
Many points may appear as noise
```

### If the neighborhood is too large

```text
eps too large
     ↓
Many regions become connected
     ↓
Separate clusters may merge
```

The handwritten notes do not specify a universal numerical rule for choosing these parameters, so parameter selection should be treated as a separate practical topic.

# 27. Interview Questions

### Q1. What is DBSCAN?

DBSCAN is an unsupervised, density-based clustering algorithm that groups points in dense regions and identifies points that do not belong to dense regions as noise.

### Q2. What does DBSCAN stand for?

**Density-Based Spatial Clustering of Applications with Noise.**

### Q3. What are the two main DBSCAN parameters?

- `eps` — neighborhood radius.
- `minPts` / `min_samples` — minimum number of points required for a dense region.

### Q4. What is a core point?

A point whose `eps` neighborhood contains at least the required number of points.

### Q5. What is a border point?

A point that is not itself sufficiently dense to be a core point but is associated with a nearby core region.

### Q6. What is noise in DBSCAN?

A point that is not reachable from any core point and therefore does not belong to a dense cluster.

### Q7. Why is DBSCAN useful for arbitrary-shaped clusters?

Because it groups points according to density connectivity rather than forcing every cluster around a centroid.

### Q8. What happens if `eps` is too small?

The neighborhoods become very small, which can cause many points to fail the density requirement and be treated as noise.

### Q9. What happens if `eps` is too large?

Distinct dense regions can become connected and may be merged into larger clusters.

### Q10. Name some applications of DBSCAN.

- Geospatial analysis
- Delivery-location clustering
- Fraud/anomaly detection
- Network intrusion detection
- Astronomy and galaxy-cluster identification
