## 🧠 Big Picture

As an ML engineer or data scientist, we usually **do not start by downloading a dataset and training a model**.

We start with a **business problem**.

```text
Business Problem
       ↓
Understand the Goal
       ↓
Frame as an ML Problem
       ↓
Choose Problem Type
       ↓
Collect the Right Data
       ↓
Build / Reuse a Model
       ↓
Evaluate Predictions
       ↓
Deploy
       ↓
Monitor + Optimize
```

### ⭐ Central idea

> **A good ML project starts with the right problem definition, not with the model.**

# 1. 🎯 What Does "Framing the Business Problem" Mean?

**Framing** means converting a real-world business requirement into a clear machine-learning problem.

A business team may say:

> "We want to reduce customer churn."

That statement is useful, but it is **not yet a complete ML problem**.

We need to ask:

- What exactly is churn?
- Which customers are likely to leave?
- How far in advance should we predict?
- What data is available?
- What type of ML problem is this?
- What prediction should the model produce?
- How will the prediction help the business?

### Business → ML translation

```text
Business requirement
        ↓
Specific business question
        ↓
ML formulation
        ↓
Input features
        ↓
Target / output
        ↓
Model prediction
        ↓
Business action
```

### Example

```text
Business Goal:
Reduce customer churn

ML Question:
Which customers are likely to leave the platform?

Input:
Customer behavior, usage, subscription information, etc.

Output:
Probability that a customer will churn

Business Action:
Target high-risk customers with retention strategies.

# 2. 🏢 Why Business Problems Come First

Imagine joining a company as a senior ML developer.

You normally would **not** start with:

```text
"Let me download a huge dataset."
"Let me train Random Forest."
"Let me try XGBoost."
```

Instead, first understand:

```text
What problem does the business have?
              ↓
Why does it matter?
              ↓
What decision needs to be improved?
              ↓
Can ML help with that decision?
```

### Example: Netflix-like platform

Suppose the business notices:

```text
Customer churn is increasing.
```

The business wants to reduce it.

Possible ML-related questions:

```text
1. Which users are likely to leave?
2. What is the probability of leaving?
3. When are they likely to leave?
4. Which intervention might retain them?
```

The first two can naturally lead to supervised ML problems.

> **The ML model is a tool for solving a business problem — it is not the business goal itself.**

# 3. 💼 Example — Increasing Prices

Consider this business situation:

```text
Current situation:
Costs are increasing.

Business decision:
Increase product/subscription prices.

Risk:
Customers may leave because of the higher price.
```

The company may have two competing goals:

```text
Increase revenue
       ↕
Avoid excessive customer churn
```

Therefore, the business problem can become:

> **How can we change pricing while minimizing customer loss and maximizing business value?**

This can lead to several ML problems depending on what the company wants to predict.

### Possible ML formulations

```text
Question:
Which users are likely to leave?
        ↓
Classification

Question:
What is the probability of a user leaving?
        ↓
Classification / probability estimation

Question:
How many customers might leave after a price change?
        ↓
Regression / forecasting, depending on formulation
```

### ⭐ Important

The same business situation can sometimes be framed as **different ML problems** depending on the exact prediction required.

# 4. 🧩 Step 1 — Identify the Actual Business Problem

Before selecting a model, clearly write the business objective.

### Good framing

Instead of:

❌ "Build a machine-learning model."

Write:

✅ "Reduce customer churn by identifying customers who are likely to leave."

The second statement gives us a business objective and a direction for the ML problem.

### Ask these questions

```text
1. What is the business trying to improve?
2. What decision needs to be made?
3. What outcome matters?
4. Who will use the prediction?
5. What action will happen after prediction?
```

### Example

```text
Business Problem:
Too many customers are leaving.

Business Goal:
Reduce churn.

Decision:
Which customers should receive retention offers?

ML Goal:
Predict which customers are likely to churn.

# 5. 🔄 Step 2 — Convert the Business Problem into a Mathematical / ML Problem

Once the business problem is understood, translate it into a mathematical formulation.

From the notes:

```text
Business Problem
       ↓
Mathematical Problem
       ↓
ML Problem
```

### Example

Business statement:

> "How do we reduce churn?"

Convert it into:

> "Which users are likely to leave the platform?"

Now we can define:

```text
Input X = customer information and behavior

Output y = churn / no churn
```

For example:

```text
X:
- Age
- Subscription type
- Usage frequency
- Number of sessions
- Payment history
- Customer support interactions

y:
- 1 → customer churned
- 0 → customer did not churn
```

This gives us a clear supervised-learning target.

# 6. 🔢 Step 3 — Decide the Type of ML Problem

A very important step is identifying whether the problem is:

```text
Classification
Regression
Clustering / Unsupervised Learning
```

The choice depends primarily on **what we want to predict**.

---

## 🟦 Classification

Use classification when the target is a **category/class**.

Examples:

```text
Will the customer leave?
→ Yes / No

Is this email spam?
→ Spam / Not Spam

Is the transaction fraudulent?
→ Fraud / Not Fraud
```

Output:

```text
Class label
```

or often:

```text
Probability for each class
```

### Churn example

```text
Customer → Model → Churn probability

Amit → 0.82
Rahul → 0.17
Priya → 0.64
```

A threshold can then be used to make a class decision.

# 7. 📈 Regression

Use regression when the target is a **continuous numerical value**.

Examples:

```text
Predict house price
Predict sales
Predict revenue
Predict demand
Predict customer lifetime value
```

### Example

Suppose we want to predict monthly revenue:

```text
Input customer/business information
             ↓
          ML Model
             ↓
      Predicted Revenue
             ↓
          ₹85,000
```

### Important distinction

```text
Classification → category / class
Regression     → numerical quantity
```

### Churn example

If the question is:

> "Will this customer leave?"

→ **Classification**

If the question is:

> "How much revenue might we lose from this customer?"

→ **Regression**, assuming the target is a continuous monetary value.

# 8. 🤖 ChatGPT / Chatbot Example

The notes also connect business/ML problem framing to chatbot systems.

Suppose we have a chatbot:

```text
User
 ↓
Chatbot
 ↓
Model
 ↓
Response
```

A question such as:

> "What is the probability of the next word occurring?"

can be formulated mathematically.

Modern language models estimate probabilities over possible tokens.

Simplified example:

```text
Input:
"The weather is"

Possible next tokens:

good      → 0.40
sunny     → 0.25
cold      → 0.15
nice      → 0.10
...
```

The model uses these probabilities as part of generating a response.

### Key idea

Natural-language behavior is ultimately represented using **mathematical operations and numerical probabilities** inside the model.

# 10. 📊 Probability of a Customer Leaving

A useful classification model can provide a probability rather than only a yes/no answer.

Example:

| Customer | Churn Probability |
|---|---:|
| A | 0.90 |
| B | 0.72 |
| C | 0.25 |
| D | 0.08 |

If the business defines a threshold of `0.50`:

```text
Probability ≥ 0.50
→ High-risk

Probability < 0.50
→ Lower-risk
```

So:

```text
A → High risk
B → High risk
C → Lower risk
D → Lower risk
```

### ⚠️ Threshold is a business decision

The best threshold is not automatically `0.50`.

For example:

- If contacting a customer is cheap, the company may contact more users.
- If retention offers are expensive, the company may target fewer, higher-risk users.

Therefore:

> **Model predictions should ultimately support a business decision.**

# 11. 🏷️ Labels / Target Variable

For supervised learning, we need a target variable.

For churn:

```text
Features X
      ↓
   Model
      ↓
Target y
```

Example dataset:

| Usage | Support Tickets | Subscription | Churn |
|---:|---:|---|---:|
| 20 | 1 | Premium | 0 |
| 3 | 8 | Basic | 1 |
| 25 | 0 | Premium | 0 |
| 4 | 6 | Basic | 1 |

Here:

```text
X = Usage, Support Tickets, Subscription
y = Churn
```

The model learns a relationship between the input features and the target.

### Core supervised-learning structure

\[
X 
ightarrow f(X) 
ightarrow \hat{y}
\]

where:

- `X` = input features
- `f` = learned model
- `ŷ` = predicted output

# 12. 🏗️ Step 4 — Building a New Model vs Reusing an Existing Model

The notes emphasize an important real-world point:

> **Building a new model from scratch can be difficult and expensive.**

In a professional environment, you should first check whether an existing solution can be reused or adapted.

### Possible approach

```text
Current solution / existing model
            ↓
Study its strengths and weaknesses
            ↓
Use available insights
            ↓
Adapt / improve / retrain
```

This is often more practical than starting from zero.

### Why?

Because building an ML system requires:

- data,
- feature engineering,
- model selection,
- training,
- validation,
- deployment,
- monitoring,
- maintenance.

If a good existing model or pipeline is available, reusing it can save significant time and resources.

# 13. 🧑‍💻 Senior Data Scientist Insight

The notes highlight an important professional practice:

```text
Current solution
       ↓
Consult experienced / senior data scientists
       ↓
Understand existing insights
       ↓
Improve the solution
```

When entering an existing ML project, investigate:

```text
What model is currently used?
Why was it selected?
What data was used?
What features were important?
What metrics were tracked?
What problems occurred?
How is it deployed?
How is it monitored?
```

### ⭐ Engineering principle

> **Do not rebuild something blindly when an existing solution and institutional knowledge are available.**

# 14. 🔍 Step 5 — Getting the Right Data

Once the ML problem is framed, we need data that actually represents the problem.

For churn prediction, relevant data might include:

```text
User profile
      +
Usage behavior
      +
Subscription information
      +
Payment behavior
      +
Support interactions
      ↓
Churn dataset
```

### What we should avoid

Collecting large amounts of irrelevant data simply because it is available.

Instead:

> **Collect data that is relevant to the business question and prediction target.**

This improves the usefulness of the resulting model.

### Important questions

```text
Does the data contain the target?
Does it contain useful predictive signals?
Is it representative?
Is it recent enough?
Is it legally and ethically usable?
Is it available at prediction time?
```

# 15. 📏 Step 6 — Define Metrics to Measure Prediction Quality

After training a model, we need to know:

> **Are the predictions actually correct/useful?**

This requires evaluation metrics.

For classification, common metrics include:

```text
Accuracy
Precision
Recall
F1-score
ROC-AUC
```

For regression, common metrics include:

```text
MAE
MSE
RMSE
R²
```

### Simple idea

```text
Actual outcome
      ↓
Compare with
      ↑
Predicted outcome
      ↓
Evaluation metric
```

### Why metrics matter

A model can produce predictions, but a prediction system is useful only when its performance is measured against an appropriate objective.

# 16. ⚖️ Choose Metrics Based on the Business Problem

Metric selection should not be arbitrary.

Suppose the business wants to identify customers likely to churn.

If missing a high-risk customer is very costly, **recall** may be especially important.

If contacting customers is expensive and false alarms are costly, **precision** may become more important.

### Example

```text
Actual churners = 100

Model catches 90
→ Recall = 90%
```

But suppose the model flags 300 customers and only 90 actually churn.

Then many contacted customers were false positives.

Therefore:

```text
Business cost
      ↓
Metric choice
      ↓
Threshold choice
      ↓
Model decision
```

> **The best model is not necessarily the one with the highest generic accuracy. It is the one that best serves the business objective under the relevant constraints.**

# 17. 🌐 Online Learning vs Batch Learning

The notes introduce two important approaches.

## 🟢 Online Learning

The model can be updated continuously or incrementally as new data becomes available.

```text
New data
   ↓
Model update
   ↓
New data
   ↓
Model update
   ↓
...
```

This is useful when data arrives continuously and the environment changes quickly.

### Conceptual example

```text
User behavior today
        ↓
Model update
        ↓
New prediction tomorrow
```

---

## 🔵 Batch Learning

The model is trained on accumulated data periodically.

```text
Collect data
     ↓
Large dataset
     ↓
Train model
     ↓
Deploy
     ↓
Collect more data
     ↓
Train again later
```

This is useful when retraining can happen periodically rather than continuously.

# 18. 🍿 Netflix-Style Example — Batch Learning

The notes use a Netflix-like example to illustrate batch learning.

Suppose recommendation data is continuously generated:

```text
User watches movies
User searches
User skips content
User rates content
```

Instead of retraining the model every second, the organization may:

```text
Collect recommendation data
        ↓
Aggregate data
        ↓
Train periodically
        ↓
Deploy updated model
```

For example:

```text
Weekly / periodic retraining
        ↓
New recommendation model
        ↓
Deploy to users
```

### Why batch learning?

Because the recommendation data can be accumulated and processed periodically.

This can make the pipeline:

- easier to schedule,
- easier to validate,
- easier to reproduce,
- more computationally manageable.

The exact retraining frequency should depend on how quickly the data distribution and business environment change.

# 23. 🎓 Interview-Ready Questions & Answers

### Q1. What is problem framing in ML?

**Answer:**  
Problem framing is the process of converting a real-world business requirement into a clearly defined machine-learning problem with specified inputs, target output, problem type, evaluation metrics, and business objective.

---

### Q2. Why is business problem framing important?

**Answer:**  
It ensures that the ML model solves a meaningful business problem rather than simply optimizing a technical metric.

---

### Q3. How do you convert a business problem into an ML problem?

**Answer:**

```text
Business goal
→ define prediction question
→ identify inputs
→ define target
→ select ML problem type
→ choose evaluation metrics
→ build model
```

---

### Q4. How do you identify classification vs regression?

**Answer:**

```text
Categorical output → Classification
Continuous numerical output → Regression
```

---

### Q5. What is churn prediction?

**Answer:**  
Churn prediction is the task of predicting which customers are likely to stop using a product or service. It is commonly framed as a binary classification problem.

---

### Q6. What is batch learning?

**Answer:**  
Batch learning trains a model periodically using an accumulated dataset rather than updating the model continuously after every new observation.

---

### Q7. What is online learning?

**Answer:**  
Online learning updates a model incrementally as new data becomes available.

---

### Q8. Why should we reuse an existing ML solution?

**Answer:**  
Reusing or improving an existing solution can save development time and leverage existing data, engineering work, domain knowledge, and lessons learned.
