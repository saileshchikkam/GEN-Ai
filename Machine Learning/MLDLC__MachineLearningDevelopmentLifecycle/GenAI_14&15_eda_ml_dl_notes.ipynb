````python
# Conceptual reconstruction of the handwritten neural-network sketch

import matplotlib.pyplot as plt

layers = [("Input", 3), ("Layer 1", 4), ("Layer 2", 4), ("Output", 2)]
coords = {}

fig, ax = plt.subplots(figsize=(10, 5))

for x, (name, n) in enumerate(layers):
    ys = [(n - 1) / 2 - i for i in range(n)]
    coords[x] = [(x, y) for y in ys]
    for y in ys:
        ax.add_patch(plt.Circle((x, y), 0.12, fill=False, linewidth=2))
    ax.text(x, max(ys) + 0.7, name, ha="center", fontsize=12, fontweight="bold")

for x in range(len(layers) - 1):
    for x1, y1 in coords[x]:
        for x2, y2 in coords[x + 1]:
            ax.plot([x1, x2], [y1, y2], linewidth=0.7)

ax.set_xlim(-0.5, 3.5)
ax.set_ylim(-2.5, 2.5)
ax.axis("off")
plt.title("Conceptual Layers of Neurons")
plt.show()
````
