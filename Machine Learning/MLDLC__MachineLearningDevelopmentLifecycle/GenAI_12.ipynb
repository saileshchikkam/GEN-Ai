````python

# Additional executable demonstration:
# A small automated missing-value cleaning workflow.

import pandas as pd
import numpy as np

data = pd.DataFrame({
    "name": ["A", "B", "C", "D"],
    "age": [20, np.nan, 22, np.nan],
    "score": [80, 85, np.nan, 90]
})

print("Before cleaning:")
display(data)

# Simple rule-based automation for demonstration.
cleaned = data.copy()

numeric_columns = cleaned.select_dtypes(include="number").columns
for column in numeric_columns:
    cleaned[column] = cleaned[column].fillna(cleaned[column].median())

print("After automated numeric missing-value filling:")
display(cleaned)
````
