# 1. 🤖 Data Gathering in Agentic AI

In traditional ML projects, data gathering means collecting data from sources such as:

```text
CSV
Database
API
Web
```

In **Agentic AI**, an AI system can actively interact with tools and external systems to gather information.

### Basic idea

```text
User
 ↓
AI Agent
 ↓
Understand the request
 ↓
Choose the required tool
 ↓
Get data
 ↓
Use the returned information
 ↓
Give result to user
```

The important change is that the AI system is not limited to information already stored inside the model.

It can use external tools, APIs, databases and other services to obtain current or task-specific information.

# 2. 🧩 AI Application vs Agentic AI

## Normal AI Application

A conventional AI application may mainly work as:

```text
User
 ↓
AI Model
 ↓
Response
```

---

## Agentic AI

An agentic system can take actions using tools.

```text
User
 ↓
AI Agent
 ↓
Reason about the task
 ↓
Select tool
 ↓
API / Database / Web / Other Tool
 ↓
Receive result
 ↓
Continue task
 ↓
Final response
```

### ⭐ Key idea

> **Agentic AI extends an AI application by allowing the system to interact with tools and external environments to accomplish a task.**

# 6. 🧾 JSON — JavaScript Object Notation

**JSON** stands for **JavaScript Object Notation**.

It is a commonly used format for exchanging structured data between software systems.

### Basic JSON

```json
{
    "movie": "Avengers",
    "date": "Sunday",
    "theater": "PVR"
}
```

JSON mainly represents information using **key-value pairs**.

```text
key        → value
movie      → Avengers
date       → Sunday
theater    → PVR
```

### Why JSON is useful

- Easy for humans to understand
- Easy for programs to parse
- Supported by many programming languages
- Widely used in APIs

# 7. 🔑 JSON Key-Value Pairs

The fundamental structure is:

```text
"key": "value"
```

Example:

```json
{
    "name": "Amit",
    "age": 22,
    "city": "Hyderabad"
}
```

Here:

```text
name → Amit
age  → 22
city → Hyderabad
```

JSON can also contain arrays and nested objects.

### Array

```json
{
    "tickets": ["A1", "A2"]
}
```

### Nested object

```json
{
    "user": {
        "name": "Amit",
        "city": "Hyderabad"
    }
}
```

# 19. 🧠 JSON vs SQL

Do not confuse JSON and SQL.

## JSON

JSON is mainly a **data representation / interchange format**.

```json
{
    "department": "Tech",
    "location": "Hyderabad"
}
```

---

## SQL

SQL is a **language used to work with relational databases**.

```sql
SELECT *
FROM employees
WHERE department = 'Tech';
```

### Simple distinction

```text
JSON → represents / exchanges data

SQL  → queries / manipulates relational database data
```

# 20. 🧪 Python Example — JSON

Python can work directly with JSON.

```python
import json

data = {
    "movie": "Avengers",
    "tickets": 2,
    "time": "5 PM"
}

json_data = json.dumps(data)

print(json_data)
```

### Convert JSON back to Python

```python
python_data = json.loads(json_data)

print(python_data["movie"])
```

# 21. 🧪 Python Example — API + JSON

A simple API interaction can look like:

```python
import requests

response = requests.get("API_ENDPOINT")

if response.status_code == 200:
    data = response.json()
    print(data)
else:
    print("API request failed")
```

For an API that expects JSON in a POST request:

```python
import requests

payload = {
    "movie": "Avengers",
    "tickets": 2
}

response = requests.post(
    "API_ENDPOINT",
    json=payload
)

print(response.json())
```

> In a real application, use the API's documented authentication, endpoint, parameters and error-handling requirements.

# 22. 🧪 Python Example — SQL

A Python application can communicate with a database.

```python
query = """
SELECT *
FROM employees
WHERE department = 'Tech';
"""

cursor.execute(query)

rows = cursor.fetchall()

for row in rows:
    print(row)
```

The exact connection code depends on the database system, such as:

```text
MySQL
PostgreSQL
SQLite
SQL Server
Oracle
```

### Agentic pattern

```text
Natural Language
       ↓
LLM
       ↓
SQL
       ↓
Database
       ↓
Rows
       ↓
LLM
       ↓
Natural Language
```

# 23. ⚠️ Safety Considerations for Text-to-SQL

Allowing an LLM to generate SQL is powerful, but generated queries must be controlled.

### Good practices

```text
✔ Validate generated SQL
✔ Use read-only database permissions when possible
✔ Restrict accessible tables/columns
✔ Use parameterized queries for user values
✔ Add query limits
✔ Log important database actions
✔ Prevent destructive operations unless explicitly authorized
```

Avoid giving an AI agent unrestricted database access.

For example, a query such as:

```sql
DROP TABLE employees;
```

could be destructive.

### ⭐ Rule

> **Agentic AI should have only the permissions it actually needs.**

# 27. 🎓 Interview Questions & Answers
