# 7. Categorical Data Encoding

Machine-learning algorithms generally require numerical representations of input data.

Categorical values therefore need to be encoded.

## Common encoding approaches

### Label Encoding

Maps categories to integer values.

```text
Category A → 0
Category B → 1
Category C → 2
```

### Ordinal Encoding

Used when categories have a meaningful order.

```text
Low    → 0
Medium → 1
High   → 2
```

### One-Hot Encoding

Creates separate binary columns.

```text
Color
-----
Red
Blue
Green
```

becomes conceptually:

```text
Color_Red   Color_Blue   Color_Green
    1           0             0
    0           1             0
    0           0             1
```

> The handwritten notes mention categorical data being converted into numerical data and refer to ordinal and one-hot encoding.

# 8. Feature Scaling

Feature scaling brings numerical features into comparable ranges.

This is especially important for algorithms that are sensitive to feature magnitude.

## Two common approaches

| Method | Main idea |
|---|---|
| **Standardization** | Centers data around mean 0 with standard deviation 1 |
| **Normalization** | Rescales values to a bounded range, commonly 0–1 |

### Standardization

$$
z = \frac{x-\mu}{\sigma}
$$

where:

- $x$ = original value
- $\mu$ = mean
- $\sigma$ = standard deviation

### Normalization

A common min-max form is:

$$
x' = \frac{x-x_{min}}{x_{max}-x_{min}}
$$

# 9. Train-Test Splitting

The dataset should be divided so that the model can be trained on one portion and evaluated on unseen data.

A common example is:

```text
Dataset
   │
   ├───────────────┐
   ▼               ▼
Training         Testing
  80%              20%
   │               │
   ▼               │
Train model        │
   │               │
   └───────┬───────┘
           ▼
       Evaluate
```

### Why?

If the model is evaluated only on the data it already saw during training, the evaluation may not represent how well it generalizes to new data.

# 13. Regression Evaluation Metrics

The notes list:

- MSE
- RMSE
- MAE

## Mean Squared Error — MSE

$$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

MSE squares the errors, so larger errors receive greater penalty.

## Root Mean Squared Error — RMSE

$$
RMSE = \sqrt{MSE}
$$

RMSE is expressed in the same units as the target variable.

## Mean Absolute Error — MAE

$$
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
$$

MAE measures the average absolute prediction error.

### Comparison

| Metric | Error treatment | Key property |
|---|---|---|
| **MAE** | Absolute error | Easy to interpret |
| **MSE** | Squared error | Penalizes large errors strongly |
| **RMSE** | Square root of MSE | Same units as target |

# 16. Classification Evaluation Metrics

The notes list:

1. Accuracy
2. Precision
3. Recall
4. F1-score
5. Confusion Matrix

## Confusion Matrix

For binary classification:

```text
                         Actual
                    Positive   Negative
                 ┌───────────┬───────────┐
Predicted Positive│    TP     │    FP     │
                 ├───────────┼───────────┤
Predicted Negative│    FN     │    TN     │
                 └───────────┴───────────┘
```

| Term | Meaning |
|---|---|
| **TP** | True Positive |
| **TN** | True Negative |
| **FP** | False Positive |
| **FN** | False Negative |

### Accuracy

$$
Accuracy = \frac{TP+TN}{TP+TN+FP+FN}
$$

### Precision

$$
Precision = \frac{TP}{TP+FP}
$$

Precision answers:

> Of the samples predicted as positive, how many were actually positive?

### Recall

$$
Recall = \frac{TP}{TP+FN}
$$

Recall answers:

> Of all actual positive samples, how many did the model correctly identify?

### F1-score

$$
F1 = 2\cdot\frac{Precision\cdot Recall}{Precision+Recall}
$$

F1 balances precision and recall.

|---|---|---|

# 28. Interview Questions

### 1. What is feature engineering?
Creating or transforming input variables so that they provide useful information to a machine-learning model.

### 2. What is feature selection?
Selecting the most relevant features from the available features.

### 3. Why do we encode categorical variables?
Many ML algorithms require numerical inputs, so categorical values need suitable numerical representations.

### 4. Why is feature scaling used?
To prevent features with larger numerical magnitudes from disproportionately influencing algorithms that are sensitive to scale.

### 5. What is the difference between normalization and standardization?
Normalization commonly maps values to a bounded range such as 0–1, while standardization transforms values using their mean and standard deviation.

### 6. Why split data into training and testing sets?
To evaluate how well the trained model generalizes to unseen data.

### 7. What is regression?
A supervised-learning task for predicting continuous numerical values.

### 8. What is classification?
A supervised-learning task for predicting discrete classes/categories.

### 9. Name common regression metrics.
MAE, MSE, and RMSE.

### 10. Name common classification metrics.
Accuracy, precision, recall, F1-score, and confusion matrix.

### 11. What is hyperparameter tuning?
Searching for suitable hyperparameter values that improve model performance.

### 12. What is unsupervised learning?
Learning patterns or structure from data without provided target labels.

### 13. What is the main difference between supervised and unsupervised learning?
Supervised learning has target labels; unsupervised learning does not have provided target labels during training.
