```python
df.info() ## we saved the some memory😁
```

```python
df['Blood Oxygen Level (%)'] = df['Blood Oxygen Level (%)'].astype('float32')
```

```python
df.info()## we saved the some memory😁
```

```python
df['Step Count'] = df['Step Count'].astype('float32')
```

```python
df.info()## Again we saved the some memory😁
```
