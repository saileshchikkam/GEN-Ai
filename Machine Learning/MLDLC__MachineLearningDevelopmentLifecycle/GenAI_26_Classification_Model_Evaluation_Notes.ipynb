# Classification Model Evaluation

Classification models do not need to be evaluated only by asking whether the prediction is correct.

A useful evaluation should answer questions such as:

- How many positive cases were predicted correctly?
- How many negative cases were predicted correctly?
- How many positive predictions were actually wrong?
- How many positive cases did the model miss?
- Is the model reliable when the classes are imbalanced?

The main concepts covered here are:

1. **Confusion Matrix**
2. **Accuracy**
3. **Precision**
4. **Recall**
5. **F-score / Fβ-score**

> The notes are reorganized into a connected study format while retaining the concepts, examples, formulas, and distinctions from the handwritten material.

# 1. Classification Evaluation — Big Picture

For a binary classification problem, the model usually predicts one of two classes.

For example:

- Positive / Negative
- Spam / Not Spam
- Disease / No Disease
- Fraud / Not Fraud

The model's prediction is compared with the actual class.

```text
                 CLASSIFICATION MODEL
                         │
                         ↓
                 Make a prediction
                         │
                         ↓
              Compare with actual value
                         │
             ┌───────────┴───────────┐
             ↓                       ↓
       Prediction is correct   Prediction is wrong
             │                       │
             ↓                       ↓
        TP / TN                  FP / FN
             │                       │
             └───────────┬───────────┘
                         ↓
                Evaluation metrics
                         │
       ┌──────────┬──────┼──────┬──────────┐
       ↓          ↓      ↓      ↓          ↓
 Confusion    Accuracy Precision Recall   F-score
   Matrix
```

# 2. Confusion Matrix

A **confusion matrix** is a table that compares:

- the **actual class**
- the **predicted class**

For binary classification, this produces four possible outcomes.

|  | **Actual Positive** | **Actual Negative** |
|---|---:|---:|
| **Predicted Positive** | **True Positive (TP)** | **False Positive (FP)** |
| **Predicted Negative** | **False Negative (FN)** | **True Negative (TN)** |

These four values form the foundation of the major classification metrics.

# 3. Understanding TP, TN, FP and FN

## True Positive (TP)

The model predicts **positive**, and the actual class is also **positive**.

```text
Prediction = Positive
Actual     = Positive
             ↓
             TP
```

Example: A patient has a disease and the model correctly predicts disease.

---

## True Negative (TN)

The model predicts **negative**, and the actual class is also **negative**.

```text
Prediction = Negative
Actual     = Negative
             ↓
             TN
```

Example: A patient does not have the disease and the model correctly predicts no disease.

---

## False Positive (FP)

The model predicts **positive**, but the actual class is **negative**.

```text
Prediction = Positive
Actual     = Negative
             ↓
             FP
```

This is a **false alarm**.

---

## False Negative (FN)

The model predicts **negative**, but the actual class is **positive**.

```text
Prediction = Negative
Actual     = Positive
             ↓
             FN
```

This means the model **missed a positive case**.

# 4. Confusion Matrix — Visual View

```text
                         ACTUAL
                    Positive   Negative
                   ┌──────────┬──────────┐
PREDICTED Positive │    TP    │    FP    │
                   ├──────────┼──────────┤
PREDICTED Negative │    FN    │    TN    │
                   └──────────┴──────────┘
```

### Easy memory trick

- **True** → prediction agrees with reality.
- **False** → prediction disagrees with reality.
- **Positive / Negative** → what the model predicted.

Therefore:

> TP = correctly predicted positive  
> TN = correctly predicted negative  
> FP = incorrectly predicted positive  
> FN = incorrectly predicted negative

# 5. Correct Predictions and Accuracy

Accuracy measures the proportion of **all predictions that are correct**.

Correct predictions are:

$$
TP + TN
$$

Total predictions are:

$$
TP + TN + FP + FN
$$

Therefore:

$$
\boxed{
Accuracy =
\frac{TP+TN}
{TP+TN+FP+FN}
}
$$

As a percentage:

$$
Accuracy(\%) = Accuracy \times 100
$$

# 6. Accuracy — Simple Example

Suppose a model makes **100 predictions**.

If **95 predictions are correct**, then:

$$
Accuracy = \frac{95}{100}
$$

$$
\boxed{Accuracy=95\%}
$$

The basic idea is:

```text
Total predictions
       │
       ├── Correct → TP + TN
       │
       └── Incorrect → FP + FN

Accuracy = Correct predictions / Total predictions
```

# 7. Limitation of Accuracy — Imbalanced Data

The notes emphasize an important limitation:

> **Accuracy can be misleading when the dataset is imbalanced.**

Suppose:

- 95 samples are negative
- 5 samples are positive

A model predicts **negative for every sample**.

It gets 95 predictions correct:

$$
Accuracy = \frac{95}{100}=95\%
$$

That looks excellent, but the model completely fails to identify the positive class.

```text
Actual data
────────────────────────────
Negative:  ███████████████████ 95
Positive:  █                    5

Model predicts "Negative" for everything.

Accuracy → 95%
Positive detection → 0%
```

Therefore, accuracy alone may not be enough for an imbalanced classification problem.

# 8. When Accuracy Can Be Misleading

| Situation | Why accuracy may be insufficient |
|---|---|
| Balanced classes | Accuracy can be a useful overall measure |
| Highly imbalanced classes | Majority-class predictions can dominate the score |
| Rare disease detection | Missing positive cases may be more important than total correctness |
| Spam detection | False positives can cause legitimate messages to be treated as spam |

This motivates metrics such as **precision** and **recall**.

# 9. Precision

**Precision** focuses on the predictions that the model classified as positive.

It answers:

> **Of all cases predicted as positive, how many were actually positive?**

Formula:

$$
\boxed{
Precision =
\frac{TP}{TP+FP}
}
$$

Here:

- $TP$ = correct positive predictions
- $FP$ = incorrect positive predictions

So precision is strongly affected by **false positives**.

# 10. Precision — Interpretation

```text
                  All Positive Predictions
                         │
             ┌───────────┴───────────┐
             ↓                       ↓
            TP                       FP
     Actually positive        Actually negative
             │                       │
             └───────────┬───────────┘
                         ↓
                     Precision
```

High precision means:

> When the model says **positive**, it is usually correct.

Low precision means:

> The model produces many false positive predictions.

# 11. Precision Example

Suppose the model predicts 100 cases as positive.

If 90 of those are actually positive:

$$
Precision = \frac{90}{100}
$$

$$
\boxed{Precision=90\%}
$$

The remaining 10 are false positives.

This shows why precision is concerned with the **quality of positive predictions**.

# 12. Precision — Practical Example

Consider **spam email detection**.

Suppose the model predicts an email as spam.

If it is actually a legitimate email, that prediction is a **false positive**.

A high false-positive rate can be undesirable because legitimate emails may be incorrectly classified as spam.

Therefore, in such a situation, reducing false positives is important, which makes **precision** an important metric.

```text
Email
  │
  ↓
Model predicts SPAM
  │
  ├── Actually spam       → TP
  │
  └── Actually legitimate → FP
                              ↑
                         False alarm
```

# 13. Recall

**Recall** focuses on the actual positive cases.

It answers:

> **Of all the cases that were actually positive, how many did the model correctly identify?**

Formula:

$$
\boxed{
Recall =
\frac{TP}{TP+FN}
}
$$

Here:

- $TP$ = positive cases correctly detected
- $FN$ = positive cases missed by the model

Therefore, recall is strongly affected by **false negatives**.

# 14. Recall — Interpretation

```text
                  All Actual Positive Cases
                           │
              ┌────────────┴────────────┐
              ↓                         ↓
             TP                         FN
       Correctly found              Missed cases
              │                         │
              └────────────┬────────────┘
                           ↓
                         Recall
```

High recall means:

> The model finds most of the actual positive cases.

Low recall means:

> The model misses many actual positive cases.

# 15. Recall — Medical Detection Example

The notes use a medical-diagnosis style example.

Suppose the task is to detect whether a person has a disease.

A **false negative** means:

```text
Actual:     Disease
Prediction: No disease
              ↓
             FN
```

This can be particularly serious because a real positive case has been missed.

Therefore, for applications where missing a positive case is dangerous, **reducing false negatives and achieving high recall can be important**.

# 16. Precision vs. Recall

| Metric | Main question | Main error of concern |
|---|---|---|
| **Precision** | Of predicted positives, how many are actually positive? | **False Positive (FP)** |
| **Recall** | Of actual positives, how many were detected? | **False Negative (FN)** |

### Memory trick

```text
PRECISION → "Can I trust my positive prediction?"

RECALL    → "Did I find the positive cases?"
```

Another shortcut:

> **Precision cares about FP.**

> **Recall cares about FN.**

# 17. Precision and Recall — Side-by-Side

```text
PRECISION
─────────
Start with what the model predicted as POSITIVE.

              TP
Precision = ───────
             TP+FP


RECALL
──────
Start with what was actually POSITIVE.

              TP
Recall = ─────────
             TP+FN
```

Both metrics use **TP**, but their denominators are different.

# 18. F1 Score

Sometimes we want a single metric that combines **precision and recall**.

The F1 score is the harmonic mean of precision and recall.

$$
\boxed{
F_1 =
\frac{2PR}{P+R}
}
$$

where:

- $P$ = Precision
- $R$ = Recall

A high F1 score generally requires both precision and recall to be reasonably high.

# 19. Fβ Score

The more general form is the **Fβ score**:

$$
\boxed{
F_{\beta}
=
(1+\beta^2)
\frac{PR}
{\beta^2P+R}
}
$$

The parameter $\beta$ controls the relative importance of recall and precision.

### Important cases

| β value | Emphasis |
|---|---|
| **β = 1** | Balanced precision and recall → F1 |
| **β > 1** | Greater emphasis on recall |
| **β < 1** | Greater emphasis on precision |

For β = 1:

$$
F_1
=
2\frac{PR}{P+R}
$$

# 20. Why Use the F-score?

Precision and recall can move in opposite directions.

A model may have:

- high precision but low recall, or
- high recall but low precision.

The F-score provides a combined measure.

```text
                Precision
                    │
                    │
                    ├──────────┐
                    │          │
                    ↓          ↓
                 F-score ←── Recall
```

For the standard F1 score, both precision and recall matter equally.

# 21. Complete Metric Comparison

| Metric | Formula | Focus | Strongly affected by |
|---|---|---|---|
| **Accuracy** | $\frac{TP+TN}{TP+TN+FP+FN}$ | Overall correctness | Class imbalance can make it misleading |
| **Precision** | $\frac{TP}{TP+FP}$ | Quality of positive predictions | False positives |
| **Recall** | $\frac{TP}{TP+FN}$ | Coverage of actual positives | False negatives |
| **F1** | $\frac{2PR}{P+R}$ | Balance of precision and recall | Both FP and FN |
| **Fβ** | $(1+\beta^2)\frac{PR}{\beta^2P+R}$ | Adjustable precision/recall balance | Depends on β |

# 22. Choosing the Right Metric

```text
                    What matters most?
                           │
          ┌────────────────┼────────────────┐
          ↓                ↓                ↓
   Overall correctness   Positive           Finding
                         prediction         positives
          │                │                │
          ↓                ↓                ↓
      Accuracy         Precision           Recall
                           │                │
                           └──────┬─────────┘
                                  ↓
                         Need one combined
                              measure?
                                  │
                                  ↓
                              F1 / Fβ
```

### Practical guide

| Goal | Useful metric |
|---|---|
| Overall percentage of correct predictions | Accuracy |
| Avoid false alarms | Precision |
| Avoid missing positive cases | Recall |
| Balance precision and recall | F1 |
| Give recall or precision extra importance | Fβ |

# 27. Common Interview / Exam Questions

### 1. What is a confusion matrix?

A table that compares predicted classes with actual classes and produces TP, TN, FP and FN.

### 2. What is accuracy?

The ratio of correct predictions to total predictions.

### 3. What is precision?

The proportion of predicted positives that are actually positive.

### 4. What is recall?

The proportion of actual positives that are correctly identified.

### 5. What is a false positive?

The model predicts positive when the actual class is negative.

### 6. What is a false negative?

The model predicts negative when the actual class is positive.

### 7. Why can accuracy be misleading?

Especially with imbalanced data, a model can obtain high accuracy by mostly predicting the majority class.

### 8. When is precision important?

When false positive predictions are costly or undesirable.

### 9. When is recall important?

When missing an actual positive case is costly or dangerous.

### 10. What is F1?

The harmonic mean of precision and recall.

### 11. What does β control in Fβ?

It controls the relative emphasis placed on recall versus precision.

````python
import numpy as np
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    fbeta_score
)

# Example binary-classification labels
y_actual = np.array([1, 1, 1, 0, 0, 0, 1, 0])
y_pred   = np.array([1, 1, 0, 0, 0, 1, 1, 0])

cm = confusion_matrix(y_actual, y_pred)

print("Confusion Matrix:")
print(cm)

print("Accuracy :", accuracy_score(y_actual, y_pred))
print("Precision:", precision_score(y_actual, y_pred))
print("Recall   :", recall_score(y_actual, y_pred))
print("F1 Score :", f1_score(y_actual, y_pred))
print("F2 Score :", fbeta_score(y_actual, y_pred, beta=2))
````

````python
# Manual calculation from the confusion matrix

tn, fp, fn, tp = cm.ravel()

accuracy = (tp + tn) / (tp + tn + fp + fn)
precision = tp / (tp + fp) if (tp + fp) else 0
recall = tp / (tp + fn) if (tp + fn) else 0
f1 = (2 * precision * recall / (precision + recall)
      if (precision + recall) else 0)

print("TN =", tn)
print("FP =", fp)
print("FN =", fn)
print("TP =", tp)

print("\nManual metrics")
print("Accuracy :", accuracy)
print("Precision:", precision)
print("Recall   :", recall)
print("F1       :", f1)
````
