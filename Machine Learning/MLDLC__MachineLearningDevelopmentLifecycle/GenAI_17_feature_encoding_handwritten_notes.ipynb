# 8. Label Encoding

## Source note

The handwritten note says:

````python
# Simple demonstration of label encoding for a categorical target.
# This is an executable example illustrating the source concept.

from sklearn.preprocessing import LabelEncoder

y = ["Yes", "No", "Yes", "No", "Yes"]

encoder = LabelEncoder()
y_encoded = encoder.fit_transform(y)

print("Original:", y)
print("Encoded :", y_encoded)
print("Classes :", encoder.classes_)
````

# 9. One-Hot Encoding — Source Example

The handwritten example contains categorical columns such as:

```text
name
education
state
job
```

For a **nominal** feature, the notes recommend:

```text
Nominal data
     ↓
One-Hot Encoder
```

### Why?

There is no meaningful numerical order between nominal categories.

For example, if the categories were:

```text
Hyderabad
Delhi
Mumbai
```

we should not interpret:

```text
Delhi = 1
Mumbai = 2
Hyderabad = 3
```

as if one city were mathematically greater than another. One-hot encoding instead represents each category using indicator columns.

````python
# Demonstration of one-hot encoding.
# Example categories are used only to make the source concept executable.

from sklearn.preprocessing import OneHotEncoder

cities = [["Hyderabad"], ["Delhi"], ["Mumbai"], ["Delhi"]]

encoder = OneHotEncoder(sparse_output=False, handle_unknown="ignore")
encoded = encoder.fit_transform(cities)

print("Categories:", encoder.categories_)
print("Encoded:")
print(encoded)
````

# 10. Ordinal Encoding — Source Example

For categories with a meaningful order, the notes recommend:

```text
Ordinal data
     ↓
Ordinal Encoder
```

### Feedback example

```text
Bad
Average
Good
Very Good
Excellent
```

The categories can be represented in an ordered numerical form.

> The handwritten notes do not specify an exact numerical mapping for these five feedback categories, so this notebook does not invent one as the source mapping.

````python
# Demonstration of ordinal encoding.
# The order below is explicitly supplied rather than learned automatically.

from sklearn.preprocessing import OrdinalEncoder

feedback = [["bad"], ["average"], ["good"], ["very good"], ["excellent"]]

ordered_categories = [["bad", "average", "good", "very good", "excellent"]]

encoder = OrdinalEncoder(categories=ordered_categories)
encoded = encoder.fit_transform(feedback)

print("Original:", [x[0] for x in feedback])
print("Encoded :", encoded.ravel())
````

# 14. Ordinal Encoder — Explicit Numerical Ordering

## Notes from March 11

The page gives a clear ordered example:

Tertiary > Secondary > Primary > Unknown
   3          2          1          0

So the source shows:

| Category | Numerical value |
|---|---:|

This is an example of **Ordinal Encoding**, because the numerical representation preserves an explicitly defined category order.

````python
# Reproducing the explicit ordinal mapping shown in the notes.

education_level = {
    "Unknown": 0,
    "Primary": 1,
    "Secondary": 2,
    "Tertiary": 3,
}

for category, value in education_level.items():
    print(category, "->", value)
````

# 15. Ordinal Encoder in Scikit-Learn

The March 11 page explicitly connects:

```text
Ordinal Encoder
        ↓
Scikit-learn
```

and gives the ordered-category idea.

A practical scikit-learn implementation is:

```text
ordered categories
       ↓
OrdinalEncoder
       ↓
numerical ordered values
```

The important requirement is that the category order must be supplied correctly when the order is meaningful.

````python
from sklearn.preprocessing import OrdinalEncoder

levels = [["Unknown", "Primary", "Secondary", "Tertiary"]]

data = [["Primary"], ["Secondary"], ["Tertiary"], ["Unknown"]]

encoder = OrdinalEncoder(categories=levels)
encoded = encoder.fit_transform(data)

for original, value in zip(data, encoded.ravel()):
    print(f"{original[0]:10s} -> {int(value)}")
````
