# 2. Why Not Simply Assign Numbers?

## Original example

The notes use a `Color` column and a `Target` column:

| Color | Target |
|---|---:|
| Yellow | 0 |
| Blue | 2 |
| Red | 1 |

The handwritten note points out that there is **no relationship / order** between the colors.


The point is:

> **Color is nominal data.**

The numerical labels `0`, `1`, and `2` should therefore not be interpreted as meaningful rankings of the colors.

# 3. One-Hot Encoder

## Original notes

For nominal data, the notes use:

```text
One-Hot Encoder
```

The example converts the three colors into vectors:

```text
Yellow → [0, 0, 1]
Blue   → [0, 1, 0]
Red    → [1, 0, 0]
```

The notes call these **encoded vectors** and state that this is done:

> to represent this data in proper numerical format.

# 4. One-Hot Encoding as Dummy Variables

The encoded representation can be written as three dummy-variable columns:

| Yellow | Blue | Red |
|---:|---:|---:|
| 0 | 0 | 1 |
| 0 | 1 | 0 |
| 1 | 0 | 0 |

The handwritten note labels these as **dummy variables**.

For each observation, exactly one category column contains `1`, while the remaining category columns contain `0`.

```text
Yellow → 0 0 1
Blue   → 0 1 0
Red    → 1 0 0
```

````python
# Basic one-hot encoding using scikit-learn

from sklearn.preprocessing import OneHotEncoder

colors = [["Yellow"], ["Blue"], ["Red"]]

encoder = OneHotEncoder(sparse_output=False)
encoded = encoder.fit_transform(colors)

print("Categories:", encoder.categories_)
print("Encoded vectors:")
print(encoded)
````

# 5. Important Property of the Encoded Columns

## Original notes

The notes observe that the dummy-variable columns have a relationship.

For the three one-hot columns:

```text
Yellow + Blue + Red = 1
```

for every row.

This means the columns are **dependent** on one another.

The notes describe this as:

> There is a relationship between all vectors.

and then connect this observation to **dependency among each column**.

# 6. Dependency Among Columns

The handwritten reasoning can be represented as:

```text
One-hot columns
      ↓
Each row contains exactly one 1
      ↓
Column values are related
      ↓
Their sum is always 1
      ↓
Dependency among columns
```

For example:

```text
[0, 0, 1] → sum = 1
[0, 1, 0] → sum = 1
[1, 0, 0] → sum = 1
```

The note explicitly records:

```text
sum → 1
always sum = 1
```

````python
# Verify the "sum = 1" property from the notes

encoded_vectors = [
    [0, 0, 1],
    [0, 1, 0],
    [1, 0, 0]
]

for vector in encoded_vectors:
    print(vector, "-> sum =", sum(vector))
````

# 8. Number of Columns vs Number of Categories

## Original notes

The handwritten relationship is:

```text
Instead of:

Columns ↔ No. of categories

do:

Columns → No. of categories - 1
```

So, if a categorical feature contains:

```text
n categories
```

the notes recommend keeping:

```text
n - 1 columns
```

instead of all `n` one-hot columns.

### Example

For three categories:

```text
Number of categories = 3

Full one-hot:
3 columns

Reduced representation:
3 - 1 = 2 columns
```

# 10. Why Two Columns Are Enough for Three Categories

## Additional explanation

This section makes the handwritten `n - 1` rule easier to see.

If the three categories are:

```text
Yellow, Blue, Red
```

and we keep only:

```text
Yellow, Blue
```

then the omitted category **Red** can be inferred:

```text
Yellow = 0 and Blue = 0
        ↓
Red = 1
```

Thus the third column does not provide independent information.

Mathematically, with three full one-hot columns:

$$
Yellow + Blue + Red = 1
$$

so:

$$
Red = 1 - Yellow - Blue
$$

Therefore, one of the three columns can be removed without losing the categorical information.

# 11. Reduced One-Hot Representation

The reduced vectors written in the notes are:

```text
Full representation       Reduced representation

Yellow → [0, 0, 1]         Yellow → [0, 0]
Blue   → [0, 1, 0]   →     Blue   → [0, 1]
Red    → [1, 0, 0]         Red    → [1, 0]
```

The important source conclusion is:

```text
n categories
     ↓
n - 1 encoded columns
     ↓
No dependency among the retained columns
     ↓
Multicollinearity eliminated
```

````python
# Reduced one-hot encoding with scikit-learn
# drop="first" removes one dummy column.

from sklearn.preprocessing import OneHotEncoder

colors = [["Yellow"], ["Blue"], ["Red"]]

encoder_reduced = OneHotEncoder(
    drop="first",
    sparse_output=False
)

reduced = encoder_reduced.fit_transform(colors)

print("Categories:", encoder_reduced.categories_)
print("Reduced encoded vectors:")
print(reduced)
````

# 18. Key Formula from the Notes

For a categorical feature with `n` categories:

$$
\boxed{\text{Number of encoded columns} = n - 1}
$$

when one dummy/category column is intentionally dropped to avoid the dependency created by the full one-hot representation.

For the color example:

$$
n = 3
$$

therefore:

$$
n - 1 = 3 - 1 = 2

encoded columns are retained.

# 19. Important Points

- **Nominal data** contains categories with no meaningful order.
- The notes use **Color** as the main nominal example.
- Simply mapping colors to `0`, `1`, and `2` can create a false sense of ordering.
- **One-Hot Encoding** converts each category into an encoded numerical vector.
- In the full one-hot representation, the dummy-variable columns have a dependency.
- For every row, the one-hot values sum to `1`.
- The notes connect this dependency with **multicollinearity**.
- The notes state that multicollinearity should be eliminated.
- Instead of `n` one-hot columns, the notes recommend:

$$
n - 1
$$

columns.
- The notes also show a **Marital** example represented by vectors.
- A categorical variable with many distinct categories can produce many encoded columns.
