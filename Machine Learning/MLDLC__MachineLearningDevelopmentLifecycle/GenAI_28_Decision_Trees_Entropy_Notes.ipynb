# 1. Decision Tree

A **Decision Tree** is a supervised-learning model that makes predictions by repeatedly applying feature-based decisions.

The notes describe it as a **flowchart-like structure** used to make decisions from features.

```text
                         Root Node
                            │
                    ┌───────┴───────┐
                    │ Feature test  │
                    └───────┬───────┘
                       Yes   │   No
                         ┌───┴───┐
                         ↓       ↓
                    Decision   Decision
                       Node       Node
                       / \        / \
                      /   \      /   \
                   Leaf  Leaf  Leaf  Leaf
```

### Core idea

> Start from the root, evaluate a condition, follow the corresponding branch, and continue until reaching a leaf.

# 2. Decision Tree as Supervised Learning

Decision Trees are supervised-learning algorithms.

The training data contains:

- **Features** — input variables
- **Target** — the value/class to predict

For classification:

```text
Features
   │
   ↓
Decision rules
   │
   ├── condition 1 ──→ branch
   ├── condition 2 ──→ branch
   └── ...
            │
            ↓
       Predicted class
```

The supplied notes focus primarily on **classification trees**.

# 3. Main Components of a Decision Tree

| Component | Meaning |
|---|---|
| **Root Node** | Starting point of the tree; represents the first major split |
| **Decision Node** | Internal node where a feature condition is evaluated |
| **Branch** | Outcome/path created by a decision |
| **Leaf Node** | Final node containing the prediction |
| **Feature** | Variable used to make a split |
| **Threshold** | Numeric cut-off used for a continuous feature |

### Example

```text
                     Age ≥ 15?
                    /         \
                  Yes          No
                  /             \
             School?          Not school
             /    \
           Yes     No
           │        │
        Class A   Class B
```

The important point is that every observation follows a path from the root to a leaf.

# 4. Decision Tree Traversal

Prediction is performed by **traversing** the tree.

```text
Input sample
     │
     ↓
[Root condition]
     │
 ┌───┴───┐
Yes      No
│         │
↓         ↓
[Next]   [Next]
│         │
└────┬────┘
     ↓
   Leaf
     │
     ↓
 Prediction
```

For a new sample:

1. Start at the root.
2. Evaluate the condition.
3. Choose the matching branch.
4. Repeat until a leaf is reached.
5. Return the leaf's prediction.

# 7. Why Do We Split the Dataset?

A good decision-tree split should make the resulting groups more homogeneous.

Before splitting:

```text
Parent
┌─────────────────────────┐
│  +   -   +   -   +   - │
│  -   +   -   +   -   + │
└─────────────────────────┘
       mixed classes
```

After a good split:

```text
Child A                  Child B
┌──────────────┐         ┌──────────────┐
│ + + + + +    │         │ - - - - -    │
│ + + + +      │         │ - - - -      │
└──────────────┘         └──────────────┘
   mostly +                 mostly -
```

This reduction in uncertainty/impurity is what the splitting criteria try to measure.

# 8. Overfitting in Decision Trees

A tree can keep splitting until it becomes extremely complex.

```text
Simple tree                     Very complex tree

       Root                         Root
      /    \                       /    \
     A      B                     A      B
                                  / \    / \
                                 C   D  E   F
                                / \ / \ / \ / \
                               ... ... ... ...
```

A very deep tree can memorize training examples instead of learning general patterns.

### Typical symptoms

- Very good training performance
- Worse performance on unseen data
- Too many branches
- Very complex decision rules

The handwritten notes connect excessive branching with **overfitting**.

# 10. Entropy

**Entropy** measures the impurity or uncertainty in a set of observations.

For a binary classification problem:

$$
\boxed{
H(S)
=
-p_+\log_2(p_+)
-p_-\log_2(p_-)
}
$$

where:

- $p_+$ = proportion of positive examples
- $p_-$ = proportion of negative examples
- $p_+ + p_- = 1$

### Intuition

```text
Pure dataset
+ + + + + + +       → Entropy = 0

Mixed dataset
+ - + - + - + -     → higher entropy

Perfectly balanced binary set
+ + + +
- - - -             → maximum entropy
```

# 11. Entropy Range for Binary Classification

For a binary target:

$$
0 \le H(S) \le 1
$$

### Entropy = 0

The set contains only one class.

```text
+ + + + + +

or

- - - - - -
```

There is no uncertainty.

### Entropy = 1

The two classes are equally represented.

```text
+ + + +
- - - -
```

This is maximum uncertainty for a binary classification problem.

```text
Entropy
1.0 |              /\
    |             /  \
    |            /    \
    |           /      \
0.0 |__________/        \__________
       0       0.5        1
             p(+)
```

The handwritten graph emphasizes the maximum around a balanced class distribution and zero entropy for a pure node.

# 12. Entropy Example

Suppose a dataset contains:

- 3 positive samples
- 1 negative sample

Then:

$$
p_+ = \frac{3}{4}
$$

$$
p_- = \frac{1}{4}
$$

Therefore:

$$
H(S)
=
-\frac34\log_2\left(\frac34\right)
-\frac14\log_2\left(\frac14\right)
$$

Approximately:

$$
\boxed{H(S)\approx0.81}
$$

This matches the worked entropy calculation shown in the notes.

# 13. Entropy Intuition Table

| Positive proportion | Negative proportion | Entropy intuition |
|---:|---:|---|
| 1.0 | 0.0 | 0 — completely pure |
| 0.75 | 0.25 | ≈ 0.81 |
| 0.50 | 0.50 | 1 — maximum binary entropy |
| 0.25 | 0.75 | ≈ 0.81 |
| 0.0 | 1.0 | 0 — completely pure |

### Memory trick

> **Pure → low entropy. Mixed → high entropy.**

# 14. Information Gain

A decision tree uses a split criterion to decide which feature/threshold should be selected.

The notes introduce **Information Gain (IG)** using entropy.

The core idea is:

$$
\boxed{
IG(S,A)
=
H(S)
-
\sum_v
\frac{|S_v|}{|S|}
H(S_v)
}
$$

where:

- $S$ = original dataset
- $A$ = candidate feature
- $S_v$ = subset produced by one value/branch of the feature
- $H(S)$ = entropy before the split
- $H(S_v)$ = entropy after the split

### Intuition

```text
Entropy before split
        │
        ↓
    Candidate split
        │
   ┌────┴────┐
   ↓         ↓
Child 1    Child 2
   │         │
   └────┬────┘
        ↓
Weighted child entropy

Information Gain
= Before − After
```

A larger information gain means a greater reduction in entropy.

# 15. Information Gain — Step-by-Step

For each candidate feature:

1. Calculate the parent entropy.
2. Split the dataset.
3. Calculate entropy for every child.
4. Calculate weighted child entropy.
5. Subtract the weighted child entropy from parent entropy.
6. Prefer the split with higher information gain.

$$
H_{after}
=
\sum_i
\frac{|S_i|}{|S|}
H(S_i)
$$

$$
IG
=
H(parent)-H_{after}
$$

```text
Parent entropy
      │
      ├── Child 1 entropy
      ├── Child 2 entropy
      └── ...
              │
              ↓
       Weighted average
              │
              ↓
       Information Gain
```

# 16. Information Gain — Worked Structure

Suppose a parent node contains a mixture of classes.

```text
              Parent
             /      \
          Child A   Child B
```

The calculation is:

```text
Parent entropy
      │
      ├── calculate Child A entropy
      │
      ├── calculate Child B entropy
      │
      └── calculate weighted child entropy
                    │
                    ↓
              Information Gain
```

The handwritten examples use this principle: compare the impurity before and after a candidate split.

# 17. Why Information Gain Is Useful

The purpose of Information Gain is to identify a split that reduces uncertainty.

```text
High entropy
    │
    │ good split
    ↓
Lower entropy
    │
    ↓
More homogeneous child nodes
```

Therefore:

> **Good split → large reduction in entropy → high information gain.**

# 18. Gini Impurity

The notes also introduce **Gini Impurity** as another way to measure impurity.

For a classification node:

$$
\boxed{
Gini(S)
=
1-\sum_i p_i^2
}
$$

For a binary problem:

$$
\boxed{
Gini(S)
=
1-(p_+^2+p_-^2)
}
$$

### Interpretation

- Gini = 0 → completely pure node
- Higher Gini → greater class mixing

# 19. Gini Example

For a node containing equal proportions of two classes:

$$
p_+=\frac12,\qquad p_-=\frac12
$$

Then:

$$
Gini
=
1-
\left[
\left(\frac12\right)^2+
\left(\frac12\right)^2
\right]
$$

$$
=
1-\left(\frac14+\frac14\right)
$$

$$
\boxed{Gini=0.5}
$$

This illustrates the maximum impurity for a binary node.

# 20. Entropy vs Gini Impurity

Both are used to measure node impurity.

| Property | Entropy | Gini Impurity |
|---|---|---|
| Measures | Uncertainty / impurity | Impurity |
| Pure node | 0 | 0 |
| Binary maximum | 1 | 0.5 |
| Formula | $-\sum p_i\log_2 p_i$ | $1-\sum p_i^2$ |
| Common use | Information-gain based splitting | Gini-based tree splitting |

### Intuition

```text
             Node impurity
                  │
          ┌───────┴───────┐
          ↓               ↓
       Entropy            Gini
          │               │
     uncertainty       impurity
          │               │
     information       squared
        gain          probabilities
```

# 21. Choosing a Decision-Tree Split

A decision tree evaluates possible feature splits.

```text
                    Dataset
                       │
          ┌────────────┼────────────┐
          ↓            ↓            ↓
       Feature A    Feature B    Feature C
          │            │            │
      calculate    calculate    calculate
       impurity     impurity     impurity
          │            │            │
          └────────────┼────────────┘
                       ↓
                 Best split
                       │
                       ↓
                 Create child nodes
```

With entropy:

> Prefer the split that gives the greatest information gain.

With Gini:

> Prefer the split that produces the lowest impurity after splitting.

# 23. Decision-Tree Stopping Intuition

A tree should not necessarily continue splitting forever.

Possible stopping conditions include:

- Node becomes sufficiently pure
- Maximum tree depth reached
- Too few samples remain
- Further splitting provides little useful improvement

The supplied notes particularly emphasize **pruning** as a way to control unnecessary tree complexity.

# 27. Random Forest

The notes then introduce **ensemble learning** and **Random Forest**.

A Random Forest combines multiple decision trees.

```text
                 Training Data
                      │
          ┌───────────┼───────────┐
          ↓           ↓           ↓
       Tree 1      Tree 2      Tree 3
          │           │           │

          └───────────┼───────────┘

                      │
```

### Core idea

> Instead of relying on one decision tree, combine multiple trees to obtain a stronger ensemble model.

# 28. Why Random Forest?

A single Decision Tree can overfit.

```text
Single tree
     │
     ↓
Very complex rules
     │
     ↓
Possible overfitting
```

An ensemble approach combines multiple trees:

```text
Tree 1 ─┐
Tree 2 ─┤
Tree 3 ─┤
Tree 4 ─┤──→ Combined prediction
Tree 5 ─┘
```

The objective is to improve robustness and generalization compared with relying on one tree.

# 29. Decision Tree vs Random Forest

| Aspect | Decision Tree | Random Forest |
|---|---|---|
| Model structure | One tree | Multiple trees |
| Complexity | Easier to visualize | More complex internally |
| Interpretability | High | Lower |
| Overfitting tendency | Can be high | Usually more robust |
| Prediction | One tree's output | Combined tree predictions |
| Main idea | Single set of rules | Ensemble of trees |

### Memory trick

> **Decision Tree = one expert.**  
> **Random Forest = many trees voting/combining.**

## One-line memory

> **A Decision Tree recursively chooses useful feature splits to create purer child nodes; entropy/information gain or Gini can guide the splits, pruning controls complexity, and Random Forest combines many trees for a more robust ensemble.**

````python
import math

def entropy_binary(p_positive):
    # Binary entropy for p(+) = p_positive.
    p = p_positive
    q = 1 - p

    value = 0.0
    if p > 0:
        value -= p * math.log2(p)
    if q > 0:
        value -= q * math.log2(q)

    return value


def gini_binary(p_positive):
    # Binary Gini impurity.
    p = p_positive
    q = 1 - p
    return 1 - (p**2 + q**2)


print("Entropy for p(+)=3/4:", entropy_binary(3/4))
print("Gini for p(+)=1/2:", gini_binary(1/2))
````
