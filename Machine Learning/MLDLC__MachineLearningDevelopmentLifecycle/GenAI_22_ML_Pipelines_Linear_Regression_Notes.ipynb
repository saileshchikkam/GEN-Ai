## 1. Machine-Learning Pipeline

A machine-learning pipeline is a sequence of preprocessing and modelling steps that can be executed together.

The notes describe the pipeline workflow as:

```text
Data
  ↓
Separate data into multiple categories
  ↓
Create separate preprocessing pipelines / transformers
  ↓
Apply the corresponding transformer to each separate part
  ↓
Create the final model pipeline
  ↓
Train/Test split
  ↓
Feed training data to the model
  ↓
Make predictions
```

### Why use a pipeline?

A pipeline helps keep the preprocessing steps connected and makes the overall workflow systematic.

Instead of manually performing every operation separately, the transformations can be chained together.


## 2. Pipeline Steps

| Step | What happens | Purpose |
|---|---|---|
| 1 | Separate data into categories | Identify different feature types |
| 2 | Create separate pipelines / transformers | Give each feature type the correct preprocessing |
| 3 | Apply the transformers | Transform each part of the data |
| 4 | Build the model pipeline | Connect transformed features to the algorithm |
| 5 | Train/test split | Separate training and testing data |
| 6 | Feed data to the model | Train the model and make predictions |

## 3. Common Transformers Mentioned

The notes refer to preprocessing operations such as:

- `SimpleImputer`
- `StandardScaler`
- Encoding transformers
- `ColumnTransformer`

A typical numerical pipeline can therefore look like:

```text
Numerical data
      ↓
SimpleImputer
      ↓
StandardScaler
      ↓
Transformed numerical features
```

For categorical data:

```text
Categorical data
      ↓
Imputation
      ↓
Encoding
      ↓
Transformed categorical features
```


````python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

numeric_pipeline
````

## 4. ColumnTransformer

When a dataset contains different types of columns, each group can be processed using its own transformer.

```text
                         Dataset
                            │
             ┌──────────────┼──────────────┐
             ↓              ↓              ↓
         Numerical        Ordinal        Nominal
             │              │              │
             ↓              ↓              ↓
        Numeric pipe   Ordinal pipe   Categorical pipe
             │              │              │
             └──────────────┼──────────────┘
                            ↓
                   ColumnTransformer
                            ↓
                  Final feature matrix
```

The idea in the notes is to avoid repeatedly separating columns manually. Instead, a `ColumnTransformer` can apply the required transformation to the selected columns.


````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

ordinal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder())
])

nominal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, ["weight", "height"]),
    ("ordinal", ordinal_pipeline, ["education_level"]),
    ("nominal", nominal_pipeline, ["city"])
])

preprocessor
````

# Linear Regression

## 8. What Is Linear Regression?

The notes describe linear regression as a **supervised machine-learning algorithm** used to predict a **continuous value**.

The key assumption is that the relationship between the input and output can be represented approximately by a straight line.

```text
y
│
│             •
│          •
│       •
│    •
│ •
└────────────────── x

        best-fit line
```

The basic form of a straight-line equation is:

$$
y = mx + c
$$

where:

- $x$ = input feature
- $y$ = predicted output
- $m$ = slope
- $c$ = intercept


## 9. Example: Weight and Height

The notes use a small example relating **weight** and **height**.

| Weight | Height |
|---:|---:|
| 35 | 150 |
| 37 | 152 |
| 42 | 158 |
| 47 | 163 |
| ? | ? |

The idea is to use the observed points to estimate a relationship and then predict the missing output for a new input.

Conceptually:

```text
Height
  ↑
  │                         •
  │                    •
  │               •
  │          •
  │      •
  └────────────────────────────→ Weight
```

The notes describe the straight line through the data as the **best-fitted line**.


## 10. Best-Fit Line

A regression line attempts to represent the relationship between the observed input and output values.

The notes describe the objective as finding the line that best fits the observed points.

```text
Observed points

      •
         •
   •
              •
        •
           ╲
            ╲
             ╲  Best-fit line
              ╲
```

The distance between an observed point and the corresponding prediction is related to the model's error.

The handwritten notes also introduce the idea of distance between points:

$$
d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}
$$

This is the Euclidean distance formula.


## 11. Slope and Intercept

For:

$$
y = mx + c
$$

### Slope

The slope describes how much the predicted output changes as the input changes.

Geometrically:

$$
m = \frac{\Delta y}{\Delta x}
$$

```text
        y
        ↑
        │       •
        │      /
        │     /
        │    /
        │   /
        │  /
        └────────────→ x

          slope = rise / run
```

### Intercept

The intercept is the value of $y$ when $x=0$.

```text
y
↑
│   •
│  /
│ /
│/________→ x
↑
c = y-intercept
```

The notes express this through the equation:

```text
y = mx + c
```


## 12. Predicting a New Value

Once the model has learned the parameters $m$ and $c$, a new input can be substituted into:

$$
y = mx + c
$$

For example, if a fitted model is represented as:

$$
y = 4x + 78
$$

then a new value of $x$ can be substituted to obtain the corresponding predicted output.

The important concept is:

```text
New input x
     ↓
Fitted equation y = mx + c
     ↓
Predicted y
```


````python
import numpy as np
from sklearn.linear_model import LinearRegression

X = np.array([[35], [37], [42], [47]])
y = np.array([150, 152, 158, 163])

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
print("Prediction for weight 50:", model.predict([[50]])[0])
````

## 13. Simple vs Multiple Linear Regression

The notes distinguish between two forms.

### Simple Linear Regression

One input feature is used to predict one output.

```text
One input
    ↓
Regression model
    ↓
One output
```

Mathematically:

$$
y = mx + c
$$

### Multiple Linear Regression

Multiple input features are used to predict one output.

```text
x₁ ──┐
x₂ ──┤
x₃ ──┼──→ Regression model ──→ y
x₄ ──┤
x₅ ──┘
```

A common mathematical form is:

$$
y = b_0 + b_1x_1 + b_2x_2 + \cdots + b_nx_n
$$

### Comparison

| Type | Inputs | Output |
|---|---|---|
| Simple linear regression | One feature | One continuous output |
| Multiple linear regression | Multiple features | One continuous output |


## 14. Linear Regression Assumptions

The notes highlight important assumptions for linear regression.

  •
 •

## 15. Distance and the Best-Fit Idea

The notes connect regression with the distance between observed and predicted values.

For two points:

$$
(x_1,y_1),\quad (x_2,y_2)
$$

the Euclidean distance is:

$$
d = \sqrt{(x_2-x_1)^2+(y_2-y_1)^2}
$$

For regression, the important error is commonly considered along the output direction through the difference between an observed value and its predicted value.

```text
Observed y
   •
   │
   │ error
   │
   •  Predicted y
   │
   └────────────→ x
```

The best-fit model is the one that provides the most appropriate overall fit to the observed data according to the chosen loss/error criterion.

## 18. Limitations Mentioned in the Notes

The handwritten notes point to several limitations / considerations:

1. **Sensitivity to outliers**
2. **A linear model is not suitable for strongly non-linear relationships**
3. **The model depends on the linearity assumption**
4. Regression is intended for outputs that vary continuously.

### Outlier illustration

```text
y
│                 ×  ← outlier
│        •
│      •
│    •
│  •
│ •
└──────────────────── x
```

An extreme point can influence the fitted line significantly.

Therefore, the data should be examined before relying on a linear regression model.
