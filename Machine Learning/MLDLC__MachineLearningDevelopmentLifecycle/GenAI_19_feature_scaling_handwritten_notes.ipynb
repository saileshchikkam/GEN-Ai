# 8. Standardization

## Definition

The notes define standardization as:

> It is a type of feature scaling where scaling is done to keep all values in a standard scale range.

The handwritten note also summarizes it as:

```text
all I/P values
```

For the scaled values, the notes state:

# 9. Standardization Formula

The formula written in the notes is:

$$
\boxed{
z=\frac{x-\bar{x}}{\sigma}
}
$$

where:

- $x$ = feature value
- $\bar{x}$ = mean
- $\sigma$ = standard deviation
- $z$ = standardized/scaled value

The notes describe this as:

```text
x - mean
──────────────
standard deviation
```

## Min-Max Scaling Formula

The formula written in the notes is:

$$
\boxed{
x'=\frac{x-\min}{\max-\min}
}
$$

The notes show:

```text
min = 23
max = 68
```

and then demonstrate the calculation beginning with:

$$
\frac{23-23}{68-23}=0
$$

For another value, the handwritten calculation shows:

$$
\frac{53-23}{68-23}
=
\frac{30}{35}
\approx 0.1
$$

The handwritten numerical working is preserved as shown, even where the written table and arithmetic are not fully consistent with one another.

````python
# Min-Max scaling formula from the notes.
# This cell demonstrates the formula independently.

def min_max_scale(x, min_value, max_value):
    return (x - min_value) / (max_value - min_value)

min_value = 23
max_value = 68

for x in [23, 53]:
    print(f"x={x} -> scaled={min_max_scale(x, min_value, max_value)}")
````

# 16. Standardization vs Normalization

| Feature | Standardization | Normalization |
|---|---|---|
| Main idea | Put values on a standard scale | Put values into a fixed normal range |
| Formula | $z=(x-\bar{x})/\sigma$ | $x'=(x-\min)/(\max-\min)$ |
| Target property | Mean = 0, standard deviation = 1 | Values in $[0,1]$ |
| Method mentioned | Standard scaling | MinMaxScaler |
| Outliers | Notes prefer this when outliers are present | Notes indicate it is less preferred when outliers are present |

# 18. Normalization — Additional Clarification

### Additional Explanation

Normalization with Min-Max scaling maps the minimum observed value to `0` and the maximum observed value to `1` when the same minimum and maximum are used for the transformation.

$$
x'=\frac{x-x_{\min}}{x_{\max}-x_{\min}}
$$

For example, if:

$$
x_{\min}=10,\qquad x_{\max}=50
$$

then for $x=30$:

$$
x'=\frac{30-10}{50-10}
=\frac{20}{40}
=0.5
$$

This makes the meaning of the `[0,1]` range explicit.

````python
# Simple Min-Max example for the explanation above.

x_min = 10
x_max = 50
x = 30

scaled_x = (x - x_min) / (x_max - x_min)
print("Scaled value:", scaled_x)
````
