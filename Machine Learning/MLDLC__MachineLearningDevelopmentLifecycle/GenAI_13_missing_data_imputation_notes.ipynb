# 2️⃣ 15. Mean / Median Imputation

## ORIGINAL NOTES

The handwritten notes list:

> **mean**

> **median**

and associate them with:

> **numerical** values.

### Concept

```text
Numerical column
      ↓
Missing value
      ↓
Choose:
   ├── Mean
   └── Median
      ↓
Fill missing value
```

### Additional Explanation

**Mean imputation** replaces a missing numerical value with the column mean.

**Median imputation** replaces it with the column median.

The median is often useful when numerical data contains outliers because it is less sensitive to extreme values than the mean.


````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Age": [20, 21, np.nan, 24, 25],
    "Salary": [30000, 35000, np.nan, 40000, 45000]
})

mean_filled = df.copy()
median_filled = df.copy()

for col in mean_filled.select_dtypes(include="number"):
    mean_filled[col] = mean_filled[col].fillna(mean_filled[col].mean())

for col in median_filled.select_dtypes(include="number"):
    median_filled[col] = median_filled[col].fillna(median_filled[col].median())

print("Mean imputation:")
display(mean_filled)

print("Median imputation:")
display(median_filled)
````

### Concept

### Additional Explanation

The **mode** is the most frequently occurring value.

For categorical variables such as:


the mode can be used as a simple imputation strategy when appropriate.


````python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Department": ["IT", "HR", "IT", np.nan, "IT", "HR"]
})

mode_value = df["Department"].mode()[0]
df["Department"] = df["Department"].fillna(mode_value)

print("Mode used:", mode_value)
display(df)
````

# 📋 22. SimpleImputer Strategies

| `strategy` | Typical use |
|---|---|
| `"mean"` | Numerical data |
| `"median"` | Numerical data |
| `"most_frequent"` | Most frequent value; useful for categorical/discrete data |
| `"constant"` | Fill with a specified constant |

> These strategy names correspond to the handwritten list: **mean, median, most frequent, constant**.


````python
from sklearn.impute import SimpleImputer
import numpy as np

X = np.array([
    [10],
    [20],
    [np.nan],
    [40]
])

imputer = SimpleImputer(strategy="mean")
X_filled = imputer.fit_transform(X)

print("Original:")
print(X)

print("\nAfter mean imputation:")
print(X_filled)
````

# 🔬 25. Complete SimpleImputer Demonstration

The following example brings together the main strategies mentioned in the notes.

### Numerical example

```python
strategy="mean"
strategy="median"
```

### Categorical/discrete example

```python
strategy="most_frequent"
```

### Constant example

```python
strategy="constant"
fill_value=...
```


````python
import numpy as np
from sklearn.impute import SimpleImputer

X = np.array([
    [10, 1],
    [20, 1],
    [np.nan, 2],
    [40, np.nan]
])

strategies = ["mean", "median", "most_frequent"]

for strategy in strategies:
    imputer = SimpleImputer(strategy=strategy)
    result = imputer.fit_transform(X)
    print(f"\nStrategy: {strategy}")
    print(result)

constant_imputer = SimpleImputer(
    strategy="constant",
    fill_value=-1
)

print("\nStrategy: constant, fill_value=-1")
print(constant_imputer.fit_transform(X))
````
