# 1. F-score and Fβ-score

The notes begin with the generalized F-score relationship.

The Fβ-score combines **precision** and **recall**:

$$
\boxed{
F_{\beta}
=
(1+\beta^2)
\frac{Precision \times Recall}
{\beta^2 \times Precision + Recall}
}
$$

For the balanced case, β = 1:

$$
F_1 =
2\frac{Precision \times Recall}
{Precision + Recall}
$$

### Special cases

| β | Emphasis |
|---:|---|
| β = 1 | Precision and recall receive equal importance |
| β > 1 | Recall receives more emphasis |
| β < 1 | Precision receives more emphasis |

```text
                 Precision
                     │
                     │
                     ▼
                  F-score
                     ▲
                     │
                     │
                   Recall

          β controls the balance
```

# 2. Support Vector Machine (SVM)

A **Support Vector Machine** is a supervised machine-learning algorithm used for classification.

The central idea in the handwritten notes is:

> Find a suitable separating hyperplane and use the closest observations to determine the boundary.

For a simple two-class problem:

```text
Class A:     ●  ●  ●       ●
                  │
                  │  margin
                  │
==================│==================  ← separating hyperplane
                  │
                  │
Class B:     ×  ×  ×       ×
```

The objective is not merely to find *any* separating line. The notes emphasize finding a boundary with a useful separation between classes.

# 3. Hyperplane

A **hyperplane** is the decision boundary used to separate classes.

In two dimensions, it is a line.

In higher dimensions, it becomes a higher-dimensional hyperplane.

A simplified equation is:

$$
w^Tx+b=0
$$

where:

- $w$ = weight vector
- $x$ = input feature vector
- $b$ = bias/intercept

The prediction can be represented conceptually by the side of the hyperplane on which the point lies.

```text
                 y
                 ↑
                 │       ● ●
                 │     ●
                 │
                 │  ───────────────  Hyperplane
                 │
                 │       × ×
                 │     ×
                 └──────────────────→ x
```

# 4. Support Vectors

**Support vectors** are the training observations closest to the separating boundary.

They are important because they determine the position of the maximum-margin boundary.

```text
Class +                    Class -

   ●  ●                         ×  ×
      ●                         ×
        ●                   ×
          ●               ×
             ●       ×
                │
                │  boundary
                │
```

The points closest to the boundary are the critical points used by the SVM.

### Key idea

> The decision boundary is strongly influenced by the observations nearest to it.

# 5. Margin

The **margin** is the separation between the decision boundary and the nearest training observations.

A conceptual picture:

```text
Class +       Support       Decision       Support       Class -
                Vector       Boundary        Vector

     ●             ●             │             ×             ×
     ●             ●             │             ×             ×
                   ←── margin ──→
```

SVM attempts to find a separating boundary with a large margin.

### Maximum-margin intuition

```text
Small margin                  Large margin

● ● | × ×                     ● ●     |     × ×
  ● | ×                         ●      |      ×
    |                            ●    |    ×
```

The second boundary provides greater separation between the classes.

# 6. Linear Classification

If the classes can be separated using a straight line, a **linear SVM** can be used.

```text
      ● ● ●
    ● ●
  ●

──────────────────────────
        ×
      × × ×
    × ×
```

The line separates the two classes.

A linear kernel is therefore suitable when the relationship can be separated approximately linearly in the original feature space.

# 7. When a Straight Line Is Not Enough

Some datasets cannot be separated effectively with one straight line.

Example:

```text
             ● ● ●
           ●       ●
          ●    ×    ●
           ●       ●
             ● ● ●
```

A simple linear boundary cannot cleanly separate the inner class from the outer class.

This motivates the use of **non-linear kernels**.

# 8. Kernel Idea

A kernel allows an SVM to model a non-linear decision boundary by operating through a suitable transformed feature representation.

Conceptually:

```text
Original feature space
        │
        │  non-linear relationship
        ↓
Kernel transformation
        │
        ↓
Higher / transformed feature space
        │
        ↓
Linear separation there
```

The important intuition is:

> A problem that is difficult to separate in the original space may become easier to separate after an appropriate feature transformation.

# 9. Common Kernels Mentioned in the Notes

The handwritten notes refer to kernel choices including:

- **Linear**
- **Polynomial**
- **RBF / Radial Basis Function**

| Kernel | Basic idea | Typical use |
|---|---|---|
| Linear | Straight decision boundary | Approximately linearly separable data |
| Polynomial | Models polynomial relationships | Curved boundaries with polynomial structure |
| RBF | Models more flexible local/non-linear relationships | Complex non-linear datasets |

In scikit-learn, these can be selected using the `kernel` argument of `SVC`.

# 10. Linear Kernel

The linear kernel produces a linear decision boundary in the input feature space.


# 13. The `C` Hyperparameter

`C` is an important SVM hyperparameter.

The notes associate `C` with controlling the trade-off between:

- maintaining a wide margin
- allowing classification errors

Conceptually:

```text
                 C
                 │
        ┌────────┴────────┐
        ↓                 ↓
   Wider margin      Fewer training
   / more tolerance   violations
        │                 │
        └────── trade-off ┘
```

A larger `C` generally makes the model place more emphasis on correctly classifying training examples.

A smaller `C` generally allows more violations in exchange for a softer, wider-margin solution.

# 14. Small `C` vs Large `C`

### Smaller C

The model is more tolerant of training errors.

```text
● ●        │        × ×
  ●        │        ×
      ●    │    ×

      wider / softer margin
```

### Larger C

The model tries harder to classify training examples correctly.

```text
● ●   ││   × ×
  ●   ││   ×
      ││
```

This can make the model more sensitive to individual training observations.

> The important study point is the trade-off between **margin size** and **training violations**.

# 15. The `gamma` Hyperparameter

The notes discuss **gamma** particularly with the RBF kernel.

Gamma controls how strongly an individual training point influences the decision boundary.

A useful intuition:

### Low gamma

A point has a broader influence.

```text
Point influence

       . . . . .
    .             .
   .       ●       .
    .             .
       . . . . .
```

The boundary tends to be smoother.

### High gamma

A point has a more localized influence.

```text
       . . .
      .  ●  .
       . . .
```

The boundary can become much more detailed.

# 16. Low Gamma vs High Gamma

| Gamma | Influence of each point | Boundary tendency |
|---|---|---|
| Low | Broader | Smoother / less sensitive |
| High | More localized | More complex / sensitive |

Conceptually:

```text
Low γ                         High γ

Smooth boundary               Highly flexible boundary

   ╭─────────╮                  ╭─╮
 ● │         │ ●              ● │ ╰─╮
   │         │                    ╰─●
 × │         │ ×                × ╭─╯
   ╰─────────╯                    ╰╯
```

A very high gamma can contribute to overfitting.

# 17. C and Gamma Work Together

`C` and `gamma` control different aspects of an SVM model.

| Parameter | Main control |
|---|---|
| **C** | Penalty / tolerance for classification violations |
| **Gamma** | Local influence and complexity of the boundary for kernels such as RBF |

```text
                   SVM Hyperparameters
                           │
                 ┌─────────┴─────────┐
                 ↓                   ↓
                 C                 gamma
                 │                   │
       margin / error trade-off   local influence
                 │                   │
                 └─────────┬─────────┘
                           ↓
                    Model complexity
```

# 19. Underfitting and Overfitting

The notes connect model complexity with the behavior of the decision boundary.

### Underfitting

The model is too simple to capture the structure of the data.

```text
Actual data:       ● ● ●
                 ●       ●
                ●   ×   ●
                 ●     ●

Simple model:
────────────────────────────
```

### Overfitting

The model becomes too closely adapted to the training data.

```text
●─╮
  ╰─╮
    ╰╮
     ╰─╮
×      ╰─╮
```

A highly flexible boundary can fit noise rather than the underlying pattern.

The notes particularly connect excessive model complexity with parameter choices such as high gamma.

# 20. Generalization

The objective is not simply to memorize the training examples.

A useful model should perform well on unseen data.

```text

     ↓

     │
     ↓

```

```text
Too simple       Good fit        Too complex
   │                │                 │
Underfit       Generalizes        Overfit
```

# 26. Simple Hyperparameter Experiment

A useful experiment is to compare several `C` and `gamma` values.

```python
from sklearn.svm import SVC

settings = [
    {"C": 0.1, "gamma": "scale"},
    {"C": 1.0, "gamma": "scale"},
    {"C": 10.0, "gamma": "scale"},
]

models = []

for params in settings:
    model = SVC(
        kernel="rbf",
        C=params["C"],
        gamma=params["gamma"]
    )
    models.append((params, model))
```

The correct values should ultimately be selected using validation or cross-validation rather than simply choosing the most complex model.

# 28. A Useful SVM Mental Model

Think of SVM as solving three connected problems:

### 1. Where should the boundary be?

Find a separating hyperplane.

### 2. How much separation should it maintain?

Maximize the useful margin while controlling violations.

### 3. How complex should the boundary be?

The kernel, `C`, and `gamma` influence the resulting boundary.

```text
             SVM
              │
      ┌───────┼────────┐
      ↓       ↓        ↓
  Boundary   Margin   Complexity
      │       │        │
      │       │      C / gamma /
      │       │      kernel
      └───────┼────────┘
              ↓
        Generalization
```

# 29. Parameter Summary Table

| Parameter / concept | Meaning | Increasing it generally tends to |
|---|---|---|
| `kernel` | Determines the type of feature relationship/boundary | Change the shape/capability of the boundary |
| `C` | Controls penalty for classification violations | Push harder toward training correctness |
| `gamma` | Controls local influence for kernels such as RBF | Make the boundary more localized/complex |
| Margin | Separation around the decision boundary | Larger margin generally encourages simpler separation |
| Support vectors | Critical observations nearest the boundary | Determine the boundary's position |

> The exact behavior depends on the selected kernel and dataset.

# 30. SVM — Exam/Interview Revision

### What is SVM?

A supervised learning algorithm that finds a separating boundary for classification while considering the margin between classes.

### What is a hyperplane?

The decision boundary separating classes.

### What are support vectors?

The training points closest to the decision boundary that strongly determine the margin/boundary.

### What is the margin?

The separation between the decision boundary and the closest observations.

### Why use kernels?

To handle relationships that cannot be separated effectively with a simple linear boundary.

### Name common SVM kernels.

- Linear
- Polynomial
- RBF

### What does `C` control?

The trade-off between margin softness and classification violations.

### What does `gamma` control?

For kernels such as RBF, the local influence of individual observations and therefore boundary complexity.

### What happens with very high gamma?

The model can become excessively localized and may overfit.

### Why is hyperparameter tuning necessary?

Because inappropriate parameter values can lead to underfitting or overfitting and poor generalization.
