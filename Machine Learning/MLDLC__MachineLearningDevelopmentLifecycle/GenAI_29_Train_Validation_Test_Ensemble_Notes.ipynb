# 14. Why Not Use Test Data for Tuning?

A common mistake is:

```text
Train → Test → change model
          ↑
          └── repeat
```

If we repeatedly use test performance to make decisions, the model-development process indirectly adapts to the test set.

Better:

```text
Training → learn
Validation → tune
Test → evaluate once at the end
```

This keeps the test set closer to a genuine unseen-data evaluation.

# 15. Cross-Validation

**Cross-validation** repeatedly divides available data into training and validation portions so that model performance can be evaluated more reliably.

The notes describe the validation portion as being taken from the training data and rotated across folds.

```text
Dataset
  │
  ↓
Split into folds
  │
  ├── Fold 1
  ├── Fold 2
  ├── Fold 3
  ├── Fold 4
  └── Fold 5
```

For each run, a different fold can act as validation data.

# 16. K-Fold Cross-Validation

In **K-fold cross-validation**, the available data is divided into `K` folds.

Example with 5 folds:

```text
Data
┌────┬────┬────┬────┬────┐
│ F1 │ F2 │ F3 │ F4 │ F5 │
└────┴────┴────┴────┴────┘
```

Five validation runs:

```text
Run 1: [VAL] [TRN] [TRN] [TRN] [TRN]
Run 2: [TRN] [VAL] [TRN] [TRN] [TRN]
Run 3: [TRN] [TRN] [VAL] [TRN] [TRN]
Run 4: [TRN] [TRN] [TRN] [VAL] [TRN]
Run 5: [TRN] [TRN] [TRN] [TRN] [VAL]
```

Each fold gets an opportunity to act as validation data.

# 17. Cross-Validation Workflow

```text
             Dataset
                │
                ↓
           Split into K folds
                │
     ┌──────────┼───────────┐
     ↓          ↓           ↓
   Fold 1     Fold 2      ... Fold K
     │          │             │
   Train      Train         Train
   on K-1     on K-1        on K-1
   folds      folds         folds
     │          │             │
   Validate  Validate      Validate
   on F1     on F2         on FK
     │          │             │
     └──────────┼─────────────┘
                ↓
         Collect scores
                ↓
       Calculate average score
                ↓
        Compare configurations
```

The notes show the idea of taking the mean performance across multiple experiments.

# 18. Example of Cross-Validation Scoring

Suppose five validation experiments produce:

```text
Experiment 1 → score₁
Experiment 2 → score₂
Experiment 3 → score₃
Experiment 4 → score₄
Experiment 5 → score₅
```

Mean validation score:

$$
\text{Mean Score}
=
\frac{
score_1+score_2+score_3+score_4+score_5
}{5}
$$

The average gives a summary of model performance across the folds.

# 21. Underfitting vs Overfitting

|---|---|---|---|
| **Underfitting** | Poor | Poor | Model is too simple / has not learned enough |
| **Good fit** | Good | Good | Useful generalization |

```text

     ↓                 ↓                   ↓
```

# 33. High-Value Interview Questions

### What is the difference between training, validation and test data?

- Training data → learn model patterns.
- Validation data → tune hyperparameters and select models.
- Test data → final evaluation on unseen data.

### Why do we need validation data?

To make model-development decisions without repeatedly using the final test set.

### What is hyperparameter tuning?

Searching for a useful set of hyperparameter values based on validation performance.

### What is cross-validation?

A method that repeatedly uses different portions of the available data for validation.

### What is K-fold cross-validation?

Divide data into K folds and rotate the validation fold across K training/validation runs.

### Why use cross-validation?

To obtain a more robust estimate of validation performance across different data partitions.

### What is overfitting?

When a model performs very well on training data but performs substantially worse on unseen data.

### What is ensemble learning?

Combining multiple models to produce a final prediction.

### Why should test data remain unseen?

To preserve it as an unbiased final evaluation set as much as possible.

# 34. Quick Formula / Calculation Reference

### Mean cross-validation score

For K validation scores:

$$
\boxed{
\text{Mean CV Score}
=
\frac{1}{K}
\sum_{i=1}^{K} Score_i
}
$$

Example:

$$
Score_1, Score_2, Score_3, Score_4, Score_5
$$

then:

$$
\text{Mean}
=
\frac{Score_1+Score_2+Score_3+Score_4+Score_5}{5}
$$

This captures the core averaging idea shown in the handwritten cross-validation example.

````python
# Simple illustration of mean cross-validation score

scores = [0.82, 0.85, 0.80, 0.88, 0.84]

mean_score = sum(scores) / len(scores)

print("Fold scores:", scores)
print("Mean CV score:", mean_score)
````
