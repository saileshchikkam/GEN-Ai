##  Pipeline

```python
import pandas as pd
```

```python
df = pd.read_csv('Bank_Marketing_Original_Data.csv')
```

```python
df.head()
```

```python
## first step ---> Seperate data into Multiple Data Categories
```

```python
Numeric_Data = ['Age', 'Balance', 'Day', 'Duration', 'Campaign', 'Pdays', 'Previous']
Nominal_Data = ['Job', 'Marital', 'Education', 'Default', 'Housing', 'Loan', 'contact', 'Month', 'poutcome']
Ordinal_Data = ['Education']
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
from sklearn.preprocessing import StandardScaler
```

```python
## second step ---> Create Pipelines for each data category
Numeric_Pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()) # no need of encoding
    ])
```

```python
Ordinal_Pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OrdinalEncoder(categories=[['unknown','primary', 'secondary', 'tertiary']])),
    ('scaler',StandardScaler())
    ])
```

```python
Nominal_Pipeline = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
    ])
```

```python
## Third step ----> to combine pipeline transformers with their respective Data
```

```python
from sklearn.compose import ColumnTransformer
```

```python
features = ColumnTransformer(transformers=[
    ('Numeric Transformation', Numeric_Pipeline, Numeric_Data),
    ('Ordinal Transformation', Ordinal_Pipeline, Ordinal_Data),
    ('Nominal Transformation', Nominal_Pipeline, Nominal_Data)
])
```

```python
features
```

```python
## Fourth Step ---> to create a model with the the respective transformers and along with their algorithm
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = Pipeline(steps=[
    ('features',features),
    ('classifier',LogisticRegression())
    ])
```

```python
model
```

```python
## Fifth Step ---> train test split my data and encoding of my Output Column data
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
X = df.iloc[:,:-1] ## all rows and all columns except the last column
Y = df['TARGET'] ## the last column is my output column
```

```python
le = LabelEncoder()
```

```python
Y = le.fit_transform(Y) ## TARGET column is encoded to 0 and 1because it is a categorical column
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2)
```

```python
## Final or sixth step ---> Model Training and getting predictions from the model
```

```python
model.fit(X_train, Y_train) ## feeding
```

```python
Y_Predict = model.predict(X_test) ## getting predictions
```

```python
Y_Predict
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(Y_test, Y_Predict)
```

```python
## now changing the algorithm from Logistic Regression to Random Forest Classifier
from sklearn.ensemble import RandomForestClassifier
```

```python
model = Pipeline(steps=[
    ('features',features),
    ('classifier',RandomForestClassifier())
    ])
```

```python
model
```

```python
model.fit(X_train, Y_train) ## feeding
```

```python
Y_Predict = model.predict(X_test)
```

```python
accuracy_score(Y_test, Y_Predict) ## accuracy score of random forest is more than logistic regression so we can say that random forest is better than logistic regression of course
```
