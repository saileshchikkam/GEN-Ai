# 5. New Feature 1 — Family Size

## ORIGINAL NOTES

The first new feature is:

> **Family-Size**

The handwritten formula is:

```text
Family Size = SibSp + Parch + 1
```

where the `1` represents the passenger themself.

The notes connect this new feature with:

> **Passenger + SibSp + Parch**

### Source-preserved idea

Instead of treating `SibSp` and `Parch` independently, create a combined feature representing the passenger's **family size**.

````python
# Feature Engineering: Family Size
# Formula reconstructed from the handwritten notes:
# Family Size = SibSp + Parch + 1

def add_family_size(df):
    df = df.copy()
    df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
    return df

# Example usage:
# df = add_family_size(df)
# df[["SibSp", "Parch", "FamilySize", "Survived"]].head()
````

````python
# Feature Engineering: Family Type
# Categories reconstructed from the handwritten conditions.

def family_type(family_size):
    if family_size < 3:
        return "small family"
    elif family_size < 5:
        return "medium family"
    else:
        return "large family"

# Example:
# df["FamilyType"] = df["FamilySize"].apply(family_type)
# df[["FamilySize", "FamilyType", "Survived"]].head()
````
