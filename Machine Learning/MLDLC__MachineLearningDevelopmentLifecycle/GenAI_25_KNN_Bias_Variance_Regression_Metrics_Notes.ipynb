# 1. K-Nearest Neighbors — Important Points

## Choosing the value of `K`

The value of `K` should not be:

- **Too small** → may cause **overfitting**
- **Too large** → may cause **underfitting**

A suitable value of `K` is normally selected using validation / cross-validation.

```text
                Choosing K
                    │
          ┌─────────┴─────────┐
          ↓                   ↓
      K too small          K too large
          │                   │
          ↓                   ↓
      Overfitting         Underfitting
          │                   │
          ↓                   ↓
  Too sensitive to       Too generalized
     local noise          / local patterns
```

### Practical idea

A good `K` should provide a useful balance between fitting the training data and generalizing to unseen data.


## 2. K-Fold Cross-Validation for Choosing `K`

The notes connect **K-fold cross-validation** with finding a suitable value for `K`.

The general process is:

```text
Training dataset
      │
      ↓
Split into folds
      │
      ↓
Try candidate K values
      │
      ↓
Train / validate repeatedly
      │
      ↓
Compare validation performance
      │
      ↓
Choose a suitable K
```

> **Important:** Here, the `K` in **KNN** and the `K` in **K-fold cross-validation** are different concepts, even though the same letter is commonly used.

# Part II — Model Performance, Bias and Variance

# 5. Model Performance

A model should not be judged only by how well it performs on training data.

We should consider:

- **Training performance**
- **Testing / validation performance**
- The gap between them

```text
Training data ──→ Model ──→ Training performance
                              │
                              ↓
                         Generalization
                              │
                              ↓
Testing data  ───→ Model ──→ Testing performance
```

A model that performs well on training data but poorly on unseen data may be **overfitting**.


# 6. Bias

**Bias** represents error caused by a model being too simple or making overly strong assumptions about the underlying relationship.

The notes associate:

> **Underfitting → High Bias**

A high-bias model may fail to capture important patterns in the data.

### Example

A complex nonlinear relationship is approximated using an overly simple linear model:

```text
Actual pattern:
       •
     •   •
   •       •
 •           •

Too-simple model:
────────────────────
```

The model does not have enough flexibility to represent the actual pattern.


# 7. Variance

**Variance** describes how much a model's learned behavior changes when the training data changes.

The notes associate:

> **Overfitting → High Variance**

A high-variance model can become too sensitive to the particular training examples.

```text
Training data
     ↓
Very flexible model
     ↓
Learns noise / individual points
     ↓
Excellent training performance
     ↓
Poor unseen-data performance
```


# 8. Bias–Variance Summary

| Situation | Bias | Variance | Typical behavior |
|---|---|---|---|
| Underfitting | High | Low | Model too simple |
| Good generalization | Low | Low | Balanced model |
| Overfitting | Low | High | Model too complex / sensitive |

### Memory rule

```text
Underfitting → High Bias
Overfitting  → High Variance
```

A useful model aims for a balance that gives good performance on unseen data.

Typical pattern:

```text
Training error → High
Testing error  → High
```

### Causes noted in the pages

1. Model is too simple
2. Weak / inadequate feature selection
3. Less training time

# 10. Overfitting

Overfitting occurs when the model learns the training data too closely, including noise or accidental patterns.

The notes describe the key idea as:

Typical pattern:

```text

```

### Causes noted in the pages

1. Dataset is too small
2. Too many features
3. Model is too complex

Examples mentioned include highly flexible models such as decision trees.

```text

Generalization becomes poor
```

# Part III — Regression Evaluation

# 16. Regression Model Evaluation

For regression, predictions are continuous numerical values.

Examples include:

- House price
- Sales
- Temperature
- Demand
- Other numerical targets

To evaluate a regression model, compare:

$$
y = \text{actual value}
$$

with

$$
\hat y = \text{predicted value}
$$

The notes introduce:

- **MSE** — Mean Squared Error
- **RMSE** — Root Mean Squared Error
- **MAE** — Mean Absolute Error


# 17. Prediction Error

For an observation:

$$
e_i = y_i-\hat y_i
$$

where:

- $y_i$ = actual value
- $\hat y_i$ = predicted value
- $e_i$ = prediction error / residual

A simple interpretation is:

```text
Actual value
     │
     ├───────────────┐
     ↓               │
Predicted value      │
     │               │
     └── difference ─┘
             ↓
          Error
```

A smaller error generally means the prediction is closer to the actual value.


# 18. Mean Squared Error (MSE)

MSE is the average of the squared prediction errors.

$$
\boxed{
MSE=
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat y_i)^2
}
$$

Where:

- $n$ = number of observations
- $y_i$ = actual value
- $\hat y_i$ = predicted value

### Why square the errors?

Squaring:

- Makes all errors non-negative
- Gives larger errors disproportionately more influence

```text
Error
  │
  ├── +2 → squared → 4
  └── -2 → squared → 4
```

So positive and negative errors do not cancel each other out.


# 19. MSE Example

Suppose:

| Observation | Actual $y$ | Predicted $\hat y$ | Error $y-\hat y$ | Squared error |
|---|---:|---:|---:|---:|
| 1 | $y_1$ | $\hat y_1$ | $e_1$ | $e_1^2$ |
| 2 | $y_2$ | $\hat y_2$ | $e_2$ | $e_2^2$ |
| 3 | $y_3$ | $\hat y_3$ | $e_3$ | $e_3^2$ |

Then:

$$
MSE=\frac{e_1^2+e_2^2+e_3^2}{3}
$$

The notes emphasize that MSE is the **average of all squared errors**.


# 20. Root Mean Squared Error (RMSE)

RMSE is the square root of MSE:

$$
\boxed{
RMSE=
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat y_i)^2
}
}
$$

### Relationship

$$
MSE \rightarrow \sqrt{\phantom{x}} \rightarrow RMSE
$$

RMSE is often easier to interpret because it has the **same units as the target variable**.

For example, if the target is measured in rupees, RMSE is also expressed in rupees.


# 21. Mean Absolute Error (MAE)

MAE is the average absolute prediction error.

$$
\boxed{
MAE=
\frac{1}{n}
\sum_{i=1}^{n}
|y_i-\hat y_i|
}
$$

### Example

If errors are:

$$
2,\;-3,\;4
$$

their absolute errors are:

$$
|2|=2,\quad |-3|=3,\quad |4|=4
$$

Therefore:

$$
MAE=\frac{2+3+4}{3}=3
$$

MAE is less sensitive to large errors than MSE because it does not square the error.


# 22. MSE vs RMSE vs MAE

| Metric | Formula | Main property |
|---|---|---|
| **MSE** | $\frac{1}{n}\sum(y_i-\hat y_i)^2$ | Strongly penalizes large errors |
| **RMSE** | $\sqrt{MSE}$ | Same units as target |
| **MAE** | $\frac{1}{n}\sum|y_i-\hat y_i|$ | Easier, more direct absolute error |

### Sensitivity to outliers

```text
Large error
    │
    ├── MAE  → grows linearly
    │
    └── MSE  → grows quadratically
```

Therefore, a very large error has a much stronger effect on MSE.


# 23. Which Metric Should Be Smaller?

For these error metrics:

> **Lower error generally indicates better predictive performance**, provided the models are evaluated on the same data and target.

```text
Lower MSE  → better
Lower RMSE → better
Lower MAE  → better
```

However, the best metric depends on the application.

If large mistakes are especially costly, MSE/RMSE can be useful because they penalize large errors more strongly.


# 24. Regression Model Evaluation Workflow

```text

        │
        ├───────────────┐
        │               │
        ↓               ↓

                          ↓

```

````python
# Regression metrics with scikit-learn
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error

y_actual = np.array([10, 12, 15, 20])
y_pred   = np.array([11, 11, 14, 23])

mae = mean_absolute_error(y_actual, y_pred)
mse = mean_squared_error(y_actual, y_pred)
rmse = np.sqrt(mse)

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
````

````python
# A simple KNN experiment:
# observe how changing n_neighbors changes predictions

import numpy as np
from sklearn.neighbors import KNeighborsRegressor

X = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 3, 4, 5, 6])

query = np.array([[3.5]])

for k in [1, 3, 5]:
    model = KNeighborsRegressor(n_neighbors=k)
    model.fit(X, y)
    print(f"K={k}: prediction={model.predict(query)[0]:.2f}")
````

**Q8. What does MSE measure?**
