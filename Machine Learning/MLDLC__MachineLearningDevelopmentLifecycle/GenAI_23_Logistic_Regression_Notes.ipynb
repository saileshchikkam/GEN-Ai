# Logistic Regression

## Overview

These notes focus on **logistic regression**, a supervised machine-learning algorithm used when the target represents a **category**, especially a binary outcome such as:

- `0 / 1`
- `No / Yes`
- `Fail / Pass`

The central idea is to start from a linear combination of input features and transform it into a **probability between 0 and 1**.

```text
Input features
      ↓
Linear combination
      ↓
Sigmoid function
      ↓
Probability
      ↓
Decision threshold
      ↓
Predicted class
```


## 1. Logistic Regression vs Linear Regression

The notes distinguish logistic regression from ordinary linear regression based on the type of output.

| Model | Typical output | Example |
|---|---|---|
| **Linear Regression** | Continuous numerical value | Salary, price, height |
| **Logistic Regression** | Category / class probability | Pass or fail |
| **Binary Logistic Regression** | Two classes | 0 / 1 |
| **Multiclass extension** | More than two classes | Class A / B / C |

### Key idea

Linear regression directly predicts a numerical value:

```text
Features → Linear equation → Numerical output
```

Logistic regression converts the linear result into a value between `0` and `1`:

```text
Features → Linear equation → Sigmoid → Probability → Class
```

## 3. Why a Sigmoid Function Is Used

A linear equation can produce any value from negative infinity to positive infinity.

```text
Linear output:
-10, -2, 0, 3, 15, ...
```

That is not directly suitable as a probability because a probability must lie between `0` and `1`.

The sigmoid function solves this by compressing any real-valued input into the range `(0, 1)`.

### Sigmoid equation

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

where:

$$
z=b_0+b_1x_1+b_2x_2+\cdots+b_nx_n
$$

Therefore:

$$
P(y=1\mid X)=\sigma(z)
$$


## 4. Sigmoid Curve

The sigmoid function has an S-shaped curve.

```text
Probability
1.0 ┤                         ●●●
    │                     ●●
    │                  ●●
0.5 ┤──────────────●●──────────────  decision region
    │           ●●
    │        ●●
    │     ●●
0.0 ┤●●●
    └──────────────────────────────→ z
                0
```

Important points:

- Very negative `z` → probability approaches `0`
- `z = 0` → probability is `0.5`
- Very positive `z` → probability approaches `1`

So the sigmoid provides a smooth mapping from the linear model's output to a probability.


## 5. Linear Equation → Sigmoid

The notes show the relationship between a straight-line model and logistic regression.

### Step 1 — Linear part

$$
z=mx+c
$$

### Step 2 — Apply sigmoid

$$
y=\frac{1}{1+e^{-z}}
$$

Therefore:

$$
y=\frac{1}{1+e^{-(mx+c)}}
$$

For multiple features:

$$
z=b_0+b_1x_1+b_2x_2+\cdots+b_nx_n
$$

and:

$$
P(y=1)=\frac{1}{1+e^{-z}}
$$

### Visual flow

```text
x
│
│      Linear model
│       z = mx + c
│             │
│             ↓
│        Sigmoid function
│             │
│             ↓
│       0 ≤ probability ≤ 1
│             │
│             ↓
│       Classification
└────────────────────────
```


# 6. Probability and Classification

The sigmoid output is interpreted as a probability.

For example:

```text
Sigmoid output = 0.80
```

This can be interpreted as a high estimated probability of class `1`.

The handwritten notes connect the probability to a classification decision using a threshold.

A common decision rule is:

| Predicted probability | Predicted class |
|---:|---:|
| `≥ 0.5` | `1` |
| `< 0.5` | `0` |

Conceptually:

```text
                 Probability
                     │
0 ───────────────────┼────────────────── 1
                     │
                    0.5
                     │
             decision threshold
                     │
              ┌──────┴──────┐
              ↓             ↓
           class 0       class 1
```

The threshold can be changed depending on the application.

# 8. Decision Boundary

The notes show a classification graph with a boundary separating two classes.

```text
Class 1
                 ● ● ●
              ● ●
           ●
───────────────┼────────────────→ x
               │
               │ decision boundary
               │
      ○ ○ ○
   ○ ○
Class 0
```

For a simple one-feature logistic model, the decision boundary occurs where:

$$
P(y=1)=0.5
$$

Because:

$$
\sigma(0)=0.5
$$

the boundary corresponds to:

$$
z=0
$$

For:

$$
z=mx+c
$$

the boundary is:

$$
mx+c=0
$$

and therefore:

$$
x=-\frac{c}{m}
$$


## 9. Slope and Intercept in Logistic Regression

The handwritten notes continue to use the linear expression:

$$
z=mx+c
$$

where:

- `m` → coefficient / slope
- `c` → intercept

The difference is that logistic regression does **not** use `z` directly as the final prediction.

Instead:

```text
z = mx + c
      ↓
Sigmoid
      ↓
Probability
```

So the coefficients control how the input changes the estimated probability.


# 10. From Probability to Class

The complete decision process can be represented as:

```text
                  Input x
                     │
                     ↓
             Linear combination
                  z = mx + c
                     │
                     ↓
              Sigmoid function
              1 / (1 + e^-z)
                     │
                     ↓
              Probability p
                     │
                     ↓
              Threshold check
              ┌──────┴──────┐
              ↓             ↓
           p < 0.5       p ≥ 0.5
              ↓             ↓
           class 0       class 1
```

This is the central workflow of binary logistic regression.

# 15. Key Mathematical Summary

### Linear score

$$
z=b_0+b_1x_1+b_2x_2+\cdots+b_nx_n
$$

### Sigmoid

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

### Probability

$$
P(y=1\mid X)=\sigma(z)
$$

### Binary decision

$$
\hat y=
\begin{cases}
1,&P(y=1\mid X)\geq 0.5\\
0,&P(y=1\mid X)<0.5
\end{cases}
$$

### Decision boundary

For the standard `0.5` threshold:

$$
z=0
$$


````python
import numpy as np
from sklearn.linear_model import LogisticRegression

# Small example based on the handwritten study-hours / pass-fail idea.
X = np.array([[0], [1], [2], [4], [5]])
y = np.array([0, 0, 0, 1, 1])

model = LogisticRegression()
model.fit(X, y)

study_hours = np.array([[3], [5]])
probabilities = model.predict_proba(study_hours)
predictions = model.predict(study_hours)

print("Probabilities:")
print(probabilities)

print("\nPredicted classes:")
print(predictions)
````

````python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

z_values = np.array([-5, -2, 0, 2, 5])
print(sigmoid(z_values))
````

# 16. Important Terms

| Term | Meaning |
|---|---|
| **Logistic regression** | A supervised classification algorithm based on a linear score followed by a sigmoid transformation |
| **Binary classification** | Classification with two possible classes |
| **Sigmoid** | S-shaped function that maps a real-valued score to `(0,1)` |
| **Probability** | Estimated likelihood of belonging to a class |
| **Threshold** | Boundary used to convert probability into a class |
| **Decision boundary** | Boundary separating predicted classes |
| **Coefficient / slope** | Controls how an input feature affects the linear score |
| **Intercept** | Constant term in the linear score |
| **Multinomial logistic regression** | Logistic-regression extension for multiclass classification |
