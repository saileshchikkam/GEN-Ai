# Tuples

### --> () -- Tuple brackets

```python
a = tuple()
print(a)
print(type(a))
```

```python
a = (1, 2, 3, 4, 5)
print(a)
print(type(a))
```

### Tuples are like list only which can store data in ordered way but not mutable

```python
a[2] = 10 # tuples are immutable
print(a)
```

## Packing and Unpacking of tuples

```python
# Packing of tuples
b = (1, 2, 3, 4, 5, 45, 67, 89, 34, 23)
print(b)
```

```python
# Unpacking of tuples
a,b,c,d,e = (1, 2, 3, 4, 5)
print(a,b,c,d,e)
```

```python
# Unpacking with a*
first_value, *middle_values, last_value = (1, 2, 3, 4, 5)
print(first_value)
print(middle_values)
print(last_value)
```

## Nested Tuple

```python
nested_tuple = ((1, 2, 3), (4, 5, 6), (7, 8, 9))
print(nested_tuple[0][:])
```

### Iterating over a tuple inside a tuple

```python
for i in nested_tuple:
    for j in i:
        print(j, sep=" ")
    print(" ")
```
