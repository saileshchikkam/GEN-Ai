## 8. JSON Library

### --> when we are going to work with NoSQL database data then the inside the NoSQL database is going to there in the JSON format and along with that when we are going to work with the datasets as well then again the datasets will be consisting of data in the json format

### ---> ***When we get the JSON Data, we get in a JSON string format so by using library we can convert the string data into the dictionary format for serializing our data

```python
import json
```

#### --> json.dumps()  === dict data to str data

```python
data = {"Name" : "sailesh",
        "Age" : 21}
str_json = json.dumps(data)
print(str_json)
type(str_json)
```

#### ---> json.loads() === str dta to dict data

```python
json_dict = json.loads(str_json)
print(type(json_dict))
```
