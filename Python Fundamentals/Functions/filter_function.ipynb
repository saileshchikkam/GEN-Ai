# Filter Function

## Filter function is going to retrieve only those items which are meeting a certain condition in an iterable

#

#### Syntax for filter function

list(filter(func, iterable))

```python
## To retrieve only those vaalues which are even
def even_function(value):
    if value%2 == 0:
        return True
```

```python
iterabel_list = [1,2,3,4,5,6,7,8,9]
print(list(filter(even_function, iterabel_list)))
```

```python
print(list(filter(lambda x:x%2==0,iterabel_list)))
```

```python
## want to retrieve the values which are greater than 5
print(list(filter(lambda x : x>5 , iterabel_list)))
```

```python
## want to retrieve all those values which are greater than 5 as well and even as well
print(list(filter(lambda x : x>5 and x%2==0 , iterabel_list)))
```

```python
employees = [{
    'Name' : 'Sailesh',
    'Age' : 21,
    'Qualification' : 'btech'
},
{
    'Name' : 'kiran',
    'Age' : 22,
    'Qualification' : 'b.com'
},
{
    'Name' : 'Roshan',
    'Age' : 27,
    'Qualification' : 'Mtech'
}]
```

```python
def age(employees):
    if employees['Age']>25:
        return True
```

```python
print(list(filter(age, employees)))
```
