# Lambda Functions

## Lambda functions are the one liner functions defined by the lambda keyword


## --> The syntax for the lambda function is

## syntax ====>>> lambda arguments : expression

## lambda --> keyword whiich simpily let the python interpreter know that the current function is our lambda function

## Expression ----> Logic(Function Body)

## lambda x:x**2

### -->> They can have any no:of arguments but only one expression or one logic 

### -->> They are used for the short operations or as arguments to the higher order functions

### -->> It can be treated as a function without the name and they don't even require a return statement

```python
## Function which calculates the square of the numbers
def square(num):
    return num**2

print(square(5))
```

```python
lambda_func=lambda x:x**2
type(lambda_func)
```

```python
lambda_func(5)
```

```python
## lambda function to calculate the sum of two numbers
sum = lambda x,y:x+y
sum(1,2)
```

```python
pro = lambda x,y,z,f,g : x*y*z+f-g
pro(1,1,1,3,4)
```
