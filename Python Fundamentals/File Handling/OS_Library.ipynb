# OS Library

## OS Libraary is able to interact with our Operating System and can lead let us work with Multiple Direcories and Files

## 1. getcwd() --> It is capable of providing us with the idea about the current working directory

## getcwd --> get current working directory

```python
import os
```

```python
print(os.getcwd())
```

## 2. mkdir() --> make directory function -->os.mkdir()

```python
## mkdir() --> make directory ---> It will be helping us out with creating the directory


new_path = "New_Folder"
os.mkdir(new_path)
print("The new directory ",new_path," is created successfully")
```

## 3. listdir() --> Listing all of the directories present


```python
os.listdir('.')
```

## Joining 2 paths Together

## 4. os.path.join() ---> It will be joining multiple paths together

```python
dir_name = "Folder_data"
file_name = "File_data"
fullPath = os.path.join(dir_name,file_name)
print(fullPath)
```

## ---> if you want to join multiple path with the current path


## 5. os.path.join(os.getcwd(), , )

```python
currentPath = os.path.join(os.getcwd(),dir_name,file_name)
print(currentPath)
```

## 6.os.path.exists()

### --> figures out whether the given path exists or not

```python
given_path = 'example/example'
print(os.path.exists(given_path)) # doesn't exists
```

## 7. os.path.isfile() -->whether if current path is a file or not
## 8. os.path.isdir() --> checks current path is directory or not

```python
path = 'Total_data.txt'
if os.path.isfile(path):
    print("Current path belongs to a file")

elif os.path.isdir(path):
    print("Current path belongs to a directory")
else:
    print("Current path is neigther a file nor a folder")
```

## The complete path <-----> Absolute path

## 9. os.path.abspath()

### It provides us with the complete path of with the absolute path of the file or directory

```python
file_name = "Total_data.txt" # Relative path
complete_path = os.path.abspath(file_name)
print(complete_path)
```
